## Oefening 1: Vind de uitweg uit het doolhof

In deze oefening moet je een recursief algoritme geschrijven om een weg uit een vierkant doolhof te vinden. 
Vanuit een beginpositie (een tuple met een x- en y-coördinaat ) moet je een weg naar 1 van de zijkanten vinden die niet geblokeerd is (door 'X'). 

Daarvoor controleer je, in elk vakje dat je bezoekt, de 4 mogelijke richtingen (onder, boven, links, rechts zoals opgegeven) of je nog verder kan lopen. Elke stap die je zet voeg je toe aan `ontsnapping` door het huidige coördinaat  (ook een tuple van een x- en y-plaats) aan te hechten.  
Als je een mogelijk einde hebt bereikt (basisoplossing) bouw je de recursie af - `ontsnapping` zal de juiste oplossing hebben als globale variabele. Indien je ergens niet verder kan moet de recursie backtracken en wordt `ontsnapping` dan ook terug aangepast (de laatste stap uitgewist).

```python
# kan je gebruiken om een stap te printen als een woord
def decodeer_stap(x, y) -> str:
    if x == 1:
        return 'Onder'
    if x == -1:
        return 'Boven'
    if y == 1:
        return 'Rechts'
    if y == -1:
        return 'Links'

# controleer of de nieuwe plaats geldig is (return = True/False)
# -> binnen het doolhof
# -> de plaats is leeg --> 1 enkele spatie ' '
# -> nog geen deel van uitweg
def check_plaats(x, y, uitweg) -> bool:
    if 0 <= x < len(doolhof) and 0 <= y < len(doolhof) and doolhof[x][y] == ' '\
            and (x, y) not in uitweg:
        return True
    else:
        return False

# Vind de uitweg door het doolhof
# Je mag er vanuit uitgaan dat er altijd een uitweg is
# Je mag enkel door 'lege' plaatsen gaan, i.e., waar er een spatie staat ' '
# Belangrijk!!! De output wordt bijgehouden in de variabele ontsnapping
# let dus heel goed op deze variabele (hint: dus ook bij backtracking)
# ontsnapping is een lijst van paren met coordinaten
def vind_uitweg(x, y, diepte) -> bool:    
    # voor de visualisatie
    indent = '\t' * diepte
    # niet aanpassen, ter controle van de recursie
    dieptes.append(diepte)
    print(indent + 'Diepte:', diepte)
    print(indent + 'Positie:', x, y)
    global ontsnapping
    
    # breid de ontsnapping uit
    ontsnapping += [(x, y)]
    print(indent + 'Voorlopige uitweg:', ontsnapping)

    # basisoplossing: je bent aan de rand van het doolhof
    # en de plaats is leeg
    # merk op: je kan hier printen 'Einde bereikt' maar je moet enkel
    # de juiste return verzorgen om de oefening juist op te lossen
    if doolhof[x][y] == ' ' \
        and (x in (0, len(doolhof)-1) 
             or y in (0, len(doolhof)-1)):
        print(indent + 'Einde bereikt! Pad:', ontsnapping)
        return True
    # je bent nog niet op het einde, zet een stap en controleer die
    else:
        # de vier mogelijk stappen
        for x_stap, y_stap in [(1, 0), (-1, 0), (0, 1), (0, -1)]:
            print(indent + 'Naar:', decodeer_stap(x_stap, y_stap))
            # controleer of dit een geldige stap is
            # ga na of de uitweg gevonden is
            # indien niet, backtrack
            if check_plaats(x + x_stap, y + y_stap, ontsnapping):
                if vind_uitweg(x + x_stap, y + y_stap, diepte + 1):
                    return True
                else:
                    print(indent + 'Backtrack')
                    ontsnapping = ontsnapping[:-1]
        # vul de juiste return hier ook in
    return False
```

```python
ontsnapping = []
dieptes = []
begin_positie = (2, 1)
doolhof = [
    [' ','X','X','X','X'],
    ['X',' ','X',' ','X'],
    ['X',' ',' ',' ','X'],
    ['X','X','X',' ',' '],
    ['X','X','X','X','X']
]
if vind_uitweg(begin_positie[0], begin_positie[1], 0):
    print('Finale uitweg:', ontsnapping)
    print('Dieptes:', dieptes)
assert ontsnapping == [(2, 1), (2, 2), (2, 3), (3, 3), (3, 4)]
assert dieptes == [0, 1, 1, 2, 3, 4]
```

# Vraag 1

Maak een recursief algoritme dat een zo lang mogelijk pad zoekt door een doolhof vanuit beginpositie `begin_positie`. 

De uitkomst van de recursie is namelijk een pad in de vorm van een lijst van coördinaten. Je zal dus van vakje naar vakje moeten springen, de 4 stappen zetten, controleren of ze mogelijk zijn en eventueel naar het volgende vakje gaan met de recursieve functie. Voor elk vakje houd je het beste pad bij dat tot dan toe gevonden is.

De coördinaten waarlangs je passeert worden uitgedrukt als tupels van 2 elementen (x en y). Er zijn 2 soorten elementen in het doolhof (dat niet per se vierkant is): 'X' en 'O' (let op, 'O' en niet '0'). 
Het pad mag maar maximaal `max_O` 'O's bevatten. 
Het doolhof is een NumPy array en zit in variabele `doolhof_np`. Dit is de enigste globale variabele die je mag gebruiken en deze moet NIET aangepast worden.

```python
# kan je gebruiken om een stap te printen als een woord
def decodeer_stap(x, y):
    if x == 1:
        return 'Onder'
    if x == -1:
        return 'Boven'
    if y == 1:
        return 'Rechts'
    if y == -1:
        return 'Links'

# vul deze functie aan zodat deze controleert of de nieuwe plaats:
# - binnen het doolhof past
# - nog niet in de uitweg zit
# - er minder dan `max_O` 'O's in de uitweg zitten
def check_plaats(x, y, uitweg):
    convert = []
    for coor in uitweg:
        convert.append(doolhof_np[coor[0], coor[1]])
    if 0 <= x < doolhof_np.shape[0] and 0 <= y < doolhof_np.shape[1] and convert.count('O') < max_O\
            and (x, y) not in uitweg:
        return True
    else:
        return False

# vul de recursieve functie verder aan
def vind_uitweg(x, y, pad, diepte):    
    # voor de visualisatie
    indent = '---' * diepte
    # niet aanpassen, ter controle van de recursie
    dieptes.append(diepte)
    print(indent + 'Diepte:', diepte)
    print(indent + 'Positie:', x, y)
    
    # breid de ontsnapping uit en houd het beste pad bij
    pad += [(x, y)]
    print(indent + 'Voorlopige uitweg:', pad)
    huidig_beste_pad = pad
    
    # de vier mogelijk stappen
    for x_stap, y_stap in [(1, 0), (-1, 0), (0, 1), (0, -1)]:
        print(indent + 'Naar:', decodeer_stap(x_stap, y_stap))
        
        # pas de coordinaten aan
        x_nieuw, y_nieuw = x + x_stap, y + y_stap

        # controleer de coordinaten en doe een recursieve oproep
        # om de volgende stap te nemen - pas ook het beste pad aan
        # indien nodig (vergeet niet het pad hard te kopieren met list())
        if check_plaats(x_nieuw, y_nieuw, pad):
            huidig_pad = list(pad)
            nieuw_pad = vind_uitweg(x_nieuw, y_nieuw, huidig_pad, diepte + 1)

            if len(nieuw_pad) > len(huidig_beste_pad):
                huidig_beste_pad = nieuw_pad
            else:
                print(indent + 'Backtrack')
                
    # vul de juiste return hier ook in
    return huidig_beste_pad
```

```python
import numpy as np

dieptes = []
begin_positie = (0, 0)
max_O = 2
doolhof = [
    ['O','O','X'],
    ['X','O','X'],
    ['X','X','X']
]
doolhof_np = np.array(doolhof)
uitweg = vind_uitweg(begin_positie[0], begin_positie[1], [], 0)

print('Finale uitweg:', uitweg)
print('Dieptes:', dieptes)
assert uitweg == [(0, 0), (1, 0), (2, 0), (2, 1), (2, 2), (1, 2), (0, 2), (0, 1)]
assert dieptes == [0, 1, 2, 3, 4, 4, 5, 6, 7, 6, 2, 1]
```
