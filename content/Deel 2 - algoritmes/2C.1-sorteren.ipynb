## Bubble sort

Het eerste algoritme dat we bekijken sorteert door getallen te laten 'bovendrijven' ('bubble to the top' in het Engels) door telkens paarswijze sortering toe te passen. In pseudocode:
- Neem een lijst van sorteerbare elementen
- Zolang er een wissel is doorgevoerd in de vorige iteratie:
    - Loop door de lijst van links naar rechts vanaf het tweede element en voor elk element:
        - Bekijk of het element links 'groter' dan het huidige element is
        - Indien zo, verwissel
- Geef de gesorteerde lijst terug.

Het voordeel is dat dit een makkelijk te implementeren algoritme is. Anderszijds zullen we later zien dat het op zich niet al te efficiënt, al was het maar omdat er op het laatste altijd nog een heel keer door de lijst moet gelopen worden om er zeker van te zijn dat er geen wissels meer nodig zijn.

Hieronder is een Python implementatie te vinden. We beginnen telkens met een variabele `verwisseld` op 'True' dit, zelfs als de lijst gesorteerd is, ervoor zorgt dat de `while` gestart wordt en de lijst al minstens 1 keer doorlopen wordt. Daarna beginnen we door de lijst te lopen op index 1, dus het tweede element. Het controleren en verwisselen op zich is niet moeilijk.

```python
def bubble_sort(reeks_in):
    verwisseld = True # begin vanuit de assumptie dat we moeten verwisselen
    aantal_rondes = 0

    while verwisseld:
        aantal_rondes += 1
        print(f'\nReeks aan begin van ronde {aantal_rondes}:', reeks_in)
        verwisseld = False # als de ronde begonnen is zetten we verwisseld op False tot het tegendeel bewezen is

        for positie in range(1, len(reeks_in)): # controleer vanaf het 2de element (op positie 1 dus)
            if reeks_in[positie] < reeks_in[positie-1]: # indien gewisseld moet worden
                print(f'Verwissel {reeks_in[positie-1]} met {reeks_in[positie]} op posities {positie-1} en {positie}')
                getal = reeks_in[positie]
                reeks_in[positie] = reeks_in[positie-1]
                reeks_in[positie-1] = getal
                verwisseld = True # er is verwisseld, hierna volgt dus nog een ronde
        print(f'Reeks na ronde {aantal_rondes}:', reeks_in)
    print('Finale output:', reeks_in)
```

Hieronder de output voor 2 lijsten:

```python
reeks_1 = [5, 1, 4, 2, 8]
bubble_sort(reeks_1)
```

```python
reeks_2 = ['johannes', '@aaa', 'aaaa', 'bbb']
bubble_sort(reeks_2)
```

```python
lijst = [7, 2, 5, 8, 10, 4, 6, 1, 3, 9]
bubble_sort(lijst)
```

## Merge sort

Een tweede algoritme dat veel populairder is om te sorteren (zoals hierboven aangegeven) is merge sort. Het is gebaseerd op recursie in een opzet die niet geheel verschilt van het eerdere zoeken naar een maximum. Het algoritme bestaat uit 2 delen: 
- Het recursief de lijst opbreken in 2 delen (wederom tot we een lijst van 1 element bekomen) 
- Het samenvoegen van de uitkomst van de 2 delen - ze zijn apart wel gesorteerd maar moeten nog met elkaar gemengd worden.
Het laatste stuk geeft het algoritme dan ook de naam.

Hieronder zullen we eerst bekijken hoe 2 lijsten kunnen samengevoegd worden in de `merge()` functie. De functie neemt als input 2 lijsten en overloopt de elementen met 2 tellers, 1 teller per lijst. De teller wordt gebruikt als huidige index per lijst en wordt verhoogt wanneer de lijst het laagste getal van de 2 heeft. Als bv. lijst [1,3,5] en [2,3,4] wordt overlopen, dan zullen de tellers eerst op 0 staan voor beide lijsten. Dan zal 1 uit de eerst lijst aan de finale `lijst_merge` worden toegevoegd, waarna `teller1` wordt verhoogt om aan te duiden dat we de volgende keer naar de tweede index zullen kijken. `teller2` staat nog op 0 en vermits 2 in de tweede lijst kleiner is dan 3 in de eerste lijst zal 2 aan `lijst_merge` worden toegevoegd en wordt `teller2` verhoogt, enz.
Dit doen we tot 1 van de 2 lijsten helemaal doorlopen is. Daarna kunnen we de resterende elementen van de andere lijst gewoon achteraan toevoegen aan `lijst_merge`, want de lijst is toch gesorteerd.

```python
def merge(lijst1, lijst2):
    print('Merge', lijst1, 'en', lijst2)
    teller1 = 0
    teller2 = 0
    n1 = len(lijst1)
    n2 = len(lijst2)
    lijst_merge = []

    while (teller1 < n1) and (teller2 < n2):
        if lijst1[teller1] < lijst2[teller2]:
            lijst_merge.append(lijst1[teller1])
            teller1 += 1
        else:
            lijst_merge.append(lijst2[teller2])
            teller2 += 1
    while teller1 < n1:
        lijst_merge.append(lijst1[teller1])
        teller1 += 1
    # alternatieve, kortere versie: 
#     lijst_merge.extend(lijst1[teller1:])
    while teller2 < n2:
        lijst_merge.append(lijst2[teller2])
        teller2 += 1
    return lijst_merge
```

Nu kunnen we onze `merge()` functie gebruiken binnen een recursie. In `merge_sort()` werken we gewoon naar het geval dat de lijst lengte 1 heeft en anders splitsen we de lijst op. Merk op de return wel bestaat uit de `merge()` van het linker- en rechterstuk:

```python
def merge_sort(lijst):
    print('Sorteren van lijst:', lijst)
    n = len(lijst)
    if n > 1:
        midden = n // 2
        lijstLinks = merge_sort(lijst[:midden])
        lijstRechts = merge_sort(lijst[midden:])
        return merge(lijstLinks, lijstRechts)
    else: # lijsten met 1 element zijn al gesorteerd
        return lijst
```

```python
lijst = [7, 2, 5, 8, 10, 4, 6, 1, 3, 9]
print("Oorspronkelijke lijst:", lijst)
lijst_gesorteerd = merge_sort(lijst)
print("Geordende lijst: ", lijst_gesorteerd)
```
