# 2B.2 - Voorbeeld dammen

Dammen kan ook gebruikt worden om recursie uit illustreren. We bouwen de recursie op. Eerst leren we de stappen zetten, daarna voegen we een controle-stap toe en ten slotte vullen we een globale variabele in die aanduidt welke stappen we hebben gezet en in welke volgorde.

## Hulp-functies

```python
def print_bord(bord):
    for rij in bord:
        output = '|'
        for plaats in rij:
            if plaats == '':
                output += '_'
            else:
                output += '%1s' % plaats
        print(output + '|')
        
# controleer dat move niet buiten spelbord valt
def check_move(pos_x, pos_y):
    if 0 <= pos_x < len(bord_begin) and 0 <= pos_y < len(bord_begin):
        return True
    else:
        return False
    
def kopieer_bord(bord):
    propere_kopij = []
    for rij in bord:
        nieuwe_rij = ['_' for i in range(0, len(bord))]
        for i in range(0, len(rij)):
            nieuwe_rij[i] = rij[i]
        propere_kopij.append(nieuwe_rij)
    return propere_kopij
```

## Deel 1: zorgeloze recusie

We zetten alle mogelijke stappen in het vakje waar we vertrekken. In elk vakje waar we komen houden we de recursie-diepte bij zodat we weten hoeveel stappen we nodig hebben om hier te geraken. In deze oplossing kunnen stappen overschreven worden in de globale variabele `bord_begin`.

```python
def vind_moves(positie_x, positie_y, bord, stap):
    bord[positie_x][positie_y] = stap
    print('Positie:', positie_x, positie_y)
    print('Stap:', stap)
    print_bord(bord)

    for (x, y) in stappen:  # haakjes moeten niet
        nieuwe_x = positie_x + x
        nieuwe_y = positie_y + y
        
        # check op 'X'...
        if check_move(nieuwe_x, nieuwe_y) and bord[nieuwe_x][nieuwe_y] == '':
            vind_moves(nieuwe_x, nieuwe_y, bord, stap + 1)

# linksboven, linksonder, rechtsboven, rechtsonder
stappen = [(-2, -2), (2, -2), (-2, 2), (2, 2)]

bord_begin = [
    ['','', '','', '','',''],
    ['','X','','','','X',''],
    ['','','','', '','',''],
    ['','X','','X','','X',''],
    ['','', '','', '','',''],
    ['','', '','', '','X',''],
    ['','', '','', '','','']
]
print('Beginbord:')
print_bord(bord_begin)

vind_moves(4, 0, bord_begin, 0)
print('Eindoplossing:')
print_bord(bord_begin)
```

## Deel 2: Toevoegen van stap-controle

We controleren of de stap mogen zetten zoals dit hoort in dammen: er moet een pion ('X') staan tussen het begin en het einde van de sprong.

```python
def vind_moves(positie_x, positie_y, bord, stap):
    bord[positie_x][positie_y] = stap
    print('Positie:', positie_x, positie_y)
    print('Stap:', stap)
    print_bord(bord)

    for (x, y) in stappen:  # haakjes moeten niet
        # mogelijkheid 1
        check_x = check_y = -1
        if x > 0: # als x=2, dan moeten we de coordinaat met 1 verhogen
            check_x = 1 
        if y > 0: # analoog aan x
            check_y = 1
        # mogelijkheid 2
#         check_x, check_y = int(x/2), int(y/2)
        pion_positie_x = positie_x + check_x
        pion_positie_y = positie_y + check_y
        # bv. van (4,0) met stap (-2,2) naar (2,2)
        # -> wat ligt er tussen (4,0) en (2,2)?
        nieuwe_x, nieuwe_y = positie_x + x, positie_y + y

        # tov deel 1 is er nu 1 extra conditie
        if check_move(nieuwe_x, nieuwe_y) \
                and bord[nieuwe_x][nieuwe_y] == '' \
                and bord[pion_positie_x][pion_positie_y] == 'X':
            vind_moves(nieuwe_x, nieuwe_y, bord, stap + 1)


# linksboven, linksonder, rechtsboven, rechtsonder
stappen = [(-2, -2), (2, -2), (-2, 2), (2, 2)]

bord_begin = [
    ['','', '','', '','',''],
    ['','X','','','','X',''],
    ['','','','', '','',''],
    ['','X','','X','','X',''],
    ['','', '','', '','',''],
    ['','', '','', '','X',''],
    ['','', '','', '','','']
]
print('Beginbord:')
print_bord(bord_begin)

vind_moves(4, 0, bord_begin, 0)
print('Eindoplossing:')
print_bord(bord_begin)
```

## Deel 3: zetten van stapnummers

Nu zoeken we naar de langste sequentie van stappen waarin we pionnen kunnen innemen.

Sommige stappen worden uitgewist als we een nog langere oplossing hebben gevonden. Dit is echte backtracking (t.o.v. deel 1 en 2).

```python
def vind_moves(positie_x, positie_y, bord, stap):
    bord[positie_x][positie_y] = stap
    print('Positie:', positie_x, positie_y)
    print('Stap:', stap)
    print_bord(bord)

    # we houden hier de beste stap bij
    beste_stap, beste_bord = stap, bord
    for (x, y) in stappen:  # haakjes moeten niet
        check_x = check_y = -1
        if x > 0:
            check_x = 1
        if y > 0:
            check_y = 1
        pion_positie_x, pion_positie_y = positie_x + check_x, positie_y + check_y
        nieuwe_x, nieuwe_y = positie_x + x, positie_y + y

        if check_move(nieuwe_x, nieuwe_y) \
                and bord[nieuwe_x][nieuwe_y] == '' \
                and bord[pion_positie_x][pion_positie_y] == 'X':
            nieuw_bord = kopieer_bord(bord)

            # we krijgen nu een tupel terug van de recursieve oproep
            stap_oplossing, bord_oplossing = vind_moves(nieuwe_x, nieuwe_y, nieuw_bord, stap + 1)
            
            # we controleren of we een betere oplossing hebben gevonden
            if stap_oplossing > beste_stap:
                beste_stap, beste_bord = stap_oplossing, bord_oplossing

    # in elk vakje geven we de beste stap terug van alle vakjes die hier bereikbaar waren
    return beste_stap, beste_bord


# linksboven, linksonder, rechtsboven, rechtsonder
stappen = [(-2, -2), (2, -2), (-2, 2), (2, 2)]

bord_begin = [
    ['','', '','', '','',''],
    ['','X','','','','X',''],
    ['','','','', '','',''],
    ['','X','','X','','X',''],
    ['','', '','', '','',''],
    ['','', '','', '','X',''],
    ['','', '','', '','','']
]
print('Beginbord:')
print_bord(bord_begin)

stap, beste_bord = vind_moves(4, 0, bord_begin, 0)
print('Eindoplossing:')
print_bord(beste_bord)
```
