```python
def print_bord(bord):
    for rij in bord:
        output = '|'
        for plaats in rij:
            if plaats == '':
                output += '_'
            else:
                output += '%1s' % plaats
        print(output + '|')
        
# controleer dat move niet buiten spelbord valt
def check_move(pos_x, pos_y):
    if 0 <= pos_x < len(bord_begin) and 0 <= pos_y < len(bord_begin):
        return True
    else:
        return False
    
def kopieer_bord(bord):
    propere_kopij = []
    for rij in bord:
        nieuwe_rij = ['_' for i in range(0, len(bord))]
        for i in range(0, len(rij)):
            nieuwe_rij[i] = rij[i]
        propere_kopij.append(nieuwe_rij)
    return propere_kopij
```

```python
def vind_moves(positie_x, positie_y, bord, stap):
    bord[positie_x][positie_y] = stap
    print('Positie:', positie_x, positie_y)
    print('Stap:', stap)
    print_bord(bord)

    beste_stap, beste_bord = stap, bord
    for (x, y) in stappen:  # haakjes moeten niet
        check_x = check_y = -1
        if x > 0:
            check_x = 1
        if y > 0:
            check_y = 1
        pion_positie_x, pion_positie_y = positie_x + check_x, positie_y + check_y
        nieuwe_x, nieuwe_y = positie_x + x, positie_y + y

        if check_move(nieuwe_x, nieuwe_y) \
                and bord[nieuwe_x][nieuwe_y] == '' \
                and bord[pion_positie_x][pion_positie_y] == 'X':
            nieuw_bord = kopieer_bord(bord)
            
            stap_oplossing, bord_oplossing = vind_moves(nieuwe_x, nieuwe_y, nieuw_bord, stap + 1)
            if stap_oplossing > beste_stap:
                beste_stap, beste_bord = stap_oplossing, bord_oplossing

    return beste_stap, beste_bord


# linksboven, linksonder, rechtsboven, rechtsonder
stappen = [(-2, -2), (2, -2), (-2, 2), (2, 2)]

bord_begin = [
    ['','', '','', '','',''],
    ['','X','','','','X',''],
    ['','','','', '','',''],
    ['','X','','X','','X',''],
    ['','', '','', '','',''],
    ['','', '','', '','X',''],
    ['','', '','', '','','']
]
print('Beginbord:')
print_bord(bord_begin)

stap, beste_bord = vind_moves(4, 0, bord_begin, 0)
print('Eindoplossing:')
print_bord(beste_bord)
```
