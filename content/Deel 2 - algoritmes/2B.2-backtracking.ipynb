# 2B.2 Recursieve backtracking

Recursie is een krachtig instrument om snel complete enumeraties op te zetten, maar zoals eerder gezien kan dit tot lange berekeningen leiden die soms overbodig zijn. In dit stuk gaan we recursieve backtracking behandelen waarbij er een zoekboom wordt opgezet waarvan er stukken zullen uitgesnoeid worden wanneer er toch geen interessante paden meer te vinden zullen zijn (bijvoorbeeld als we weten dat het optimale getal er niet meer kan inzitten of het doorlopen pad te lang is geworden, we buiten een schaakbord gaan, etc.). 
Backtracking slaat erop dat we de zoekboom doorlopen en soms teruggaan naar een vorig punt en van daaruit verderzoeken. Denk bijvoorbeeld aan het doorlopen van een doolhof: we blijven alle richtingen proberen tot we tegen een muur lopen en gaan dan op onze stappen terug tot we terug ergens een andere richting uit kunnen. Deze analogie komt goed van pas in onderstaande voorbeelden.

## a) Schaken

Schaakproblemen vormen ideale voorbeelden voor recursieve backtracking. In dit voorbeeld wordt er gekeken of alle vakjes op een schaakbord kunnen bezocht worden met een paard zonder ergens 2x te komen (dit wordt ook de Knight's tour genoemd, zie: https://en.wikipedia.org/wiki/Knight%27s_tour). Een paard heeft 8 verschillende zetten (zie: https://www.chess.com/terms/chess-knight) en het is de bedoeling om stapsgewijs na te gaan of we of het einde hebben bereikt (de stopconditie) of geen verdere stappen meer kunnen nemen en stappen terug moeten zetten in de zoekboom opgebouwd door de recursie omdat we geen geldige stappen meer kunnen zetten die in het spelbord zijn of niet naar plaatsen leiden die al gezocht zijn in de huidige oplossing.

We hebben dan ook verschillende elementen nodig in onze recursie. Ten eerste maken we een functie die controleert of de stap die we gaan nemen binnen het bord valt. Het bord is een globale variabele `bord` en bestaat uit een lijst van lijsten om een vierkante matrix te bekomen met het `n\times n` spelbord. We zoeken of de x- en y-coördinaten dus groter of gelijk zijn aan 0, kleiner dan `n` en of het bord er nog niet bezocht is (we beginnen met een bord vol met -1 als placeholder voor de toekomstige volgnummers voor de stappen):

```python
def stap_is_correct(x, y, bord):
    if x >= 0 and y >= 0 and x < n and y < n and bord[x][y] == -1:
        return True
    return False
```

Nu gaan we de eigenlijke recursieve functie bekijken. 

Het eindcriterium, zoals gezegd, is het bereiken van het finale volgnummer. Als we een bepaalde plaats bezoeken zetten we er de huidige stapnummer. Dit betekent dat als we uiteindelijk de `n^2` stap kunnen zetten, we heel het bord hebben kunnen vullen met onze stappen. 

Indien we nog niet op het einde zijn, gaan we alle 8 stappen proberen in de huidige positie. Het eerste wat we doen is berekenen van de nieuwe coördinaten door de sprong bij de huidige x/y-waarden te tellen. Dan gebruiken we de functie van hierboven om te bekijken of de nieuwe coördinaten geldig zijn. We gebruiken variabele `pos` om het huidige volgnummer bij te houden. Indien we de stap kunnen zetten dan zetten we op het bord op die positie het volgnummer. Daarna gaan we verder met de volgende stap door de functie recursief op te roepen met de nieuwe coördinaten, een hoger volgnummer, en het aangepaste bord. 

Als we alle stappen in een positie hebben uitgetest en nooit een oplossing hebben gevonden (want dan zou 1 van de zoekpaden een 'True' hebben teruggeven), dan geven we in die positie een 'False' terug, waardoor dat stuk van de recursie stopt. Als we dan op lijn 15 geen `return` oproepen omdat we geen oplossing gevonden hebben verder in de zoekboom zullen we beginnen backtracken en op onze stappen terugkomen. Dit doen we door het bord daar terug leeg te maken met een `-1` waarde.

```python
def recursieve_oplossing(n, bord, huidige_x, huidige_y, sprong_x, sprong_y, pos):
    prefix = '\t' * pos
    print(prefix + f'Recursieve oplossing voor: {huidige_x}, {huidige_y}')
    # eindoplossing en stopcriterium: alles bezocht
    if (pos == n ** 2):
        return True

    # probeer alle 8 mogelijke sprongen
    for i in range(8):
        nieuwe_x = huidige_x + sprong_x[i]
        nieuwe_y = huidige_y + sprong_y[i]
        
        if stap_is_correct(nieuwe_x, nieuwe_y, bord):
            # zet de stapnummer in het vakje
            bord[nieuwe_x][nieuwe_y] = pos
            if recursieve_oplossing(n, bord, nieuwe_x, nieuwe_y, sprong_x, sprong_y, pos + 1):
                return True
            else:
                # backtracking - stap terug uitwissen
                bord[nieuwe_x][nieuwe_y] = -1
                print(prefix, 'Backtracking')
    return False
```

Ten slotte moeten we ook nog alles opzetten. Hieronder wordt het bord aangemaakt (de globale variabele `bord`), de sprongen vastgelegd, de beginpositie vastgelegd alsook het volgnummer. Indien we geen oplossing vinden (wat kan voor kleine borden met $n<5$) dan wordt dit zo meegegeven, anders wordt de finale oplossing als een ingevuld bord met stapvolgorde afgeprint:

```python
def los_op(n):
    # maak bord aan van grootte n x n met -1
    bord = [[-1 for i in range(n)] for i in range(n)]

    # dit zijn de 8 mogelijke sprongen die een paard kan maken
    sprong_x = [2, 1, -1, -2, -2, -1, 1, 2]
    sprong_y = [1, 2, 2, 1, -1, -2, -2, -1]

    # beginpositie paard
    bord[0][0] = 0

    pos = 1

    # zoek de oplossing
    if not recursieve_oplossing(n, bord, 0, 0, sprong_x, sprong_y, pos):
        print("Oplossing bestaat niet")
    else:
        for i in range(n):
            for j in range(n):
                print(bord[i][j], end=' ')
            print()
```

Deze code zal slechts alles in gang zetten:

```python
n = 4
los_op(n)
```

Merk op dat de backtracking dus bestaat uit het afsnijden van bepaalde paden die zouden leiden via het nemen van stappen buiten het bord, of waar de huidige oplossing al is geweest. Dit maakt dat we niet alle mogelijkheden moeten overlopen. In de eerder vermelde Wikipedia-entry staan ook het aantal te doorlopen mogelijkheden. Desondanks het gebruik van recursie zal het toch oppassen zijn met deze code. Bij een $n$ groter dan 5 zal dit algoritme al heel wat tijd nodig hebben om af te lopen.

## Sudoku oplossen

Een sudoku is een simpele numeriek puzzel waarbij men meestal een 9x9 matrix moet invullen waarbij de cijfers 1-9 maar 1 keer per rij/kolom mogen voorkomen en ook slechts 1 maal in de 3x3 matrices die in de 9x9 vervat zitten. Om een sudoku in te vullen moet je ene bepaalde strategie toepassen gebaseerd op eliminatie door te kijken welke getallen er niet kunnen voorkomen om zo de overblijvende getallen in te vullen. Dit kan perfect in een backtracking algoritme geïmplementeerd worden. We gaan proberen zoveel mogelijk getallen in te vullen, todat heel het bord is ingevuld, of er ergens een fout in onze oplossing is. In dat laatste geval moeten we backtracken en getallen uitwissen zodat we opnieuw verder kunnen zonder fouten.

Ten eerste implementeren we de controle. Hiervoor maken we een functie gebaseerd op de sudoku matrix, de huidige rij, kolom, en het getal dat daar ingevuld zal worden. Hieronder worden eerst de rijen overlopen om te zien of het getal al in de rij voorkomen (`(sudoku[i][kol] == getal)`) en dat we niet in de rij waarin we willen invullen zelf controleren (`i != rij`). Indien we het getal dus tegenkomen zullen we meteen 'False' teruggeven om aan te geven dat dit geen geldige zet is binnen de sudoku.
Hetzelfde gebeurt daarna voor de kolommen.
Dan komt er een functie die zal kijken of het getal al voorkomt in de huidige 3x3 submatrix waarin deze `rij`/`kolom` inzitten door de indices de juiste 'start' mee te geven zodat ze juist worden vermeerderd. Zo zal bijvoorbeeld voor de matrix startende op de derde rij en de zesde kolom met indices [3,4,5] horizontaal en [6,7,8] verticaal `boxrij` 3 zijn en `boxkol` 6 (indien onzeker, print deze variabelen eens af) zodat we telkens toch van 0-3 kunnen loopen in de dubbele for-lus doorheen de 3x3 submatrix.

```python
def controleer(sudoku, rij, kol, getal):
    for i in range(9):  # check langs de rijen
        if (sudoku[i][kol] == getal) and (i != rij):
            return False
    for j in range(9):  # check langs de kolommen
        if (sudoku[rij][j] == getal) and (j != kol):
            return False
    boxrij = (rij // 3) * 3
    boxkol = (kol // 3) * 3
    for i in range(3):  # check nu de 3x3 matrices
        for j in range(3):
            if (sudoku[boxrij + i][boxkol + j] == getal) and ((boxrij + i != rij) or (boxkol + j != kol)):
                return False
    return True  # als alle tests goed zijn klopt de move
```

Nu gaan we de spelstrategie zelf implementeren. We doen dit door de sudoku voor een bepaalde positie in te vullen.
- We kijken eerst of we naar de volgende rij moeten gaan (indien `kol`>8, de laatste index van een rij). Dan bekijken we of we het einde bereikt hebben (basisoplossing) en we een cascade van `True` gaan teweegbrengen zodat de recursie zichzelf afbreekt. 
- Indien niet, dan gaan we verder met het huidige vakje. 
    - Eerst kijken we of het vakje leeg is en ingevuld moet worden. Indien al ingevuld kunnen we gewoon het volgende vakje invullen door de recursie voor het volgend vakje (`kol+1`) op te roepen.
    - Anders gaan we een cijfer tussen 1 en 10 proberen, controleren (met `controleer()`), en indien het mag ingevuld worden invullen in de sudoku. Dan kunnen we net zoals hierboven de recusieve voor het volgende vakje oproepen.
        - Als dit ergens tot een juiste oplossing heeft geleid krijgen we vanuit de basisoplossing hier een `True` terug. Wij roepen zelf ook `return True` op waardoor alle uitstaande recursieve oproepen worden afgebroken (test dit eens met een debugger).
        - Als geen enkel van de cijfers in dit vakje tot een oplossing heeft geleid gaan we het vakje terug leegmaken en via de `return False` de vorige functieoproepen laten weten om hier niet verder te zoeken. Zij kunnen dan lokaal kijken of er wel nog andere getallen passen in het vakje.

```python
def los_op(sudoku, rij, kol, diepte):
    
    if kol > 8:  # werk kolom per kolom, als kolom klaar is ga naar volgende rij
        rij += 1
        kol = 0
    if rij > 8:  # alle rijen zijn gedaan - basisoplossing
        return True
    else:
        if sudoku[rij][kol] != 0:  # indien ingevuld ga verder
            return los_op(sudoku, rij, kol + 1, diepte)
        else:  # indien nog niet ingevuld
            for cijfer in range(1, 10):  # probeer 1-9
                if controleer(sudoku, rij, kol, cijfer):  # als dit een geldige zet is
                    sudoku[rij][kol] = cijfer   # vul in
                    print("Diepte: %4d" % diepte, rij, kol, cijfer)
                    resultaat = los_op(sudoku, rij, kol + 1, diepte+1)  # ga verder
                    if resultaat:  # als op het einde, begin recursie af te bouwen
                        return True
            sudoku[rij][kol] = 0 # als er geen oplossing in dit vakje wordt gevonden via recursie, wis uit
            print("Diepte: %4d" % diepte, rij, kol, "backtrack")
            return False
```

Hieronder wordt de code gebruikt met een sudoku als lijst van lijsten:

```python
sudoku = [
[7, 8, 0, 5, 0, 0, 0, 4, 0],
[0, 0, 1, 0, 0, 4, 8, 9, 0],
[0, 0, 9, 0, 3, 0, 0, 0, 1],
[2, 9, 0, 0, 5, 1, 0, 7, 4],
[0, 0, 0, 0, 0, 0, 0, 0, 0],
[8, 4, 0, 9, 7, 0, 0, 3, 2],
[5, 0, 0, 0, 9, 0, 4, 0, 0],
[0, 7, 2, 1, 0, 0, 3, 0, 0],
[0, 1, 0, 0, 0, 7, 0, 2, 6]
]

diepte = 0
if los_op(sudoku, 0, 0, diepte):
    print("Matrix opgevuld")
    for lijn in sudoku:
        print(lijn)
else:
    print("Deze sudoku heeft geen oplossing")
```
