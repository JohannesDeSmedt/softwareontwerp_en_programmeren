## Voorbeeld 1

Tel af tot 0 vanaf een bepaald getal `n` mbv recursie. Je moet geen getal teruggeven, enkel het huidige getal printen (de basisoplossing zal uiteraard wel een `return` nodig hebben).

```python
def tel_af(n):
    print('Het huidige nummer is:', n)
    
    if n == 0:
        return
    else:
        tel_af(n-1)
        
tel_af(10)
```

## Voorbeeld 2

Gebruik recursie om alle getallen van een reeks op te tellen. Doe dit door telkens een rij in 2 te splitsen en de som van de twee rijen weer recursief te berekenen, tot de rij maar 1 element bevat en de som triviaal is. 

```python
def som_reeksen(reeks):
    print('Neem som van reeks:', reeks)
    if len(reeks) == 1:
        return reeks[0]
    else:
        midden = len(reeks) // 2
        tussensom = som_reeksen(reeks[midden:]) + som_reeksen(reeks[:midden])
        print(f'Tussensom {reeks[midden:]} en {reeks[:midden]}:', tussensom)
        return tussensom
    
som_reeksen([10, 2, 33, 3, 22, 4])
```

## Voorbeeld 3

Zoek een opgegeven getal `n` in een opgegegeven lijst - geef `True` terug als het getal in de lijst zit, `False` anders. Gebruik hiervoor een binaire zoekboem met recursie.

```python
def vind_getal(reeks, n):
    print('Reeks:', reeks)
    if len(reeks) == 1:
        print('Basisoplossing', reeks, reeks[0] == n)
        if reeks[0] == n:
            return True
        else:
            return False
    else:
        midden = len(reeks) // 2
        links = vind_getal(reeks[:midden], n)
        rechts =  vind_getal(reeks[midden:], n)
        print(reeks, 'getal in linkerstuk:', links)
        print(reeks, 'getal in rechterstuk:', rechts)
        return links | rechts
```

```python
n = 10
lijst = [1, 0, 9, 10, 11]

vind_getal(lijst, n)
```
