## a) Zoek de laagste getallen in een lijst

Zoek de 2 laagste getallen in een lijst. Hoe gaan we te werk? Pseudocode:
- Maak 2 variabelen aan met een hoog getal als tijdelijke oplossing (of neem de eerste 2 getallen uit de lijst, dit zijn zeker valabele oplossingen)
- Ga door de lijst en doe voor elk element $e$:
    - Als $e$ lager is dan het huidige laagste getal, maak $e$ het laagste getal en maak het huidige laagste getal het tweede laagste getal
    - Als $e$ hoger is dan het huidige laagste getal, maar lager dan het tweede laagste getal, maak $e$ het tweede laagste getal
- Geef de 2 variabelen terug.

Dit hier wordt hieronder geïmplementeerd in Python:

```python
def zoek_kleinste_getallen(lijst: list) -> (int, int):
    kleinste_getal = 1000
    tweede_kleinste_getal = 1000
    teller = 0

    print("%2d" % teller, "%5s" % "--", "%5d" % kleinste_getal, "%5d" % tweede_kleinste_getal)
    for getal in lijst:
        teller += 1 # we zouden ook enumerate kunnen gebruiken
        if getal < kleinste_getal:
            kleinste_getal, tweede_kleinste_getal = getal, kleinste_getal
        elif getal < tweede_kleinste_getal:
            tweede_kleinste_getal = getal
        print("%2d" % teller, "%5d" % getal, "%5d" % kleinste_getal, "%5d" % tweede_kleinste_getal)
    return kleinste_getal, tweede_kleinste_getal

lijst_0 = [7, 5, 3, 1, 9, 4, 10, 2, 6, 9]

kleinste, tweede_kleinste = zoek_kleinste_getallen(lijst_0)
print("Het kleinste getal is", kleinste)
print("Het tweede kleinste getal is", tweede_kleinste)
```

Merk op dat algoritmes ook gevoelig zijn aan uitzonderingen en grensgevallen. Deze maken dikwijls het vinden van een juiste oplossing en het debuggen wat moeilijker. Denk bijvoorbeeld aan de volgende gevallen:
- 1) Wat als er maar 1 element in de lijst zit?
- 2) Wat als er geen elementen in de lijst zitten?
- 3) Wat als er negatieve getallen in de lijst zitten?
- 4) Wat als het eerste en tweede getal hetzelfde zijn?
- 5) Moet de hele lijst doorlopen worden tot het einde?

## b) Zoek het laagste cijfer boven een ondergrens

Een variatie op het klassieke 'zoek het laagste getal in een reeks' is het probleem van het zoeken van het laagste getal gegeven een ondergrens.

De pseudocode ziet er als volgt uit:
- Maak een variable aan met een hoog getal als placeholder (of neem het eerste uit de lijst)
- Ga door de lijst en doe voor elk element $e$:
    - Als $e$ groter dan of gelijk is aan de gegeven ondergrens en lager dan het huidige laagste getal, maak het het nieuwe laagste getal
- Geef de variabele terug.

Dit hier wordt hieronder geïmplementeerd in Python:

```python
lijst = [7, 5, 3, 1, 9, 4, 10, 2, 6, 9]
ondergrens = 3.4
kleinsteGetal = 1000
teller = 0
print("%2d" % teller, "%5s" % "--", "%5d" % kleinsteGetal)

for getal in lijst:
    teller += 1
    if getal >= ondergrens and getal < kleinsteGetal:
        kleinsteGetal = getal
    print("%2d" % teller, "%5d" % getal, "%5d" % kleinsteGetal)
print("Het kleinste getal groter dan de gegeven ondergrens is", kleinsteGetal)
```

Let op, het is makkelijk te zien dat de code hierboven bijvoorbeeld niet zal werken voor een lege lijst (probeer dit eens zelf). Ook hier is het weer belangrijk om in te zien dat er heel wat grensgevallen kunnen opduiken zoals: Wat als er geen getal gevonden wordt?
