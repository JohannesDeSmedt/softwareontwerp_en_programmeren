# Vraag 4

Gegeven zijn 2 dataframes:
- `machines_df` bevat 3 kolommen: de naam van een machine, de looptijd en omsteltijd.
- `taken_df` bevat 2 kolommen: een taaknaam en een tupel met daarin de namen van de machines die nodig zijn om de taak uit te voeren.

Vervolledig de functie `bereken_totale_doorlooptijd()` die als lambda-functie wordt uitgevoerd over de kolom 'Combinatie' in `taken_df` om een nieuwe kolom 'Totale Doorlooptijd' te maken voor `taken_df`. Het argument `taak_combinatie` is de tupel per rij die moet gebruikt worden, in combinatie met `machine_df` waarin je de totale doorlooptijd (omsteltijd + looptijd van elke machine opgeteld) te berekenen per taak.

```python
import pandas as pd
import numpy as np
import random

random.seed(10)
np.random.seed(10)

def bereken_totale_doorlooptijd(taak_combinatie, machines_df):
    print('Taak combinatie:', taak_combinatie)
    print('Machine_df:\n', machines_df)
    totale_doorlooptijd = 0
    
    for machine in taak_combinatie:
        print('Machine:', machine)
#         print('Info machine:', machines_df[machines_df['Machine'] == machine])
        rij_machine = machines_df[machines_df['Machine'] == machine]
        omstel = rij_machine['Omsteltijd']
        looptijd = rij_machine['Looptijd']
        print(rij_machine)
        print(omstel.values[0], omstel.iloc[0])
        print(looptijd.values[0], looptijd.iloc[0])
        totale_doorlooptijd = omstel.values[0] + looptijd.values[0]
    # overloop machine per taak
        # haal de looptijd/omsteltijd van elke taak per machine op in machines_df
        # update totale doorlooptijd
    # ...
   #     print('Totale doorlooptijd', machine, ':', totale_doorlooptijd)
        
    return totale_doorlooptijd
```

```python
machines_data = {'Machine': ['A', 'B', 'C'],
                 'Looptijd': [10, 15, 20],
                 'Omsteltijd': [5, 8, 10]}
machines_df = pd.DataFrame(machines_data)

taken_data = {'Taak': ['Taak1', 'Taak2', 'Taak3'],
              'Combinatie': [('A', 'B'), ('B', 'C'), ('A', 'C')]}
taken_df = pd.DataFrame(taken_data)

print(machines_df)
print(taken_df)

taken_df['Totale Doorlooptijd'] = taken_df['Combinatie'].apply(lambda x: bereken_totale_doorlooptijd(x, machines_df))
taken_df
```

```python
assert np.allclose(taken_df['Totale Doorlooptijd'].values, [38, 53, 45])
```
