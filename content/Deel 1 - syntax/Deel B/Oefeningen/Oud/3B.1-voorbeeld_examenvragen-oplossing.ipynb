# Vraag 1
Opgegeven: een matrix als NumPy `array` object. Vervolledig de functie `vind_matrix_eigenschappen()` die controleert of de matrix vierkant is. Indien zo moet er ook gecontroleerd worden of alle cijfers op de diagonaal van de matrix (van linksboven tot rechtsonder) hetzelfde zijn.

Stop je antwoorden (beide booleans en niet strings zoals 'True' en 'False') in de variabelen `is_vierkant` en `diagonaal_juist` respectievelijk.

```python
def vind_matrix_eigenschappen(np_matrix):
    is_vierkant = False
    diagonaal_juist = False
    
    print('Vorm van de matrix:', np_matrix.shape)
    if np_matrix.shape[0] == np_matrix.shape[1]:
        is_vierkant = True
    
    if is_vierkant:
        diagonaal_juist = True
        te_zoeken = np_matrix[0,0]
        for i in range(len(np_matrix)):
            if not np_matrix[i,i] == te_zoeken:
                diagonaal_juist = False
                break
    
    return is_vierkant, diagonaal_juist
```

```python
import numpy as np

lijsten_in_lijst = [[1]]
matrix = np.array(lijsten_in_lijst)
print(matrix)

is_v, diag = vind_matrix_eigenschappen(matrix)
print('Vierkant?', is_v)
print('Diagonaal juist?', diag)
assert is_v == True
assert diag == True
```

```python
lijsten_in_lijst = [[1, 1, 1], [0, 0, 0], [3, 2, 1]]
matrix = np.array(lijsten_in_lijst)
print(matrix)

is_v, diag = vind_matrix_eigenschappen(matrix)
print('Vierkant?', is_v)
print('Diagonaal juist?', diag)
assert is_v == True
assert diag == False
```

# Vraag 2

Vervolledig de functie `vermenigvuldig()` met als argumenten een vierkante `matrix`, een `factor` en een `ondergrens`. Het is de bedoeling om een matrix te maken die dezelfde vorm heeft als de originele `matrix`, maar met de `factor` op de links-rechts diagonaal en `factor-1` (factor min 1) op al de andere plaatsen.
Maak eerst die `factors` matrix aan, en vermenigvuldig dan `matrix` met die `factors` matrix volgens het kruisproduct. Doe deze vermenigvuldiging tot het maximum van elke rij minstens het minimum `ondergrens` haalt.

```python
import numpy as np

def vermenigvuldig(matrix, factor, ondergrens):
    factors = np.zeros((matrix.shape))
    factors[:] = factor - 1
    for i in range(len(matrix[0])):
        factors[i, i] = factor
    print(factors)
        
    while np.min(np.max(matrix, axis=1)) < ondergrens:
        matrix = np.dot(matrix, factors)

        print(matrix)
        print(np.max(matrix, axis=1))
    return np.round(matrix, 1)
```

```python
np.random.seed(10)
data = np.random.randint(1, 150, (5, 5))

uitkomst = vermenigvuldig(data, 1.2, 200)
uitkomst
```

```python
assert np.allclose(uitkomst, [[208.6, 324.6, 214.6, 263.6, 312.6],
       [317.2, 307.2, 202.2, 267.2, 194.2],
       [290. , 365. , 266. , 350. , 389. ],
       [317.8, 351.8, 370.8, 385.8, 325.8],
       [203.2, 270.2, 224.2, 247.2, 183.2]])
```
