# 3B.1 NumPy oefeningen

Gegeven: de punten van 10 studenten voor 6 vakken:

```python
import numpy as np

np.random.seed(15)

punten = np.random.randint(1, 20, (10, 6))  # random nummers tussen 1 en 20
print(punten)
```

## Vraag 1

Vind de som van elke rij

```python
som_elke_rij = punten.sum(axis=1)

print('Som van elke rij:', som_elke_rij)
```

## Vraag 2

Vind de som van de de originele matrix en de nieuwe matrix met dezelfde vorm die de bonuspunten bevat:

```python
bonuspunten = np.random.randint(1, 3, (10, 6))
print('Bonuspunten:\n', bonuspunten)

som = punten + bonuspunten

print('Som van alle punten:\n', som)
```

Sommige studenten hebben nu teveel punten. Verander de `som`-matrix zodat de punten boven de 20 worden omgezet naar 20:

```python
som_nieuw = np.where(som > 20, 20, som)

som_nieuw
```

## Vraag 3

Voeg de bonuspunten toe als aparte kolommen:

```python
totaal = np.concatenate([punten, bonuspunten], axis=1)

print('Alle punten over 2 jaar:\n', totaal)
print('Vorm van de matrix:', totaal.shape)
```

## Vraag 4

Geef het gemiddelde van elke kolom uit `totaal` (per vak):

```python
gemiddeldes_kolom = totaal.mean(axis=0)

print(gemiddeldes_kolom)
```

Geef het gemiddelde van elke rij in `totaal` (per student):

```python
gemiddelde_student = totaal.mean(axis=1)

print(gemiddelde_student)
```

## Vraag 5

Vermenigvuldig elke even kolom van `totaal` (beginnend bij 0) uit het totaal met 5:

```python
nieuw_totaal = totaal[:,0:totaal.shape[1]:2]

nieuw_totaal = nieuw_totaal * 5

print(nieuw_totaal)
```
