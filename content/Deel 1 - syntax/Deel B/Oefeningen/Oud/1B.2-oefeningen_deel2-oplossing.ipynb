# Pandas oefeningen: deel 2

```python
import numpy as np
import pandas as pd

np.random.seed(10)

# gegeven: de punten van 10 studenten voor 6 vakken
punten = np.random.randint(7, 20, (10, 6))
print(punten)
studenten = ['otis', 'maeve', 'eric', 'adam', 'aimee', 'ruby', 'ola', 'cal', 'vivienne', 'rahim']
```

## Vraag 1

Maak een pandas DataFrame aan met de gegeven studenten.
Tip: geef via het argument 'columns' een lijst van vakken mee (we zetten de index gelijk aan de lijst met studenten):

```python
df = pd.DataFrame(data=punten, columns=['vak1', 'vak2', 'vak3', 'vak4', 'vak5', 'vak6'])
df.index = studenten
df
```

## Vraag 2

Wie is er niet van vak 2 gebuisd?

```python
beste_studenten = df[df.vak2 >= 10]
beste_studenten
```

## Vraag 3

Wie is er voor geen enkel vak gebuisd?

```python
goede_studenten = set(studenten)
for vak in df.columns:
    uitkomst = df[df[vak] < 10]
    goede_studenten = goede_studenten.difference(set(uitkomst.index))

print('Goede studenten:', goede_studenten)
```

## Vraag 4

Wie is de beste student (heeft de hoogste punten)?

Uitkomst: geef de naam van de student als een string.

Tip: gebruik ofwel `idxmax()` (zoek op op internet) ofwel een for-loop.

Met for-loop:

```python
beste_student = ''
beste_punt = 0
for student, punten in df.iterrows():
    som = punten.sum()
    print(student, som)
    if som > beste_punt:
        beste_punt = som
        beste_student = student
print('Beste student:', beste_student)
```

Met `idmax()`:

```python
print('Beste student:', df.sum(axis=1).idxmax())

# alternatief
print('Beste student:', df[df.sum(axis=1) == df.sum(axis=1).max()].index[0])
```
