## Vraag 1: vind_matrix_eigenschappen

Opgegeven: een matrix als NumPy `array` object. Deze functie moet controleren of de matrix vierkant is. Indien zo, moet er ook gecontroleerd worden of alle cijfers op de diagonaal van de matrix (van linksboven tot rechtsonder) hetzelfde zijn. De antwoorden (beide booleans) staan in `is_vierkant` en `diagonaal_juist`.

**Antwoord:** Twee onafhankelijke fouten. (1) De vierkant-check `np_matrix.shape[-1] == np_matrix.shape[-1]` vergelijkt een waarde met zichzelf — dat is altijd `True`, ongeacht de echte vorm van de matrix (het had `np_matrix.shape[0] == np_matrix.shape[1]` moeten zijn). Voor de eerste testmatrix (die toevallig wél vierkant is) valt dit niet op, maar bij de niet-vierkante matrix (2×3) geeft de functie toch `True` terug. (2) De diagonaalcheck gebruikt `is` in plaats van `==` om `np_matrix[i, i]` te vergelijken met `te_zoeken`. `is` test object-identiteit, niet waarde-gelijkheid — en twee keer dezelfde cel van een NumPy-array opvragen levert telkens een nieuw scalar-object op, ook al is de waarde identiek. Daardoor faalt de diagonaalcheck zelfs voor de matrix met één element, die uiteraard aan zichzelf gelijk is. Fix: gebruik `np_matrix.shape[0] == np_matrix.shape[1]` en vervang `is` door `==`.

```python
import numpy as np

def vind_matrix_eigenschappen(np_matrix):
    is_vierkant = False
    diagonaal_juist = False

    print('Vorm van de matrix:', np_matrix.shape)
    if np_matrix.shape[0] == np_matrix.shape[1]:
        is_vierkant = True

    if is_vierkant:
        diagonaal_juist = True
        te_zoeken = np_matrix[0, 0]
        for i in range(len(np_matrix)):
            if not np_matrix[i, i] == te_zoeken:
                diagonaal_juist = False
                break

    return is_vierkant, diagonaal_juist

matrix1 = np.array([[1]])
is_v, diag = vind_matrix_eigenschappen(matrix1)
print('Vierkant?', is_v)
print('Diagonaal juist?', diag)
assert is_v == True
assert diag == True

matrix2 = np.array([[1, 2, 3], [4, 5, 6]])
is_v, diag = vind_matrix_eigenschappen(matrix2)
print('Vierkant?', is_v)
print('Diagonaal juist?', diag)
assert is_v == False
assert diag == False
```

## Vraag 2: vermenigvuldig

Deze functie maakt een `factors`-matrix met dezelfde vorm als `matrix`: `factor` op de diagonaal, `factor - 1` overal elders. Vervolgens wordt `matrix` herhaaldelijk met die `factors`-matrix vermenigvuldigd (kruisproduct), tot het maximum van élke rij minstens de `ondergrens` haalt.

**Antwoord:** Twee samenhangende fouten. (1) `np.max(matrix, axis=0)` neemt het maximum per **kolom** in plaats van per **rij** (`axis=1`) — de opgave vraagt net het maximum van élke rij. Met de verkeerde as vergelijk je dus kolom-maxima met de ondergrens, niet rij-maxima. (2) De matrixvermenigvuldiging gebeurt als `np.dot(factors, matrix)` in plaats van `np.dot(matrix, factors)` — matrixvermenigvuldiging is niet commutatief, dus de volgorde van de argumenten bepaalt een volledig ander resultaat. Beide fouten zijn op zich al genoeg om de assert te doen falen, en samen leveren ze een uitkomst op die op het eerste gezicht nog plausibel oogt (dezelfde vorm, gelijkaardige grootteorde), maar cijfer voor cijfer verkeerd is. Fix: gebruik `axis=1` in de `np.max`-aanroep en `np.dot(matrix, factors)` voor de vermenigvuldiging.

```python
import numpy as np

def vermenigvuldig(matrix, factor, ondergrens):
    factors = np.zeros((matrix.shape))
    factors[:] = factor - 1
    for i in range(len(matrix[0])):
        factors[i, i] = factor

    while np.min(np.max(matrix, axis=1)) < ondergrens:
        matrix = np.dot(matrix, factors)

    return np.round(matrix, 1)

np.random.seed(10)
data = np.random.randint(1, 150, (5, 5))
uitkomst = vermenigvuldig(data, 1.2, 200)
print(uitkomst)
assert np.allclose(uitkomst, [[208.6, 324.6, 214.6, 263.6, 312.6],
       [317.2, 307.2, 202.2, 267.2, 194.2],
       [290. , 365. , 266. , 350. , 389. ],
       [317.8, 351.8, 370.8, 385.8, 325.8],
       [203.2, 270.2, 224.2, 247.2, 183.2]])
```
