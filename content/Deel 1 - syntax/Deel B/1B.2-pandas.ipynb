# 1B.2 Pandas

Pandas is een van de meestgebruikte dataopslag pakketten beschikbaar voor Python. Het vormt een schil van metadata (kolommen, kolomtypes, etc.) rond rauwe data opgeslagen in NumPy matrices en laat het toe die makkelijk te bevragen (met SQL-achtige methodes) en te bewerken.
Er worden veelal 2 objecten gebruikt om dit te bewerkstelligen: de `Series` en de `DataFrame`. De eerste wordt gebruikt om een enkele kolom te vatten met diens data en datatype. Over een enkele kolom kunnen ook statistieken worden berekend zoals `sum()` of `min()`. Een `DataFrame` daarentegen zorgt voor een volledige tabulaire representatie en is een verzameling van verschillende `Series`-objecten aan elkaar. Beiden maken uitgebreid gebruik van een index zoals hieronder te merken. Standaard worden de rijnummers meegenomen, maar soms worden deze overschreven, bv. met een klanten ID of zelfs een datum.

We importeren Pandas als `pd`:

```python
import pandas as pd
import numpy as np
```

## Series

De `Series` vat dus een enkele kolom (of vlak gezien, een rij). We kunnen ze aanmaken dmv bv. een lijst zoals hieronder:

```python
een_lijst = ['robin', 'ted', 'barney', 'lilly', 'marshall']
een_reeks = pd.Series(een_lijst)
print(een_reeks)
```

Hieronder is te zien dat we dus een `Series` object hebben aangemaakt. We kunnen de rauwe datawaarden (een NumPy array) opvragen met de methode `values`:

```python
print('De reeks is van het type:', type(een_reeks))
print('De rauwe waarden:', een_reeks.values)
print('De waarden zijn van het type `ndarray` uit NumPy:', type(een_reeks.values))
```

Pandas zal automatisch proberen om het datatype te herkennen. Hierboven gaven we strings mee die werden ondergebracht onder het type 'object' wat staat voor niet-geordende categorische waarden. Hieronder geven we nummers mee die automatisch herkend worden als integer uitgedrukt in 64 bits (`int64`):

```python
een_andere_lijst = [10, 20, 30, 40, 50, 50, 50]
een_andere_reeks = pd.Series(een_andere_lijst)
print('De andere reeks:\n', een_andere_reeks)
```

Pandas kan ook goed met lege cellen om (die worden bijgehouden als `np.nan` constanten). Zo zullen ze niet meegenomen worden in functies zoals `min()` of `mean()`:

```python
een_andere_lijst = [10, 20, 30, 40, 50, 50, 50, np.nan]
een_andere_reeks = pd.Series(een_andere_lijst)
print('NumPy methodes zijn mogelijk (min):', een_andere_reeks.min())
print('NumPy methodes zijn mogelijk (gem.):', een_andere_reeks.mean())
```

Lege cellen tellen mee voor `len()`, maar niet voor `value_counts()` - een handige functie om het aantal voorkomens van (categorische/discrete) variabelen op te vragen als dan niet met `normalize` als argument om de relatieve frequentie te verkrijgen - of `count()`:

```python
print('Aantal elementen:', len(een_andere_reeks))
print('Aantal voorkomens:\n', een_andere_reeks.value_counts())
print('Proportioneel voorkomen:\n', een_andere_reeks.value_counts(normalize=True))
print('Aantal niet-lege cellen:', een_andere_reeks.count())
```

We kunnen ook de lijst indexering van Python blijven gebruiken op `Series`:

```python
print('Eerste 2 elementen:\n', een_andere_reeks[:2])
```

Wat ook werkt is de `iloc()` functie voor index-localisatie. `DataFrame` objecten kunnen niet met de standaard-indexering gebruikt worden en moeten altijd met `iloc()` gebruikt worden.

```python
print('Eerste 2 elementen met iloc:\n', een_andere_reeks.iloc[:2])
```

Lambda-functies werken ook op `Series`, zoals hieronder geïllustreerd:

```python
print('Lambda functies werken ook:\n', een_andere_reeks.apply(lambda x: x**2))
```

## DataFrame

Een `DataFrame` zal dus meerdere `Series` aan elkaar rijgen in een tabel.

### Aanmaken van een DataFrame

Een `DataFrame` kan aangemaakt worden uit verschillende (even lange) lijsten die de aparte kolommen voorstellen. Kolommen moeten natuurlijk wel geïdentificeerd kunnen worden. Ofwel geven we een dictionary mee waarin de lijsten de waarden zijn en de kolomnamen de sleutels, ofwel geven we de lijsten mee in een lijst en geven we de kolomnamen mee als een aparte lijst aan de `DataFrame` functie onder het argument `columns`. Jupyter notebooks zijn ook voorzien op Pandas dataframes, zou moet de `print()` functie niet gebruikt worden om ze gestructureerd te visualiseren:

```python
namen_lijst = ['robin', 'ted', 'barney', 'lilly', 'marshall']
aantal_lijst = [10, 20, 30, 40, 50]

personages_dict = {'personages': namen_lijst, 'aantal': aantal_lijst}
df = pd.DataFrame(personages_dict)
df
```

```python
andere_data = [['johannes', 10], ['giovanni', 2], ['john', 3], ['jan', 30]]
df_2 = pd.DataFrame(andere_data, columns=['naam', 'aantal'])
df_2
```

De index kan als volgt worden opgevraagd:

```python
df_2.index
```

De index is van het type `Index` en heeft ook een NumPy array als waarden:

```python
print(type(df_2.index))
print(type(df_2.index.values))
df_2.index.values
```

We kunnen de index ook aanpassen, bv. door de namen te gebruiken als index:

```python
df_2.index = df_2['naam']
df_2
```

`iloc()` werkt ook nog:

```python
df_2.iloc[:2]
```

We kunnen bij een `DataFrame` ook `loc()` gebruiken door elementen van de index op te vragen:

```python
df_2.loc[['johannes', 'jan']]
```
