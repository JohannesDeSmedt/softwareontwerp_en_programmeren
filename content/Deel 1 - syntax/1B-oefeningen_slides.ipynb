# Oefeningen 1B.2

```python
mat = np.array([[1, 2, 3],
                [4, 5, 6]])
print(mat[1, 2], mat[0])
print(mat[:, 1])
print(mat.T.shape)

# VERBETER: haal de laatste kolom op (deze lijn crasht):
laatste_kolom = mat[2]
assert list(laatste_kolom) == [3, 6]
```

```python
mat = np.array([[1, 2, 3],
                [4, 5, 6]])
print(mat.sum())
print(mat.sum(axis=0))
print(mat.sum(axis=1))

# VERBETER: we willen het gemiddelde per rij:
gem_per_rij = mat.mean(axis=0)
print(gem_per_rij)
assert list(gem_per_rij) == [2.0, 5.0]
```

```python
punten = np.array([12, 17, 9, 15])
print(np.max(punten), np.argmax(punten))
print(punten.min(), np.mean(punten))

# VERBETER: welke index heeft de laagste score?
index_laagste = np.min(punten)
assert index_laagste == 2
```

```python
a = np.array([[1, 2], [3, 4]])
b = np.array([[1, 0], [0, 1]])
print(a * b)
print(a @ b)
print(a + np.array([10, 20]))

# VERBETER: bereken het matrixproduct van a en c:
c = np.array([[2, 0], [0, 2]])
product = a * c
assert (product == np.array([[2, 4], [6, 8]])).all()
```

# Oefeningen 1B.4

```python
r = np.array([4, 9, 1, 7, 3])
print(r > 3)
print(r[r > 3])
print((r > 3).sum())

# VERBETER: selecteer de elementen strikt tussen 2 en 8 (deze lijn crasht):
tussen = r[r > 2 and r < 8]
assert list(tussen) == [4, 7, 3]
```

```python
r = np.array([4, 9, 1, 7, 3])
print(np.where(r > 3, 1, 0))
print(np.where(r > 3, r, -1))
print(np.where(r > 3))

# VERBETER: vervang alle waarden kleiner dan 5 door 0, behoud de rest:
nieuw = np.where(r < 5, r, 0)
print(nieuw)
assert list(nieuw) == [0, 9, 0, 7, 0]
```

```python
kopie = r
kopie[kopie > 5] = 0
print(r)
print(np.unique(np.array([2, 1, 2, 3, 1])))

# VERBETER: r mag niet veranderen, maak een echte kopie:
r = np.array([4, 9, 1, 7, 3])
kopie = r
kopie[kopie > 5] = 0
assert list(r) == [4, 9, 1, 7, 3] and list(kopie) == [4, 0, 1, 0, 3]
```

# Oefeningen 1B.7

```python
df = pd.DataFrame({
    'provincie': ['Antwerpen', 'Limburg', 'Antwerpen', 'Limburg', 'Namen'],
    'SEX': ['M', 'F', 'F', 'M', 'F'],
    'CASES': [12, 3, 25, 8, 4]})

print(len(df[df['CASES'] > 5]))
print(list(df[df['CASES'] > 5]['provincie']))
print(df[(df['provincie'] == 'Limburg') & (df['CASES'] > 5)])

# VERBETER: selecteer de rijen uit Limburg of Namen (deze lijn crasht):
sel = df[df['provincie'] == 'Limburg' | df['provincie'] == 'Namen']
assert len(sel) == 3
```

```python
print(list(df.sort_values(by='CASES')['CASES']))
print()
print(df.groupby('provincie')['CASES'].sum())
print()
print(df.groupby('SEX')['CASES'].mean())
print()

# VERBETER: zoek de provincie met in totaal de meeste gevallen:
totaal = df.groupby('provincie')['CASES'].sum()
print(totaal)
print()
top = totaal.sort_values().index[0]
print(top)
assert top == 'Antwerpen'
```

```python
df.drop(['SEX'], axis=1)
print(list(df.columns))
print(len(df.drop([0, 1])))

# VERBETER: verwijder de kolom 'SEX' (deze lijn crasht):
zonder_sex = df.drop(['SEX'])
assert list(zonder_sex.columns) == ['provincie', 'CASES']
```

# Oefeningen 1B.8

```python
klanten = pd.DataFrame({'naam': ['an', 'bo', 'cis'],
                        'uitgaven': [600, 120, 300],
                        'loyaliteit': [3, 9, 5]})

def actie(rij):
    if rij['uitgaven'] > 500:
        return 'exclusief'
    elif rij['loyaliteit'] > 7:
        return 'loyaliteit'
    return 'standaard'

print(list(klanten.apply(actie, axis=1)))

# VERBETER: voeg de actie toe als nieuwe kolom (deze lijn crasht):
klanten['actie'] = klanten.apply(actie)
assert list(klanten['actie']) == ['exclusief', 'loyaliteit', 'standaard']
```

```python
print(list(klanten['uitgaven'].apply(lambda x: x * 12)))
print(klanten[klanten['naam'].isin(['an', 'cis'])]['uitgaven'].sum())

# VERBETER: bereken de jaarlijkse uitgaven als nieuwe kolom 'jaar' (deze lijn crasht):
klanten['jaar'] = klanten.apply(lambda x: x * 12)
assert list(klanten['jaar']) == [7200, 1440, 3600]
```

```python
korting = np.where(klanten['loyaliteit'] > 4, 0.5, 0.0)
print(korting)
print(list(klanten['uitgaven'] * (1 - korting)))

# VERBETER: segment is 'hoog' als de uitgaven minstens 300 zijn, anders 'laag':
klanten['segment'] = np.where(klanten['uitgaven'] > 300, 'hoog', 'laag')
assert list(klanten['segment']) == ['hoog', 'laag', 'hoog']
```
