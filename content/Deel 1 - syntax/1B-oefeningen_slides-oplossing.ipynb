# Oefeningen 1B.2

```python
mat = np.array([[1, 2, 3],
                [4, 5, 6]])
print(mat[1, 2], mat[0])
print(mat[:, 1])
print(mat.T.shape)

# VERBETER: haal de laatste kolom op (deze lijn crasht):
laatste_kolom = mat[:, 2]  # of mat[:, -1]
assert list(laatste_kolom) == [3, 6]
```

```python
print(mat.sum())
print(mat.sum(axis=0))
print(mat.sum(axis=1))

# VERBETER: we willen het gemiddelde per rij:
gem_per_rij = mat.mean(axis=1)  # langs de kolommen -> 1 waarde per rij
assert list(gem_per_rij) == [2.0, 5.0]
```

```python
punten = np.array([12, 17, 9, 15])
print(np.max(punten), np.argmax(punten))
print(punten.min(), np.mean(punten))

# VERBETER: welke index heeft de laagste score?
index_laagste = np.argmin(punten)
assert index_laagste == 2
```

# Oefeningen 1B.3

```python
a = np.array([[1, 2], [3, 4]])
b = np.array([[1, 0], [0, 1]])
print(a * b)
print(a @ b)
print(a + np.array([10, 20]))

# VERBETER: bereken het matrixproduct van a en c:
c = np.array([[2, 0], [0, 2]])
product = a @ c  # of np.dot(a, c)
assert (product == np.array([[2, 4], [6, 8]])).all()
```

```python
r = np.arange(6)
print(r.reshape((2, 3)))
print(r.reshape((3, 2)).shape)
print(np.concatenate((r[:2], r[-2:])))

# VERBETER: zet de 6 elementen om in een matrix met 3 rijen (deze lijn crasht):
m = r.reshape((3, 2))  # 3 x 2 = 6 elementen
assert m.shape == (3, 2)
```

```python
m = np.array([[3, 7, 2],
              [9, 1, 8]])
print(np.sort(m, axis=1))
print(np.sort(m, axis=0))
print(np.sort(m, axis=None))

# VERBETER: plak de twee matrices onder elkaar (4 rijen):
x = np.ones((2, 3))
y = np.zeros((2, 3))
samen = np.vstack((x, y))
assert samen.shape == (4, 3)
```

# Oefeningen 1B.4

```python
r = np.array([4, 9, 1, 7, 3])
print(r > 3)
print(r[r > 3])
print((r > 3).sum())

# VERBETER: selecteer de elementen strikt tussen 2 en 8 (deze lijn crasht):
tussen = r[(r > 2) & (r < 8)]  # & i.p.v. and, en haakjes rond elke conditie
assert list(tussen) == [4, 7, 3]
```

```python
print(np.where(r > 3, 1, 0))
print(np.where(r > 3, r, -1))
print(np.where(r > 3))

# VERBETER: vervang alle waarden kleiner dan 5 door 0, behoud de rest:
nieuw = np.where(r < 5, 0, r) # we moesten 0 en r omdraaien
assert list(nieuw) == [0, 9, 0, 7, 0]
```

```python
kopie = r
kopie[kopie > 5] = 0
print(r)
print(np.unique(np.array([2, 1, 2, 3, 1])))

# VERBETER: r mag niet veranderen, maak een echte kopie:
r = np.array([4, 9, 1, 7, 3])
kopie = r.copy() # we moeten een echte kopie maken, anders verwijst kopie naar hetzelfde object als r, dit is ingebouwd in numpy
kopie[kopie > 5] = 0
assert list(r) == [4, 9, 1, 7, 3] and list(kopie) == [4, 0, 1, 0, 3]
```

# Oefeningen 1B.5

```python
import pandas as pd
import numpy as np

s = pd.Series([10, 20, 20, np.nan])
print(len(s), s.count())
print(s.mean())
print(s.value_counts())

# VERBETER: tel het aantal niet-lege cellen:
aantal = s.count() # count() ipv len() want len telt ook de lege cellen mee
assert aantal == 3
```

```python
s = pd.Series(['an', 'bo', 'cis'], index=[10, 20, 30])
print(s.iloc[0], s.loc[20])
print(s.values)

# VERBETER: haal het eerste element op (deze lijn crasht):
eerste = s.iloc[0]  # s[0] zoekt naar het label 0 in de index
assert eerste == 'an'
```

```python
s = pd.Series([1, 2, 3])
print(s.apply(lambda x: x ** 2))
print(list(s * 10))
print(list(s[s > 1]))

# VERBETER: zet elk getal om in een tekst met een euroteken ervoor (deze lijn crasht):
euro = s.apply(lambda x: '€' + str(x)) # we moesten een lambda functie gebruiken, want str(getal) werkt niet, getal is niet gedefinieerd
assert list(euro) == ['€1', '€2', '€3']
```

# Oefeningen 1B.7

```python
df = pd.DataFrame({
    'provincie': ['Antwerpen', 'Limburg', 'Antwerpen', 'Limburg', 'Namen'],
    'SEX': ['M', 'F', 'F', 'M', 'F'],
    'CASES': [12, 3, 25, 8, 4]})

print(len(df[df['CASES'] > 5]))
print(list(df[df['CASES'] > 5]['provincie']))
print(df[(df['provincie'] == 'Limburg') & (df['CASES'] > 5)])

# VERBETER: selecteer de rijen uit Limburg of Namen (deze lijn crasht):
sel = df[(df['provincie'] == 'Limburg') | (df['provincie'] == 'Namen')]
# of: sel = df[df['provincie'].isin(['Limburg', 'Namen'])]
assert len(sel) == 3
```

```python
print(list(df.sort_values(by='CASES')['CASES']))
print(df.groupby('provincie')['CASES'].sum())
print(df.groupby('SEX')['CASES'].mean())

# VERBETER: zoek de provincie met in totaal de meeste gevallen:
totaal = df.groupby('provincie')['CASES'].sum()
top = totaal.sort_values(ascending=False).index[0]  # of totaal.idxmax() # of .index[-1]
assert top == 'Antwerpen'
```

```python
df.drop(['SEX'], axis=1)
print(list(df.columns))
print(len(df.drop([0, 1])))

# VERBETER: verwijder de kolom 'SEX' (deze lijn crasht):
zonder_sex = df.drop(['SEX'], axis=1)
assert list(zonder_sex.columns) == ['provincie', 'CASES']
```

# Oefeningen 1B.8

```python
klanten = pd.DataFrame({'naam': ['an', 'bo', 'cis'],
                        'uitgaven': [600, 120, 300],
                        'loyaliteit': [3, 9, 5]})

def actie(rij):
    if rij['uitgaven'] > 500:
        return 'exclusief'
    elif rij['loyaliteit'] > 7:
        return 'loyaliteit'
    return 'standaard'

print(list(klanten.apply(actie, axis=1)))

# VERBETER: voeg de actie toe als nieuwe kolom (deze lijn crasht):
klanten['actie'] = klanten.apply(actie, axis=1)  # axis=1: de functie krijgt 1 rij per keer
assert list(klanten['actie']) == ['exclusief', 'loyaliteit', 'standaard']
```

```python
print(list(klanten['uitgaven'].apply(lambda x: x * 12)))
print(klanten[klanten['naam'].isin(['an', 'cis'])]['uitgaven'].sum())

# VERBETER: bereken de jaarlijkse uitgaven als nieuwe kolom 'jaar' (deze lijn crasht):
klanten['jaar'] = klanten['uitgaven'].apply(lambda x: x * 12)  # of klanten['uitgaven'] * 12, of klanten.apply(lambda x: x['uitgaven'] * 12)
assert list(klanten['jaar']) == [7200, 1440, 3600]
```

```python
korting = np.where(klanten['loyaliteit'] > 4, 0.5, 0.0)
print(korting)
print(list(klanten['uitgaven'] * (1 - korting)))

# VERBETER: segment is 'hoog' als de uitgaven minstens 300 zijn, anders 'laag':
klanten['segment'] = np.where(klanten['uitgaven'] >= 300, 'hoog', 'laag')
assert list(klanten['segment']) == ['hoog', 'laag', 'hoog']
```
