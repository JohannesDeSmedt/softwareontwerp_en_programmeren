# 1A.7 Conditionals

Condities vormen de basisblokken van programmeertalen en computers in het algemeen. Denk maar aan de volgende voorbeelden:
- Moet dit datapakket naar adres X of Y gestuurd worden over het Internet?
- Kan deze gebruiker deze films zien gegeven zijn abonnement bij onze service?
- Welke berichten moeten een trilsignaal geven op een smartphone?
- Moeten deze uitgaven bij het vorige jaarresultaat geteld worden?
- ...

Condities worden gemaakt door bv. elementen te vergelijken, zien of bepaalde data beschikbaar is in een bepaalde collectie zoals een set, enz. Ze kunnen daarna gebruikt worden om het programma te sturen en slechts bepaalde code uit te voeren (of net niet) als er aan bepaalde voorwaarden voldaan is. Hiervoor worden `if`-statements gebruikt, in samenwerking met `else` en `elif` (een verkorting van else if).

Hier wordt een klein voorbeeld gegeven waarbij er iets anders afgeprint wordt naar gelang de waarde van variabele `a`:

```python
a = 10

# merk op, het dubbelpunt is verplicht
if a > 2:
    print('a is groter dan 2')
else:
    print('a is kleiner dan of gelijk aan 2')
```

## Indents

Het wordt ook duidelijk dat indents in Python (de inspringingen verkregen met tabs of 4 spaties) een grote impact hebben op de uitvoering. Alles op de indent 1 verder dan de `if` zal tijdens de `if` uitgevoerd worden, de code zonder indent (het `print()` statement) wordt terug na de `if` uitgevoerd:

```python
a = 10

if a > 2:
    print('\tBinnen het if-statement')
    print('\ta is groter dan 2')
print('Dit komt na het if-statement')
```

```python
if a > 2:
    print('\tBinnen het if-statement')
print('a is groter dan 2')
print('Dit komt na het if-statement')
```

## Geneste `if`-statements

Condities kunnen elkaar ook opvolgen en samengebracht worden in een gelaagde structuur (geneste `if`-statements). In dat geval zal de indent dus weer bepalen welke code er onder welke laag wordt uitgevoerd, zoals hier beneden te zien is:

```python
if a > 2:
    if a >= 4:
        print('a is groter dan of gelijk aan 4')
    else:
        print('a is groter dan 2 maar kleiner dan 4')
    print('Na de geneste if')
else:
    print('a is kleiner dan of gelijk aan 2')
```

Merk op dat hier dus eerst wordt gecontroleerd of `a` groter is dan 2, en daarna of `a` groter of gelijk is aan 4. M.a.w., enkel als $4>a>2$ is wordt de `else` van de binnenste `if` uitgevoerd.

Daarnaast kunnen we dit ook korter schrijven door een `elif`, een else if, te gebruiken die pas wordt gecontroleerd indien de conditie van de eerste `if` niet houdt. Zo kan er dus trapsgewijs worden gewerkt, ideaal voor intervallen of een reeks discrete waarden die overlopen moeten worden:

```python
if a >= 4:
    print('a is groter dan of gelijk 4')
elif a > 2:
    print('a is groter dan 2 maar kleiner dan 4')
else:
    print('a is kleiner dan of gelijk aan 2')
```

Geneste `if`-statements kunnen wel een gevaar vormen voor consistentie. Zo is het mogelijk dat bepaalde condities elkaar tegenspreken, of niet overeen komen. Hieronder bijvoorbeeld zal de eerste geneste `if` conditie nooit kunnen uitgevoerd worden:

```python
if a >= 10:
    if a < 5:
        print('dit wordt nooit uitgevoerd')
    else:
        print('a is groter dan of gelijk aan 10')
else:
    print('a kleiner dan 10')
```

Hieronder wordt de `elif` wel uitgevoerd, maar de volgorde is misschien niet ideaal:

```python
if a >= 10:
    print('a is groter dan of gelijk aan 10')
elif a <= 5:
    print('a is kleiner dan of gelijk aan 5')
else:
    print('a is kleiner dan 10 maar groter dan 5')
```

## Logische operatoren

Nu wordt het ook duidelijk waarom we de logische operatoren nodig hebben. Ze worden gebruikt als bouwblokken in condities en vormen de basis van algoritmes. Hieronder worden er enkele voorbeelden gegeven waarbij `and`, `or` en `not` gebruikt worden om vergelijkingen aan elkaar te lijmen. Ze worden steeds van links naar rechts geïnterpreteerd, vandaar dat het meestal aangeraden is om de conditie die de hoogste kans heeft om te falen (dus de laagste kans om waar te zijn, bv. het zoeken van een 107-jarige in een personendatabase) het meest links staat bij een `and`. Als allebei de statements moeten houden gaan we het minst moeten rekenen als we meteen een tegenvoorbeeld vinden. Bij `or` is het net andersom, we zetten de conditie die de hoogste kans heeft om waar te zijn links omdat we dan zo snel mogelijk de `if` kunnen evalueren - want 1 juist voorbeeld is al genoeg om de conditie waar te maken en anders moeten we de volgende conditie ook beginnen uitrekenen. 

Als we met lange lijsten en grote sets werken willen we ook meestal vermijden dat we daarmee aan de slag moeten als het niet nodig is, dus zullen we die het meest rechts plaatsen (indien mogelijk uiteraard).

```python
a = 30
b = 10
c = False

if a > 20 and b < 2:
    print('Eerste deel')
elif not c or a == 10:
    print('c houdt niet')
else:
    print('Else')

a = 10
b = 10
if a == b:
    print('a == b')
if a is b:
    print('a is b')
```
