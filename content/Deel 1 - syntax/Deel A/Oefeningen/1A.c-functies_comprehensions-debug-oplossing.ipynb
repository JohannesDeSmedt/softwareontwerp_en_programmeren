## Fout 1: vergeten return

Dit moet `True` printen voor een woord langer dan 5 tekens.

**Antwoord:** De functie berekent `resultaat` wel, maar geeft die nooit terug met `return`. Zonder expliciete `return` geeft een functie standaard `None` terug.

```python
def is_lang(woord):
    if len(woord) > 5:
        resultaat = True
    else:
        resultaat = False
    return resultaat

print(is_lang('programmeren'))
```

## Fout 2: mutable default argument

Dit moet telkens een **nieuwe** lijst met 1 item teruggeven.

**Antwoord:** Een default argument (`lijst=[]`) wordt maar één keer aangemaakt, bij het definiëren van de functie — niet bij elke aanroep. Elke aanroep zonder `lijst`-argument hergebruikt dus dezelfde lijst, die blijft aangroeien (`['a']`, dan `['a', 'b']`). Gebruik `lijst=None` en maak binnen de functie een nieuwe lijst aan.

```python
def voeg_toe(item, lijst=None):
    if lijst is None:
        lijst = []
    lijst.append(item)
    return lijst

print(voeg_toe('a'))
print(voeg_toe('b'))
```

## Fout 3: off-by-one in een comprehension

Dit moet de kwadraten van **alle** elementen in `lijst` teruggeven.

**Antwoord:** `range(len(lijst) - 1)` loopt maar tot index 3, dus het laatste element (5) wordt nooit verwerkt. Als je gewoon over de elementen zelf itereert (`for getal in lijst`) heb je bovendien geen indexering nodig en kan deze fout niet meer optreden.

```python
lijst = [1, 2, 3, 4, 5]
kwadraten = [getal ** 2 for getal in lijst]
print(kwadraten)
```

## Fout 4: comprehension voor neveneffecten

Dit moet `resultaten` vullen met de teruggegeven waarden van `print(x)` voor elke `x`.

**Antwoord:** `print()` print wel iets naar het scherm, maar geeft zelf `None` terug. De comprehension verzamelt dus de *return-waarden* van `print(x)`, niet de waarden van `x` zelf — het resultaat is `[None, None, None]`. Een comprehension is bedoeld om een nieuwe lijst van waarden op te bouwen, niet om een lus met neveneffecten (zoals printen) te vervangen.

```python
resultaten = [x for x in [1, 2, 3]]
print('resultaten:', resultaten)
```

## Fout 5: botsende sleutels in een dict comprehension

Dit moet elk woord in `woorden` koppelen aan zijn lengte.

**Antwoord:** De sleutel is hier `len(woord)`, en 'bank' en 'huis' hebben toevallig dezelfde lengte (4). Een dictionary kan een sleutel maar één keer bevatten, dus de tweede match (`'huis'`) overschrijft stilzwigend de eerste (`'bank'`) — zonder foutmelding. Het woord moet net de sleutel zijn (uniek), en de lengte de waarde.

```python
woorden = ['bank', 'straat', 'huis']
resultaat = {woord: len(woord) for woord in woorden}
print(resultaat)
```

## Fout 6: filter() zonder list()

Dit moet het aantal niet-lege strings in `data` printen.

**Antwoord:** `filter()` geeft geen lijst terug, maar een lui (`lazy`) `filter`-object dat pas bij het overlopen waarden produceert. Zo'n object heeft geen lengte. Wrap het resultaat in `list(...)` om er een gewone lijst van te maken.

```python
data = ['a', '', 'b', '']
schoon = list(filter(lambda s: s != '', data))
print(len(schoon))
```

## Fout 7: waarom lost een comprehension dit op?

Beide fragmenten filteren `lijst` op getallen ≤ 10. Voorspel de output van **beide** — waarom geeft de eerste (lus-gebaseerde) versie een fout resultaat, en de tweede (comprehension) wel het juiste?

*(uit 1D.oefeningen_op_fouten, oefening 10)*

**Antwoord:** De lus-versie past `lijst` aan terwijl ze er nog over itereert (zie ook fout 9 in 1B) — elementen schuiven op en `145` wordt overgeslagen. Een comprehension bouwt in de achtergrond een volledig **nieuwe** lijst op, één element per iteratie, zonder de originele `lijst` ooit aan te passen. Daardoor kan dit mutatie-probleem bij een comprehension niet optreden.

```python
lijst = [0, 10, 20, 145, 2]
nieuwe_lijst = [el for el in lijst if el <= 10]
print('Met een comprehension:', nieuwe_lijst)
```
