### Oefening 3
Laat deze code werken door 1 teken te veranderen:

```python
lijst = [1, 2, 3, 4]

print(lijst.remove(4))
```

We veranderen 5 naar 4, dit element zit wel in de lijst. Merk op, hier wordt het resultaat van `remove()` geprint, een methode van de klasse/type `list` die geen resultaat geeft, vandaar dat we dus `None` terugkrijgen. Merk op dat al deze functies geen resultaat teruggeven maar wel de lijst aanpassen:

```python
print(lijst.append(5))
print(lijst)
print(lijst.extend([10, 100]))
print(lijst)
```

Dit is verschillend van bv. `lower()` of `upper()`, methodes van het `str` type die een aangepaste versie van de string waarover ze worden opgeroepen teruggeven en de originele string niet aanpassen. Let dus altijd goed op wat je wilt bekomen en wat de methode doet: iets teruggeven, of het object bewerken:

```python
mijn_string = 'GioVaNNI'
print(mijn_string.lower())
print(mijn_string)
print(mijn_string.replace('NN', 'nn'))
print(mijn_string)
```

### Oefening 10
Hier wordt er geen fout gemaakt die de code tegenhoudt, maar toch wordt de lijst hieronder niet juist gefilterd, kan jij dit verhelpen?

```python
lijst = [0, 10, 20, 145, 2]

nieuwe_lijst = lijst.copy()
for el in lijst:
    if el > 10:
        nieuwe_lijst.remove(el)
lijst = nieuwe_lijst
print('Lijst op het einde:', lijst)
```

We gebruiken een nieuwe lijst die een kopij is van de originele.

Het helpt ook om met `enumerate` de lijst te doorlopen om te zien dat de index gewoon verderspringt:

```python
lijst = [0, 10, 20, 145, 2]

for ind, el in enumerate(lijst):
    print('Index:', ind, 'Element:', el, 'lijst voor verwijderen:', lijst)
    if el > 10:
        print('element groter dan 10, verwijderen:', el)
        lijst.remove(el)

print('Lijst op het einde:', lijst)
```

Hier vermijden we dat de lijst wordt aangepast tijdens het loopen, wat resulteert in een verkeerde verwijdering (de lijst wordt telkens anders, maar de loop en dus de waarde die aan `el` gegeven wordt lopen ondertussen al verder). Daarom maken we een nieuwe lijst `nieuwe_lijst` die we eerst alle elementen van `lijst` geven en maken we die leeg. Deze wordt niet gebruikt om over te loopen. Hier zien we hoe het in verschillende gevallen fout gaat:

```python
lijst = [20, 30, 50, 40]

for el in lijst:
    print(lijst)
    if el > 10:
        lijst.remove(el)
print('Lijst op het einde:', lijst)

lijst = [20, 30, 50, 40, 60]

for el in lijst:
    print(lijst)
    if el > 10:
        lijst.remove(el)
print('Lijst op het einde:', lijst)
```

Een alternatief is een lijst behouden met de elementen die verwijderd moeten worden:

```python
lijst = [0, 10, 20, 145, 2]

te_verwijderen = []
for el in lijst:
    if el > 10:
        te_verwijderen.append(el)
for element in te_verwijderen:
    lijst.remove(element)
        
print('Lijst op het einde:', lijst)
```

Merk op dat we niet gewoon de lijst kunnen gelijkstellen aan `lijst.remove()` vermits de methode van de klasse lijst `None` teruggeeft:

```python
lijst = [0, 10, 20, 145, 2]

for el in lijst:
    if el > 10:
        lijst = lijst.remove(el)
print('Lijst op het einde:', lijst)
```

Een comprehension doet dit wel juist (waarom?):

```python
lijst = [0, 10, 20, 145, 2]

nieuwe_lijst = [el for el in lijst if el <= 10]
print('Nieuwe lijst:', nieuwe_lijst) 
```

De comprehension maakt in de achtergrond een nieuwe lijst aan per iteratie, dit vermijdt dat de originele variabele `lijst` wordt aangepast.
