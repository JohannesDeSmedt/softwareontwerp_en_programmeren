## Voorbeeld 3

Gebruik een filter + lambda functie die enkel de woorden uit een zin overhoudt die geen cijfers bevatten:

```python
voorbeeldzin = 'Ik heb 100 peren gekocht aan 20 euro in 2 winkels'

resultaat = filter(lambda x: x.isalpha(), voorbeeldzin.rsplit())
print('Woorden zonder cijfers in de zin:', list(resultaat))

# For-lus equivalent
resultaat = []
for woord in voorbeeldzin.rsplit():
    if woord.isalpha():
        resultaat.append(woord)
        
print('Woorden zonder cijfers in de zin:', resultaat)
```

## Voorbeeld 4

Maak een comprehension om een lijst te maken met enkel getallen deelbaar door 3 (onder 50):

```python
deelbaar_door_drie = [x for x in range(0, 50) if x % 3 == 0]

print('Getallen onder 50 deelbaar door 3:', deelbaar_door_drie)

# For-lus equivalent
deelbaar_door_drie = []
for i in range(0, 50):
    if i % 3 == 0:
        deelbaar_door_drie.append(i)
        
print('Getallen onder 50 deelbaar door 3:', deelbaar_door_drie)
```

## Voorbeeld 5

Maak een comprehension om een dictionary aan te maken die telt hoeveel getallen er deelbaar zijn door 3, 6 en 9 onder de 30 (je hebt hiervoor een comprehension in een comprehension nodig)"

```python
nieuwe_dictionary = {x: [y for y in range(0, 30) if y % x == 0] for x in [3, 6, 9]}
print('Getallen onder 30 deelbaar door 3/6/9:', nieuwe_dictionary)

# For-lus equivalent
nieuwe_dictionary = {}
for x in [3, 6, 9]:
    getallen = []
    for getal in range(0, 30):
        if getal % x == 0:
            getallen.append(getal)
    nieuwe_dictionary[x] = getallen
    
print('Getallen onder 30 deelbaar door 3/6/9:', nieuwe_dictionary)
```

## Voorbeeld 6

Maak een comprehension om een dictionary aan te maken die onder het deelnemersnummer (oplopend) het bmi opslaat ($\frac{gewicht}{lengte ^ 2}$).
Gegeven: een lijst van tuples met daarin een lengte en gewicht.
Rond af tot 2 cijfers na de komma.

```python
observaties = [(1.78, 65), (1.86, 75), (1.5, 55), (1.65, 62)]

bmis = {nummer: round(y / x ** 2, 2) for nummer, (x, y) in enumerate(observaties)}
print('Bmi van deelnemers:', bmis)

# For-lus equivalent
bmis = {}
for nummer, (lengte, gewicht) in enumerate(observaties):
    bmis[nummer] = round(gewicht/lengte ** 2, 2)
    
print('Bmi van deelnemers:', bmis)
```
