### a) While
Schrijf een while loop die de index geeft voor het eerste getal dat kleiner is dan het getal in opgegeven variabele `a`. Gebruik een variabele `index` om bij te houden waar je in de lijst zit. De boolean `verdergaan` stopt de while lus als de oplossing gevonden is en `resultaat` houdt de uiteindelijke index bij (we zetten deze eerst gelijk aan de laatste index van de lijst als default oplossing):

```python
lange_lijst = [10, 88, 2992, 20, 8, 10, 53]
a = 10
verdergaan = True
index = 0
resultaat = len(lange_lijst) - 1

# JOUW CODE
# loop
while verdergaan:
    print(f'Getal op index {index} is {lange_lijst[index]}')
    # controleer huidig element en indien < a
    if lange_lijst[index] < a:
        print('Kleiner dan a, we stoppen')
        # bewaar resultaat
        # stop de lus   
        resultaat = index
        verdergaan = False
    index += 1

    
print('Index van eerste getal kleiner dan a:', resultaat)
```
