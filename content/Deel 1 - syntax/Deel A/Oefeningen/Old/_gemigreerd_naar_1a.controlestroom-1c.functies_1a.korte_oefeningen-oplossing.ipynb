## Oefening 1

### a) While
Schrijf een while loop die de index geeft voor het eerste getal dat kleiner is dan het getal in opgegeven variabele `a`. Gebruik een variabele `index` om bij te houden waar je in de lijst zit. De boolean `verdergaan` stopt de while lus als de oplossing gevonden is en `resultaat` houdt de uiteindelijke index bij (we zetten deze eerst gelijk aan de laatste index van de lijst als default oplossing):

```python
lange_lijst = [10, 88, 2992, 20, 8, 10, 53]
a = 10
verdergaan = True
index = 0
resultaat = len(lange_lijst) - 1

def een_functie():
    print('Doet niets')
    print('Doet echt niets')

while verdergaan:
    print(lange_lijst[index])
    if lange_lijst[index] < a:
        verdergaan = False
        resultaat = index
        een_functie()
    index += 1
print('Index van eerste getal kleiner dan a:', resultaat)
```

### b) For

Doe nu hetzelfde met een for-loop:

```python
a = 10
resultaat = len(lange_lijst)

for i in range(0, len(lange_lijst)):
    if lange_lijst[i] < 10:
        resultaat = i
        break
print('Index van eerste getal kleiner dan a:', resultaat)
```

Alternatief met enumerate:

```python
a = 10
resultaat = len(lange_lijst)

for i, element in enumerate(lange_lijst):
    if element < 10:
        resultaat = i
        break
print('Index van eerste getal kleiner dan a:', resultaat)
```

Doorloop de volgende lijst vanaf het 2de element (index 1) en geef telkens het 2de volgende element terug:

## Oefening 3

Gebruik de `format` functie en het %-teken om de volgende string te vervolledigen en het eerste kommagetal tot 3 cijfers en het tweede kommagetal tot 2 cijfers na de komma af te ronden:

```python
getal_1 = 0.022292
getal_2 = 0.30993

print('Ik print 2 getallen af, getal 1 is %.3f en getal 2 is %.2f' % (getal_1, getal_2))
```

Doe nu hetzelfde maar enkel met de `format` functie:

```python
getal_1 = 0.022292
getal_2 = 0.30993

print('Ik print 2 getallen af, getal 1 is {0:.3f} en getal 2 is {1:.2f}'.format(getal_1, getal_2))
```

# Oefening 4

Maak een functie die een ongespecifieerd aantal getallen en een deler neemt als argument en een boolean teruggeeft om aan te geven of de som van de getallen geheel deelbaar is door de deler.

```python
def mijn_eerste_functie(*args, deler) -> bool:
    som = 0
    for a in args:
        som += a    
    return som % deler == 0


getal_1 = 10
getal_2 = 22
getal_3 = 20
print('1. Restdeling van som 0?', mijn_eerste_functie(getal_1, deler=10))
print('2. Restdeling van som 0?', mijn_eerste_functie(getal_1, getal_2, deler=10))
print('3. Restdeling van som 0?', mijn_eerste_functie(getal_1, getal_2, getal_3, deler=52))
```
