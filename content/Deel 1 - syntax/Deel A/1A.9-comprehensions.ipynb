# 1A.9 Comprehensions

Comprehensions, of samentrekkingen (niet de juiste vertaling, maar het leek de auteur de meest gepaste) laten toe om Python code wat korter te maken. Ze zijn vooral nuttig om snel nieuwe collecties aan te maken zoals lijsten, sets en dictionaries zoals hieronder duidelijk zal worden.

## Lijst comprehensions

We weten allemaal ondertussen hoe lijsten werken. Hieronder wordt een lege lijst gemaakt en gevuld met de kwadraten van een reeks getallen:

```python
eerste_lijst = []
for i in range(5):
    eerste_lijst.append(i ** 2)
print('Eerste lijst:', eerste_lijst)
```

Het gebruik van de for-lus is nogal omslachtig. Daarom kan dit in Python korter met een comprehension door de de `for` binnen vierkante haken te gebruiken waarbij de elementen die in de lijst moeten komen als eerste komen, gevolgd door de for-lus:

```python
een_lijn_lijst = [i ** 2 for i in range(5)]
print('1-lijn lijst:', een_lijn_lijst)
```

Merk op dat we nu ook geen lege lijst meer moeten maken en die vullen.

We kunnen ook condities inbouwen bij onze lus. Die condities kunnen dan achteraan geplaatst worden, na de for-lus. Deze kunnen bijvoorbeeld als filter dienen om een lijst te overlopen en enkel bepaalde getallen te weerhouden:

```python
een_andere_lijst = [i ** 2 for i in range(20) if i % 2 == 0 if i > 10]
print('Kwadraten van even getallen onder 20 maar boven 10:', een_andere_lijst)
```

We kunnen zelfs condities inbouwen aan het begin van de comprehension om zo te wisselen tussen eindwaarden naar gelang deze condities. Hieronder zal bv. `i` enkel in de lijst geplaatst worden zolang het een even getal is, anders wordt er een 0 geplaatst in de lijst:

```python
# met alternatieve waarde
een_laatste_lijst = [i ** 2 if (i % 2 == 0) else 0 for i in range(10)]
print('Een laatste lijst:', een_laatste_lijst)
```

We kunnen onze comprehensions ook verrijken met (zelfgemaakte) functies, zoals hieronder:

```python
def doe_iets_met_een_element(element: str) -> str:
    return '__' + element.lower() + '__'

string = 'legendary'
een_nieuwe_string = [doe_iets_met_een_element(i) for i in string]
print('Een nieuwe string:', een_nieuwe_string)
```

## Set comprehensions

Ook sets kunnen aangemaakt worden met comprehensions. In dit geval zullen dan de accolades gebruikt worden. Voor de rest blijft alles hetzelfde als bij lijst comprehensions met het verschil dat dubbele elementen eruit gefilterd worden:

```python
woord_met_rare_tekens = 'joh@ann3s'
letters_uit_alfabet = {i for i in woord_met_rare_tekens if i.isalpha()}
print('Letters van het woord:', letters_uit_alfabet)
```

## Dictionary comprehensions

Ten slotte zijn er ook nog dictionary comprehensions mogelijk. Die gebruiken, net als set comprehensions, accolades maar zullen dus zowel een sleutel als waarde moeten specifiëren gescheiden door een dubbelpunt. Hieronder wordt er een dictionary aangemaakt die telt hoeveel keer elk teken in een string voorkomt:

```python
# net hetzelfde maar met een dubbelpunt ertussen
woord_met_rare_tekens = 'joh@ann3s'
letter_count = {letter: woord_met_rare_tekens.count(letter) for letter in woord_met_rare_tekens}
print('Aantal voorkomens van de letters:', letter_count)
```
