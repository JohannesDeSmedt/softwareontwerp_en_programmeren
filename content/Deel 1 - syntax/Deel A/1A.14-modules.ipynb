# 1A.14 Modules

Python is vooral groot geworden door de integratie met verschillende externe modules/pakketten. Hieronder wordt aangetoond hoe ze makkelijk kunnen toegevoegd worden aan de code en daarna gebruikt om functies en attributen/variabelen in te laden.

Externe modules worden geïmporteerd met het keyword `import`, gevolgd door de naam. Er zijn enkele standaard modules die in Python zitten, maar niet meteen voor elk project/notebook worden opgeroepen (vooral om geheugen te besparen). Hieronder wordt het `math` pakket opgehaald. Daarna kan `math` dan met een punt bepaalde methodes zoals `cos` en attributen zoals `pi` oproepen: 

```python
import math

print('Pi:', math.pi)
print('cosinus(10):', math.cos(10))
```

We kunnen een overzicht krijgen van alle code die beschikbaar is binnen de 'directory' - afgekort `dir` - van het pakket. Kort gezegd zien we zo welke functies/attributen er beschikbaar zijn binnen de opgevraagde module:

```python
for m in dir(math):
    print(m)
```

Merk op dat we wel niet weten welke elementen van `dir` attributen en welke methodes zijn. Hier kunnen we het wat afleiden uit de context (`sin` is een methode, `e` is een attribuut/getal).

We kunnen ons ook beperken tot het importeren van bepaalde elementen van een module. Dit vermijdt dat we telkens alle code in het geheugen laden (en het maakt ook dat onze eigen functies niet overschreven worden omdat we ze bv. 'sin' hadden genoemd). Hieronder worden enkel `pi` en `cos` binnengehaald. Merk op dat we met de `type()`-functie nagaan of we met een methode of getal te maken hebben, want methodes zijn ook objecten van het type/de klasse functie:

```python
from math import pi, cos
print('Type van pi:', type(pi))
print('Type van cos:', type(cos))
print('Cosinus(0):', cos(0))
```

Zo zien we dus dat `cos` een 'builtin_function_or_method' is en `pi` gewoon een floating point nummer.

Een andere handige module die in Python zit is `time`. Deze kan de huidige tijd opvragen met behulp van de functie `time()` (beetje verwarrend). We kunnen de module bv. gebruiken om de code 1 seconde te laten pauzeren:

```python
import time

eerder = time.time()
print('Huidige tijd (aantal seconden sinds 01/01/1970):', eerder)

# pauseer het programma 1 seconde
time.sleep(1)
later = time.time()
print('Duur van uitvoeren (s):', later - eerder)
```

Er is een klein verschil met exact 1 seconde omdat de computer uiteraard de code ook nog moet draaien en al de andere processen in de achtergrond (jouw andere browser tabs, Spotify, Word, etc.) ook draaiende moet houden.

Ten slotte kijken we hier al eens naar NumPy. Deze module komt later nog uitvoerig terug in het 3de deel van deze cursus. Merk op dat NumPy meestal met een alias wordt binnengehaald zodat niet telkens 'numpy' voluit moet gebruikt worden. Enkele conventies binnen data analytics qua alias:
- NumPy -> `np`
- pandas -> `pd`
- TensorFlow -> `tf`

Hieronder gebruiken we dan het alias van de module om de methode die het gemiddelde van een lijst berekent op te vragen en daarna ook het attribuut `pi`:

```python
import numpy as np

gem = np.mean([1, 2, 3])
pi = np.pi
print(round(gem * pi, 3))
```

Ten slotte kunnen we ook geïmporteerde attributen of methodes een alias geven. Hieronder wordt bv. `mean` hernoemd naar `gemiddelde`. Daarna zal `gemiddelde` als functie beschikbaar zijn (omdat `mean` een functie is):

```python
from numpy import mean as gemiddelde

print(type(gemiddelde))
print('Gemiddelde van:', gemiddelde([10, 12, 14]))
```
