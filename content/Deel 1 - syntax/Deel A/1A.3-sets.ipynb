# 1A.3 Sets

Sets (of verzamelingen) zijn een handig instrument om logica uit te bouwen en een belangrijk onderdeel van veel algoritmes.

Sets hebben bepaalde eigenschappen. Ten eerste moet elk element uniek zijn (maar niet noodzakelijk van hetzelfde type). Als je een element 2 keer in een set stopt, dan zal de set nog altijd slechts 1 keer dat element bevatten. Daarnaast laten sets verschillende operaties toe tussen sets, die je hopelijk nog kent uit verzamelingenleer:
- De unie ($A \cup B$): is de verzameling van alle elementen die of in $A$ of in $B$ zitten.
- De doorsnede ($A \cap B$): is de verzameling van alle elementen die zowel in $A$ als in $B$ zitten.
- Het verschil ($A\setminus B$): is de verzameling van alle elementen die enkel in $A$ zitten en niet in $B$.

Nog een belangrijke eigenschap: sets hebben, in tegenstelling tot bv. lijsten, geen volgorde.

Een set wordt gemaakt dmv de ingebouwde `set()`-functie (ook wel de constructor van het type `set`, zie later bij objecten), om een lege set aan te maken, of door meteen een set te vullen met de meegegeven elementen tussen accolades:

```python
mijn_set = {1, 2, 3}
print(mijn_set)
nieuwe_set = mijn_set - {2}
print(nieuwe_set)
print(type(mijn_set))
print('Aantal elementen in mijn_set:', len(mijn_set))
```

```python
kleine_set = set() # lege set
print(kleine_set)
```

Elementen kunnen toegevoegd worden met de `add()` methode (een functie die eigen is aan alle variabelen van het type/de klasse `set`):

```python
kleine_set.add(1)
kleine_set.add('johannes')
kleine_set.add(1)
kleine_set.add(2)

# merk op dat 1 maar 1 keer aanwezig is
print(kleine_set)
```

Deze set wordt gemaakt van een lijst (die gebruiken vierkante haken). Lijsten kunnen wel duplicaten bevatten, deze bewerking haalt dus bv. alle duplicaten uit een lijst:

```python
andere_set = set([1, 2, 3, 2])
print(andere_set)
```

Elementen moeten ook niet van hetzelfde type zijn, hier mixen we integers en strings:

```python
laatste_set = {1, 5, 'johannes'}
print(laatste_set)
print(type(laatste_set))
```

Je kan controleren of een bepaald element in een set zit met het `in` keyword (analoog aan de $\in$ operator uit wiskunde). Deze oproep zal een boolean teruggeven (of het element dus in de set zit ja/nee).
Dit zullen we vaak gebruiken in combinatie met een `if`-statement later. 

```python
nieuwe_set = {1, 2, 3, 5} # nog een manier om een set aan te maken
print(1 in nieuwe_set)
print(4 in nieuwe_set)
```

Let wel, je kan niet zomaar eender welk type in een set stoppen. Omdat Python moet kunnen controleren of er duplicaten inzitten, moeten de elementen/objecten 'hashable' zijn. Om niet te technisch te gaan mag je ervan uit gaan dat alle eenvoudige types zoals strings en integers erin kunnen, maar bv. geen lijsten of sets zelf (als element, je kan wel een lijst meegeven aan een constructor zoals voorheen gezien, dan wordt die lijst omgezet in een set):

```python
foute_set = {1, 2, 3}
foute_set.add([1, 2])
```

Hieronder worden de verschillende setoperatoren geïllustreerd met behulp van de verschillende methodes beschikbaar voor een set:

```python
set_1 = {1, 2, 3, 4, 5}
set_2 = {1, 2, 3}
set_3 = {4, 6, 7}
print('Unie (geeft set terug):', set_1.union(set_3))
print('Verschil (geeft set terug):', set_1.difference(set_2))
print('Doorsnede (geeft set terug):', set_1.intersection(set_2))
print('Controleer op subsets (geef boolean terug):', set_2.issubset(set_1))
```

Je kan deze bewerkingen ook uitvoeren met behulp van enkele symbolen ipv met methodes, waardoor je ook meer dan 2 sets kan gebruiken:

```python
set_1 = {1, 3, 4}
set_2 = {1, 2}
set_3 = {4, 6}
print('Unie:', set_1 | set_2 | set_3, 'gelijk aan', set_1.union(set_2).union(set_3))
print('Doorsnede:', set_1 & set_2, 'gelijk aan', set_1.intersection(set_2))
print('Verschil:', set_1 - set_2, 'gelijk aan', set_1.difference(set_2))
```

Ten slotte kunnen we sets ook makkelijk uitbreiden en verkleinen mbv enkele methodes:

```python
set_4 = {1, 4, 'Johannes'}
# voeg een element toe
set_4.add('Johanna')
set_4.update([1, 2, 3])  # voeg meerdere elementen uit een itereerbaar object toe
print(set_4)
# verwijder een willekeurig element
set_4.pop()
print(set_4)
# verwijder element 'Johannes', of het aanwezig is of niet
set_4.discard('Johannes')
print(set_4)
set_4.discard('Johannes')  # geen foutmelding hoewel element niet (meer) aanwezig
print(set_4)
# verwijder element 'Johannes' als het bestaat en geef een foutmelding als het niet bestaat in de set
set_4.remove('Johannes')  # dit geeft een foutmelding
```

Met operatoren:

```python
# werkt niet: '+'
set_4 = {1, 4, 'Johannes'}
set_4 = set_4 + {'Johanna'}
print(set_4)
```

```python
# werkt wel: '-' (zoals discard, niet remove): 
set_4 = {1, 4, 'Johannes'}
set_4 = set_4 - {'Johannes'}
print(set_4)
set_4 = set_4 - {'Johannes'}
print(set_4)
```
