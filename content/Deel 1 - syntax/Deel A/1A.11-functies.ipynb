# 1A.11 Functies

We hebben ondertussen al heel wat functies zien voorbijkomen die standaard in Python zitten. Daarnaast hebben we ook al heel wat methodes gezien die eigen zijn aan bepaalde types/klasses zoals strings of lijsten. Nu gaan we leren om zelf functies te maken. Deze hebben verschillende voordelen:
- Ze houden code overzichtelijk
- Ze kunnen op verschillende plaatsen herbruikt worden
- Ze kunnen logica afschermen om inkapseling te vergemakkelijken (niet echt relevant hier, behalve dat je bv. niet moet weten hoe `mean()` in NumPy wordt berekend)

Functies hebben, zoals voorheen gezien, argumenten die we meegeven, en meestal ook een output. Let op, vermits Python geen sterk-getypeerde taal is zijn er geen eisen om de types van een argument (en de output) vast te leggen. Hierdoor kan je snel fouten maken door bv. een string mee te geven aan een afrondingsfunctie specifiek voor floats, e.d. In andere talen worden deze fouten er meestal uitgehaald zelfs voor de compilatie vermits er al een type check wordt uitgevoerd tijdens het schrijven van de code in een editor. Let dus altijd goed op en controleer van welk type de argumenten zijn en ook wat het type is van de output die je krijgt/zelf wegschrijft. Om dit toch iets of wat te vergemakkelijken, gaan we gebruik maken van *type hinting*. We geven het beoogde type van een argument mee achter de naam van het argument, gescheiden met een `:`, bv. `input_string: str`. Type hinting is louter informatief en wordt niet afgedwongen.

Hieronder wordt er een kleine functie gegeven die de lengte van 2 strings optelt. Een functie begint met het `def` keyword, gevolgd door een naam met daarachter ronde haakjes waartussen eventueel argumenten kunnen. Ten slotte volgt een dubbelpunt en gebruiken we wederom inspringingen/indents om de code aan de functie te linken. Ten slotte wordt er ook iets teruggeven, namelijk een integer met het aantal tekens in beide strings. Die wordt aangeduid door het `return` keyword, wat altijd als laatste in de control flow moet (let op, het kan dus ook dat er meerdere `return` keywords voorkomen als we bv. m.b.v. condities verschillende uitkomsten willen geven - zie later). Type hinting kan hier ook door het beoogde type van de output van een functie meegeven door `-> str` achter de haakjes van een functie te plaatsen.

```python
def geef_lengte_van_strings(string_1: str, string_2: str) -> str:
    print('String 1:', string_1)
    print('String 2:', string_2)
    return len(string_1) + len(string_2)

a = 'Johannes'
b = 'De Smedt'
totale_lengte = geef_lengte_van_strings(a, b)
print(f'Lengte van {a} en {b} opgeteld:', totale_lengte)
```

Merk op, type hinting wordt dus niet afgedwongen, want hieronder geven we gewoon 2 integers mee en krijgen een integer als resultaat:

```python
def geef_lengte_van_strings(string_1: str, string_2: str) -> str:
    print('String 1:', string_1)
    print('String 2:', string_2)
    return string_1 + string_2

a = 10
b = 20
totale_lengte = geef_lengte_van_strings(a, b)
print(f'{a} en {b} opgeteld:', totale_lengte)
```

## Scope

Belangrijk bij het gebruik van functies is dat iedere variabele een 'scope' heeft. Een scope bepaalt waar een variabele gekend is en gebruikt kan worden. Een variabele in het 'hoofdstuk' - dus het meest links gealigneerd - van het programma zal overal gekend zijn, terwijl een variabele die in een functie wordt aangemaakt (dus daar een eerste keer wordt toegewezen) alleen daar gekend zal zijn. 
Dit gaat soms wat verloren in Jupyter notebook omdat alles in meerdere cellen wordt gepropt.

Dit wordt hieronder duidelijk gemaakt m.b.v. `a`, een variabele die 'globaal' beschikbaar is, en `variabele_binnen_functie`, een variabele die niet buiten de functie kan opgeroepen worden en dus alleen 'lokaal' binnden de functie beschikbaar is:

```python
def geef_lengte_van_strings(string_1: str, string_2: str) -> str:
    print('a is hier ook beschikbaar:', a)
    variabele_binnen_functie = 'functiestring'
    print(variabele_binnen_functie)
    return len(string_1) + len(string_2)

a = 'Johannes'
b = 'De Smedt'
totale_lengte = geef_lengte_van_strings(a, b)
print(f'Lengte van {a} en {b} opgeteld:', totale_lengte)
try:
    print(variabele_binnen_functie) # dit geeft een fout
except NameError as e:
    print(f"Fout: {e}")
```

Over het algemeen wordt het afgeraden om globale variabelen te gebruiken. Als je een variabele van buiten de functie nodig hebt binnen een functie, is het beter om deze mee te geven als argument aan de functie. Dit vermijdt verwarring, fouten en zorgt ervoor dat functies 'self-contained' blijven. Indien we echt een globale variabele willen gebruiken kan deze best zo worden aangeduid met het `global` keyword (dit is verplicht als een globale variabele in een functie voor het eerst wordt aangemaakt - we zullen dit later nodig hebben voor recursie).

We hebben al gezien dat argumenten kunnen meegegeven worden. Ze krijgen hun eigen variabelenaam toegewezen in de functie tussen de haakjes. Daarnaast kunnen we ook default waarden meegeven aan onze argumenten. Let wel, deze moeten dan als laatste worden meegegeven tussen de haakjes. Hieronder wordt eerst een juist voorbeeld meegegeven en daarna een voorbeeld dat zondigt tegen deze laatste regel (we krijgen een `SyntaxError`, deze code wordt zelfs tegengehouden nog voor die kan beginnen draaien omdat ze fout is geschreven):

## Argumenten doorgeven bij referenties

Hieronder wordt ook duidelijk gemaakt waarom het geen goed idee is om zomaar een lijst te beginnen manipuleren met een functie. Als een lijst als argument wordt meegegeven, zal alleen de verwijzing worden meegegeven en niet een kopij van het object. M.a.w., alles wat er in de functie gebeurt zal op de oorspronkelijke lijst uitgevoerd worden en dus ook een effect hebben op de originele variabele, desondanks het feit dat de lijst een andere variabelenaam had als argument:

```python
def verwijder_eerste_element_uit_lijst(lijst_in: list) -> list:
    del lijst_in[0]

een_lijst = [0, 1, 2, 3]
print('Lijst in het begin:', een_lijst)
verwijder_eerste_element_uit_lijst(een_lijst)
print('Lijst na de functie:', een_lijst)
```

Hieronder wordt `copy()` gebruikt om de lijst eerst te kopiëren: 

```python
def geef_lijst_zonder_eerste_element_uit_lijst(lijst_in: list) -> list:
    lijst_2 = lijst_in.copy()
    del lijst_2[0]
    return lijst_2

een_lijst = [0, 1, 2, 3]
print('Lijst in het begin:', een_lijst)
nieuwe_lijst = geef_lijst_zonder_eerste_element_uit_lijst(een_lijst)
print('Lijst na de functie:', een_lijst)
print('Resultaat van de functie:', nieuwe_lijst)
```

En Python zou Python niet zijn moest dit niet belangrijk zijn voor basistypes, waar wel automatisch kopijen worden meegegeven aam functies:

```python
def verdubbel_integer(x: int):
    print('x voor:', x)
    x = x * 2
    print('x na:', x)

y = 10
print('y eerst:', y)
verdubbel_integer(y)
print('y na functie:', y)
```

## `return`

Een `return` moet op het einde van de control flow van een functie komen, anders zou het kunnen dat de functie soms wel iets en soms niets teruggeeft. Functies moeten niet noodzakelijk iets teruggeven, maar als er gevallen zijn waarin ze wel iets teruggeven kunnen ze niet kiezen om dit niet te doen voor uitzonderingen.
Hieronder heeft `vind_getal_in_lijst()` verschillende `return` mogelijkheden. Eerst een `return` die aan een conditie voldoet en daarna een standaard `return` die uitgevoerd wordt indien de conditie niet voldaan is (dit is uiteraard een heel simpel voorbeeld - als oefening, zoek een interessanter voorbeeld).

Daarnaast is het ook mogelijk dat `return` gebruikt wordt om een functie vroeger te doen stoppen, want vanaf de `return` bereikt wordt, wordt de functie afgebroken. Hieronder wordt er een lijst doorlopen met een `for`, die vroeger wordt gestopt omdat het laagste getal in de lijst al gevonden wordt voor alle elementen van de lijst doorlopen zijn:

```python
def vind_getal_in_lijst(lijst_in: list, getal: int) -> bool:
    if getal in lijst_in:
        return True
    return False

def vind_lager_getal_in_lijst_1(lijst_in: list, getal: int) -> int:
    for element in lijst_in:
        print('Element:', element)
        if element < getal:
            print('Return')
            return getal

def vind_lager_getal_in_lijst_2(lijst_in: list, getal: int):
    for element in lijst_in:
        print('Element:', element)
        if element < getal:
            print('Gevonden:', getal)
            return
        
lijst_1 = [100, 5, 100, 2]
getal = 10
print('Getal in lijst?', vind_getal_in_lijst(lijst_1, getal))
print('-----')
output = vind_lager_getal_in_lijst_1(lijst_1, getal)
print(output)
print('-----')
vind_lager_getal_in_lijst_2(lijst_1, getal)
```

Let op, je kan functies makkelijk overheen cellen gebruiken. Ze blijven beschikbaar:

```python
def doe_iets_met_getallen(getal_1: int, getal_2: int) -> int:
    return getal_1 * getal_2
```

## Argumenten

We weten dat sommige functies geen vast aantal argumenten neemt, zoals bv. de `print()`-functie die we kunnen blijven vullen met strings gescheiden door komma's. We kunnen dit ook in onze eigen functies gebruiken door `*args` als het eerste argument te gebruiken (`args` als naam is een conventie en de asterisk duidt aan dat `args` een tupel is). `args` wordt dus gewoon opgevuld met alle argumenten die dan beschikbaar zijn en teruggegeven in een tupel. Indien er nog andere argumenten nodig zijn moeten deze achter `args` gezet worden en in de functieoproep vast met hun naam aangeduid (zoals hieronder waar `minimum` expliciet moet worden bijgevoegd als argumentnaam wanneer `som_getallen()` wordt opgeroepen), anders zouden ze bij de tupel worden gevoegd:

```python
def som_getallen(*args: tuple, minimum: float) -> float:
    print('Aantal getallen meegegeven aan args:', len(args))
    return sum(args) + minimum

print('Uitkomst 1:', som_getallen(1, 2, 3, 4, minimum=10))
print('Uitkomst 2:', som_getallen(1, 4, minimum=100.2))
```

Type hinting wordt niet afgedwongen, maar toch is deze info zichtbaar in bepaalde programmeeromgevingen (zoals Spyder, maar niet Jupyter notebooks). Ten slotte kunnen we ook, m.b.v. van 3 aanhalingstekens, een korte descriptie voor een functie meegeven.

Ze kunnen wel worden opgevraagd door het attribuut `__annotations__` op te vragen (let op, dit zijn variabelen eigen aan een functie en zijn vergelijkbaar met functies, maar dan zonder haakjes om ze uit te voeren - ze zijn slechts data en geen code):

```python
def som_van_getallen(getal_1: 'integer', getal_2: 'integer') -> 'integer':
    """Deze functie telt 2 getallen op"""
    return getal_1 + getal_2

print(som_van_getallen(1, 4))
print(som_van_getallen('een', ' string!'))
print(som_van_getallen.__annotations__)
```
