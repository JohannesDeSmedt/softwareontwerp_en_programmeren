# Oefeningen 1.8

```python
def dubbel(x):
    print(x * 2)
y = dubbel(3)
print(y)

# VERBETER: laat de functie de waarde teruggeven:
def dubbel(x):
    return x * 2 # de return wordt toegevoegd zodat de functie de waarde teruggeeft in plaats van enkel te printen
y = dubbel(3)
assert y == 6
```

```python
teller = 0
def verhoog():
    teller = 10
verhoog()
print(teller)

# VERBETER: laat functie iets teruggeven en ken toe aan teller (zonder global):
def verhoog(t): 
    t = t + 10
    return t # toegevoegd zodat de functie de waarde teruggeeft
teller = verhoog(teller) # daarna roepen we de functie aan en ken de waarde toe aan teller
assert teller == 10
```

```python
def info(naam, jaar=1):
    return naam, jaar
print(info("An"))
print(info(jaar=2, naam="Bo"))

# VERBETER: haal de twee returnwaarden uit elkaar:
n, j = info("An", 2)
assert n == 'An' and j == 2
```

```python
class Student:
    def __init__(self, naam, nummer):
        self.naam = naam
        self.nummer = nummer
s = Student('An', 1234)

# VERBETER: maak nieuwe student s aan:
s = Student('Bo', 4321) 
assert s.naam == 'Bo' and s.nummer == 4321
```

```python
class Vak:

    def roep():
        return titel.upper()

# VERBETER: pas de class aan zodat de assert uitkomt:
class Vak:
    def __init__(self, titel):
        self.titel = titel

    def roep(self):
        return self.titel.upper() # de self is toegevoegd zodat de methode de titel van het vak in hoofdletters teruggeeft

vak = Vak("python")
assert vak.roep() == 'PYTHON'
```

```python
class Teller:
    def __init__(self):
        self.n = 0
a = Teller()
b = a
b.n = 5
print(a.n, b.n, a is b)

# VERBETER: a.n moet 0 blijven (in object b):
a = Teller()
b = Teller() 
b.n = 5
assert a.n == 0 and b.n == 5
```
