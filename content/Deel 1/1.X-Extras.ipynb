```python
rgb_im = img.convert('RGB')

r, g, b = rgb_im.getpixel((100, 100))
r,g,b
```

```python
kleuren_set = set()
for h in range(rgb_im.height):
    for w in range(rgb_im.width):
#         print(rgb_im.getpixel((w, h)))
        kleuren_set.add(rgb_im.getpixel((w, h)))
kleuren_set
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

img = mpimg.imread('feblogo.png')   

img2 = img.copy()

print(type(img))
print(img.shape)
print(img[0][1])


for lengte in range(len(img)):
    for breedte in range(len(img[0])):
        for i in range(0,3):
            img2[lengte, breedte, i] = min(1, max(0, img[lengte,breedte,i]* 0.45))

plt.imshow(img2)
```
