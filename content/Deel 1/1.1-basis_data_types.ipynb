# 1.1 Basis data types

In deze notebook worden de verschillende basistypes 1-voor-1 overlopen met korte voorbeelden en veelgebruikte ingebouwde functies (die worden later nog uitgebreid behandeld).
Deze basistypes vormen de bouwstenen voor al de andere objecten die we ooit in de code willen gebruiken. Studenten kunnen bijvoorbeeld voorgesteld worden als een tupel (rij van $n$ elementen) van een studentennummer (geheel getal/integer) en een naam + voornaam (strings). Merk op dat foto's of beelden uiteindelijk ook maar een matrix van getallen zijn. Alle complexe types kunnen dus uitgedrukt worden in de basis types.

## Integers

Gehele getallen. Hieronder wordt een nieuwe variabele van het type/klasse integer (`int`) met een waarde van 5 aangemaakt. De ingebouwde functie `type()` laat toe om het type/de klasse van een variabele op te vragen:

```python
a = 5
a = 10

print(a)
print(type(a))
```

## Floating point numbers

Kommagetallen. Hieronder worden zowel de standaard komma (in Python punt) gebruikt alsook de wetenschappelijke notatie met de macht `e` (voor exponent, niet de constante):

```python
b = 5.3
c = 2.21e-10
d = 2.21e10

print(b)
print(c)
print(type(b))
print(type(c))
print(int(d))
```

Zoals te zien, is er geen onderscheid in type, beide formaten zijn in de achtergrond opgeslagen als klasse `float`.

## Functies voor cijfers

Hieronder worden enkele functies meegegeven die in Python standaard zitten ingebakken. Functies gebruiken ronde haakjes met 0 tot meerdere argumenten.
Afronden kan met `round()` door een cijfer mee te geven als eerste argument, samen met het aantal gewenste getallen na de komma als tweede argument. Daarnaast kan je het maximum of minimum van getallen berekeken met `max()` en `min()` respectievelijk (merk op: dit werkt ook op lijsten later - tenminste als daar enkel getallen inzitten), en een getal tot een macht verheffen met `pow()` (dit kan ook met `**`, bv. `2 ** 4`):

```python
a = 2
b = 2.0202
print(round(a, 2))  # rond af tot 2 cijfers na komma
print(round(b, 2))
print(round(a)) # tot geheel getal
print(max(a, b))
print(min(a, b))
print(pow(a, 10))  # machtsverheffing
print(a ** 11)
```

## Booleans

Binaire waar/vals waarden.

```python
waar = True
vals = False

print(waar)
print(type(waar))
```

Deze zullen zeker handig zijn in algorithmes, bv. om bij te houden of we een getal al eens zijn tegengekomen, of om te zien of andere condities voldaan zijn.

## Strings

Sequenties van tekens/karakters. Komen sterk overeen met lijsten en kunnen verschillende functionaliteiten van lijsten overnemen (zie later). Kunnen zowel tussen ' (enkele aanhalingstekens) als " (dubbele aanhalingstekens) voorkomen, maar je moet consistent hetzelrde aanhalingsteken gebruiken aan het begin en einde van een string (dus bv. niet  `'Programmeren"`).

```python
ch1 = 'b'
ch2 = "b"
str1 = 'blabla'

print(ch1)
print(ch2)
print(str1)
print(type(ch1))
print(type(ch2))
print(type(str1))
```

In andere talen (zoals Java) worden tekens als een aparte klasse beschouwd, waardoor een string een lijst van karakters kan beschouwd worden. In Python is dit niet het geval. Later zullen we nog zien dat de strings indices hebben per teken (die op zich ook gewoon een string zijn dus) die we makkelijk kunnen opvragen met vierkante haken. Een voorproefje:

```python
mijn_string = 'johannes'
# het 2de element van de string (van 0 te tellen)
print(mijn_string[1])
print(type(mijn_string[1]))
```

### Escape characters

Er zijn verschillende speciale tekens en combinaties van tekens die gebruikt worden om tekst in te delen en op te smukken. De forward slash '\\' wordt gebruikt om zo'n combinaties te maken. Zo zijn er:
- `\'` en `\"` om een apostrof/aanhalingstekens aan te duiden (gezien ze anders worden aanzien als een teken om een string af te bakenen)
- `\n` voor een nieuwe lijn ('\r\n' wordt meestal ook gebruikt op Windows)
- `\t` om een tab toe te voegen.

Zo kunnen we de volgende voorbeelden bouwen:

```python
kleine_zin = 'Mama, ik ben daarstraks \"naar de les geweest\"'
print(kleine_zin)

zin_met_slash = 'Ik heb mijn foto\'s bekeken'
print(zin_met_slash)

zin_over_meerdere_lijnen = 'Ik ben daarstraks gaan:\n\tfietsen\n\tzwemmen'
print(zin_over_meerdere_lijnen)
```

### Concatenatie

Om stukken tekst aan elkaar te lijmen kunnen we het plusteken gebruiken. Let op, een plusteken heeft verschillende invullingen naar gelang de types van de variabelen waarmee het combineerd wordt en is in dit geval gewoon een ingekorte versie voor een `join()` functie:

```python
string1 = 'eerste'
string2 = 'tweede'
concatenatie = string1 + ' ' + string2
print(concatenatie)
# hieronder wordt join gebruikt om een lijst van strings aan elkaar te zetten met een 
# spatie ertussen
print('Alternatief:', ' '.join([string1, string2]))

string1 += ' en nog wat erachter'
# += is equivalent aan string1 = string1 + ' en nog wat erachter'
print(string1)
```

## Dynamische typering

Uitermate belangrijk om geen fouten te maken binnen Python is het doorgronden van de dynamische typering die Python aanbiedt. Een standaardtoewijzing met de `=` operator zal een bepaalde variabele slechts naar een bepaald object laten verwijzen. Daarna kan je de variabele gewoon een andere verwijzing laten maken in het geheugen zonder dat dit naar een object van hetzelfde type moet zijn. Maw, een variabele kan eerst naar een integer verwijzen en later naar een string, maar dat betekent dat je natuurlijk moet opletten dat je je variabele daarna niet meer afrondt (want een string kan niet afgerond worden).

Dit komt later nog uitgebreider aan bod. Let hier zeker voor op in Jupyter notebooks, gezien de cellen in eender welke volgorde kunnen uitgevoerd worden.

```python
a = 4.2
print(round(a))
a = 'vier'
# dit geeft een fout
print(round(a))
```

## Conversies van types

Het is ook mogelijk om (basis)types in elkaar om te zetten via ingebouwde functies. Zo kan je een string omzetten in een getal, maar enkel als er daadwerkelijk enkel numerieke tekens te vinden zijn. 'True' kan Python ook nog herkennen, maar het Nederlandse 'Waar' uiteraard niet. Let dus zeker op met deze conversies en vermijd ze waar mogelijk.

```python
a = 'True'
print('a:', type(a))
a2 = bool(a)
print('a2:', type(a2))

a = '5'
print('a:', type(a))
a2 = int(a)
print('a2:', type(a2))
print('a:', a, 'a2:', str(a2))
```

Een kommagetal kan makkelijk worden afgerond (naar onder) naar een geheel getal mbv `int()`. Beter is om `round()` te gebruiken:

```python
a = 5.6
print(int(a))
a = 5.4
print(int(a))
a = 5.5
print(int(a))
print(round(a))
```
