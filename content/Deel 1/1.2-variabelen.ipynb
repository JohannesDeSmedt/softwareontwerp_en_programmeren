# 1.2 Variabelen

Ondertussen kunnen we al variabelen aanmaken via een toewijzing met het `=`-symbool. Maar wat gebeurt er eigenlijk in de achtergrond? Kort gezegd, bij een toewijzing zal er letterlijk aan de variabelenaam (zoals hieronder `a` en `b`) een geheugenplaats toegewezen worden waar er zich een object van een bepaald type bevindt samen met een waarde. 

Hieronder wijzen we `a` en `b` beide toe aan dezelfde geheugenplaats (die kan worden opgevraagd met functie `id`) waar een object van type `int` opgeslagen is met een waarde van 100 als volgt:

```python
a = b = 100
print('a:', a)
print('b:', b)
print(id(a))
print(id(b))
```

Als we nu `a` aanpassen, dan zal `b` nog altijd naar de oude plaats in het geheugen verwijzen en `a` naar een ander object van het type `int` met waarde 200 verwijzen:

```python
a = a + 100
print('a - nu:', a)
print('b - nu:', b)
print(id(a))
print(id(b))
```

Als we nu een variabele `c` zouden aanmaken met waarde 100 zien we het volgende:

```python
c = 100
print(id(c))
```

Dit is hetzelfde geheugenadres als `b`! In Python worden bij integers/booleans/floats/strings de waarde en de variabele met bijbehorende geheugenplaats gelijkgesteld.

Bij samengestelde types zoals strings, die bestaan uit een reeks van karakters/tekens, krijgen we het volgende:

```python
a = b = 'johannes'
print('a:', a)
print('b:', b)
print(id(a))
print(id(b))
```

Als we nu `a` aanpassen en een variabele `c` maken met dezelfde inhoud als `b` gebeurt er dit:

```python
a += ' de smedt'
b = 'johannes'
c = 'johannes'
print('a - nu:', a)
print('b - nu:', b)
print('c - nu:', c)
print(id(a))
print(id(b))
print(id(c))
```

Merk op dat `a` nu op een nieuwe plaats wordt bijgehouden en dat `c` nog altijd dezelfde geheugenplaats als `b` heeft omdat hun waarde hetzelfde is.
Het goede nieuws is dat er bij basis types dus automatisch een kopij werd gemaakt van de waarde van `a` voor die in een andere geheugenplaats een aanpassing kreeg op die waarde (zowel voor integers als strings).

Bij andere niet-basistypes zoals lijsten (details volgen in de volgende notebooks), echter, zal dit niet zomaar gebeuren. De verwijzing naar dezelfde geheugenplaats heeft dan ook gevolgen voor beide variabelen tegelijkertijd zoals hieronder duidelijk wordt:

```python
a = b = [1, 2, 3]
print('a:', a)
print('b:', b)
a.pop()  # verwijdert laatste element
print('a - nu:', a)
print('b - nu:', b)
print(id(a))
print(id(b))
```

Hier zien we dat als we een element uit de lijst van variabele `a` verwijderen, die ook uit `b` verwijderd wordt omdat ze naar hetzelfde object verwijzen. Hier moeten we dus goed voor opletten en eventueel een deep copy voor maken (zie later).
