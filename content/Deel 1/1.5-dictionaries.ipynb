# 1.5-dictionaries

Dictionaries (of woordenboeken, maar klinkt wat vreemd in het Nederlands, denk misschien eerder datamaps) zijn een uitgelezen bouwsteen bij de constructie van gestructureerde data. Dictionaries maken gebruiken van key-value (sleutel-waarde) paren waarbij elke sleutel uniek is. De sleutels vormen samen de sleutelset/keyset en zijn ook weer niet-muteerbaar zoals de elementen van een set, ze kunnen dus geen sets, dictionaries, lijsten, etc. zijn. Er kan wel eender welke waarde bij een sleutel horen (dus ook lijsten, sets, andere dictionaries, etc.). Waarden die bij een sleutel horen kunnen ook meerdere keren voorkomen vermits ze identificeerbaar zijn via de sleutel. Merk op dat dit sterk lijkt op een (relationele) database waarbij we een tabel hebben met unieke rijen, waarbij elke rij een unieke waarde heeft zoals een ID of volgnummer om die uniek te identificeren. 

Een woordenboek gebruikt, net als sets, accolades bij aanmaak. Dit omdat de sleutels een set vormen. Een dictionary kan aangemaakt worden door sleutels en hun waarden gescheiden door een dubbelpunt tussen accolades te zetten met een komma tussen de verschillende sleutel/waarde paren. Een lege dictionary kan ook gemaakt worden met de ingebouwde `dict()` functie.
De waarde van een sleutel kan opgevraagd worden tussen vierkante haken achter de variabele die de dictionary toegewezen kreeg. Let wel, als de sleutel niet bestaat krijg je een foutenmelding (een `KeyError`). Je kan de waarde van die sleutel wel aanmaken (of veranderen) door een toewijzing, ook als die nog niet bestaat, zoals hieronder:

```python
woordenboek = {'Johannes': 'SOEP1', 'Jan': 'SOEP2', 2: 'SOEP3'}
print(woordenboek)

woordenboek_2 = dict()
print(woordenboek_2)
# wijzigen van de waarde van sleutel 'Johannes'
woordenboek_2['Johannes'] = 'SOEP1'
print(woordenboek_2)
woordenboek_2['Johannes'] = 'SOEP2'
print(woordenboek_2)
# opvragen van niet-bestaande sleutel geeft een foutmelding
print(woordenboek_2['Johanes'])
```

We kunnen de sleutels en waarden ook apart opvragen met de `keys()` en `values()` methodes. Let op: de sleutels zitten in een object van het type `dict_keys`, dat kan omgezet worden naar een set zoals hieronder aangegeven. De waarden worden teruggegeven als `dict_values` en kunnen naar een lijst of een set omgezet worden. De key-value paren kunnen opgevraagd worden via `items()`. Let wel, je krijgt dan een itereerbaar object terug van het type `dict_items` dat je kan gebruiken in een for-lus en de elementen zijn allemaal tuples (paren).

```python
dict_1 = {'Johannes': 'SOEP1', 'Johanna': 'SOEP2'}
print('De sleutels:', dict_1.keys())
print('De key-value paren:', dict_1.items())
print('Het type van de sleutels:', type(dict_1.keys()))
print('De sleutels als set:', set(dict_1.keys()))
print('De key-value paren als lijst:', list(dict_1.items()))
print('Type van het eerste key-value paar:', type(list(dict_1.items())[0]))
print('De waarden:', dict_1.values())
print('De waarden als lijst:', list(dict_1.values()))
```

Zoals gezegd kunnen we de items overlopen met een for-lus, deze zal je later beter begrijpen, maar kan nu al van pas komen. Er wordt over alle sleutel-waarde paren gegaan en deze worden afgeprint:

```python
for key, value in dict_1.items():
    print('Sleutel:', key, '\twaarde:', value)
```

We kunnen dictionaries ook uitbreiden en aanpassen. 2 dictionaries kunnen samengevoegd worden met `update()` waarbij alle sleutels en hun waarden worden samengebracht (tenzij de ontvangende dictionary al een waarde voor de sleutel heeft - dan wordt er overschreven). Met `del` kan er zoals bij lijsten een bepaalde sleutel worden verwijderd (indien die bestaat). `pop()` laat ook toe om sleutels te verwijderen en geeft als resultaat indien gewenst de waarde bij die sleutel terug.

```python
dict_1 = {'Belgie': 10, 'Nederland': -10}
dict_2 = {'Belgie': 2, 'Duitsland': 2}
print(dict_1)
print(dict_2)
dict_1.update(dict_2)
print(dict_1)

del dict_1['Belgie']
print(dict_1)

# pop verwijdert de sleutel en geeft de waarde als resultaat
waarde = dict_1.pop('Nederland')
print(dict_1)
print(waarde)

del dict_1['Belgie']
print(dict_1)
# dit geeft een fout omdat de sleutel niet meer bestaat
```

Ten slotte nog een kleine opmerking die verder gaat dan dictionaries en ook relevant is voor andere sorteerbare objecttypes zoals sets, tuples, lijsten en dictionaries. Al deze objecten zijn itereerbaar en dus rechtstreeks te gebruiken met een for-lus, zoals we later zullen zien. Verder zijn ze ook sorteerbaar wat betekent dat ze inherent een volgorde kunnen toegewezen worden (mbv van bv. ><-operatoren bij getallen, of het alfabet bij tekst). Een lijst of set van getallen zal bijvoorbeeld gesorteerd worden op basis van grootte. Een lijst van strings zal alfabetisch worden behandeld. Dit kan op 2 manieren:
- Via de ingebouwde `sorted()` functie: deze zal een gesorteerde versie van de meegegeven variabele teruggeven.
- Via de `sort()`-methode: deze zal de inhoud van de variabele zelf sorteren en de volgorde van de variabele dus voor altijd wijzigen (let wel, dit geldt niet voor sets).

Let wel, dit zal enkel lukken indien er slechts types die met elkaar vergelijkbaar zijn in de collectie zitten. Voor floats en integers is dit geen probleem, maar dus wel voor bv. een mix van strings en integers.

```python
lijst = [0, 4, 1, -5, 2]
print(lijst)
print(sorted(lijst))
# de originele lijst blijft ongewijzigd:
print(lijst)
lijst.sort()
# de lijst is nu definitief veranderd
print(lijst)

mijn_set = [1.1, 2, 4.4]
print(sorted(mijn_set))
print(mijn_set.sort())

# dit kan dus niet, want strings en getallen kunnen niet vergeleken worden:
print(sorted(['Johannes', 3, 2, 'Aster']))
```

Lege dictionaries kunnen gemaakt worden met `dict()` of met `{}`. Merk op dat je dus nooit een 'gewone' lege set kan maken met `{}`:

```python
lege_dict = dict()
print(type(lege_dict))
lege_dict = {}
print(type(lege_dict))
```
