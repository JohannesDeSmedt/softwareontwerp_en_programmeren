# 1.13 Debuggen

Tijdens het programmeren zal je gegarandeerd fouten maken. Dat is geen probleem, zolang je ze snel opgelost krijgt. Daarvoor moet je de fouten snel doorgronden. Hiervoor heeft Python een hele reeks aan voorgedefinieerde foutenboodschappen die worden opgeroepen wanneer een error/fout in de vorm van een `Exception` zich voordoet. Deze zijn ingebouwd in de code die je gebruikt - je kan ook je eigen `Exception` opstellen of oproepen maar dit gaat voorbij de scope van de cursus. Hieronder volgt een overzicht.

Er zijn uiteraard nog een groot aantal andere foutenboodschappen, maar deze zouden al het merendeel van jouw problemen binnen deze cursus moeten dekken. Hieronder zullen we elk van hen kort belichten zodat je ze leert herkennen wanneer je ze zelf tegenkomt in jouw code. Eerst overlopen we enkele syntax errors die worden opgeroepen als de code fout geschreven is. Daarna overlopen we enkele exceptions die worden opgeroepen als een logische fout zich voordoet (bv. deling door 0) die vanuit de code zelf worden opgeroepen. Veel van de errors geven ook wat extra informatie mee over wat er juist fout is gegaan.

### 2. `IndentationError`
Deze doet zich voor wanneer de code niet goed gestructureerd is qua uitlijning.

```python
for i in range (0, 10):
for j in range(0, 10):
print(i)
```

Merk op dat je ook fouten kan maken tegen uitlijning die niet worden belicht door Python omdat de code syntactisch en logisch correct is. Hieronder roepen we bv. het `print()`-statement niet onder de geneste `for` op, waardoor het enkel na de `for` wordt opgeroepen en we niet elk index-paar overlopen (wat natuurlijk wel de bedoeling kan zijn):

```python
matrix = [[1,2,3], [3,4,5]]

for i in range(len(matrix)): # overloopt de rij indices
    for j in range(len(matrix[0])): # overloop de kolom indices (gegeven dat de lijsten in de lijst even lang zijn)
        print('i en j', i,j)
    print('Element op plaats i,j:', matrix[i][j])
```

### 5. `ValueError`
Deze error wordt opgeroepen wanneer het juiste type variabele wordt meegegeven aan een functie, maar de waarde niet klopt met de berekening die binnen de functie gebeurt. Zo zal het bijvoorbeeld wel mogelijk zijn om de vierkantswortel van een getal te nemen, maar niet van een negatief getal:

```python
import math
a = 10
b = -10
c = math.sqrt(a)
d = math.sqrt(b)
```

```python
values = []
max_value = max(values)  
min_value = min(values)
```

Een andere veelvoorkomende context voor een `ValueError` is wanneer een itereerbaar object zoals een lijst of set wordt overlopen, maar de inhoud van dat object niet mapt op de variabelen die we reserveren voor de inhoud, zoals bv. in een `for`-lus waar we 3 resultaten verwachten `a, b, c`, maar de paren in onze lijst maar bestaan uit 2 element (hier (1,2) en (3,4)):

```python
for a, b, c in [(1,2), (3,4)]:
    print(a)
```

```python
for a, b in [(1, 2, 3), (3, 4, 5)]:
    print(a, b)
```

### 6. `KeyError`
Deze error wordt opgeroepen wanneer we een bepaalde sleutel uit een dictionary willen halen die niet bestaat:

```python
a = {'a': 1, 'b':2}
print(a['a'])
print(a['c'])
```

### 7. `IndexError`
Sterk gerelateerd aan de `KeyError` is de `IndexError` die wordt opgeroepen als je een index uit een lijst opvraagt die niet bestaat. Deze zal je heel dikwijls tegenkomen.

```python
lijst = [0, 1, 2, 3]
for i in range(0, 5):
    print(lijst[i])
```

Om dit vermijden, gebruik `len()` om de lengte van de lijst op te vragen:

```python
lijst = [0, 1, 2, 3]
for i in range(0, len(lijst)): # of zelfs gewoon range(len(lijst))
    print(lijst[i])
```

### 8. `AttributeError`
Deze error wordt opgeroepen wanneer een attribuut van een variabele wordt opgeroepen dat niet bestaat (voor het type van het object waar die variabele naar verwijst). Dit klinkt misschien abstract, maar beschouw even dit voorbeeld: een string heeft ingebouwde methodes zoals `lower()` of `format()`. Deze methodes zijn attributen van de string klasse. Als we een variabele eerst naar een string object laten verwijzen en daarna naar een integer object en we letten niet goed op, dan kan het zijn dat we toch bv. `lower()` aanroepen. Deze methode bestaat niet voor het huidige type van die variabele, namelijk het type integer. Enkele voorbeelden zijn hieronder te vinden:

```python
a = 'Johannes'
print(a.lower())
a = 10
print(a.lower())
```

```python
een_lijst = [10, 20]
een_set = set()

print(een_lijst.append(10))
print(een_set.append(10))
```

### 9. `TypeError`
Deze error wordt opgeroepen als een functie of operatie wordt opgeroepen over een object dat een verkeerd type heeft, bv. het optellen van een string en een integer, of het verminderen van een string:

```python
a = 'johannes' + 2
```

```python
a = [1, 2, 3]
b = a - 1
```

Werkt ook niet (wel met `+`):

```python
a = [1, 2, 3]
b = a - [1]
```

### 10. `OverflowError`
In het tweede deel van de cursus komt recursie aan bod. Het is makkelijk om dan per ongeluk een oneindige lus te maken en heel je geheugen vol te proppen. In dit geval zal waarschijnlijk de kernel van je Python interpreter crashen. Het kan ook zijn dat je een `MemoryError` of een `OverflowError` krijgt.

Eerst maken we een gigantisch grote variabele die niet meer in het geheugen past:

```python
j = 10.0
for i in range(1, 10000):
    j = j ** i
    print(j)
```

Dit misschien best niet uitvoeren:

```python
gigantische_lijst = []
for i in range(0,10**1000):
    for j in range(0, 10**100):
        for k in range(0, 10*100):
            gigantische_lijst.append('Een kleine string'*1000)
```
