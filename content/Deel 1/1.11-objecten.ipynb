# 1.11 Objecten

Zoals eerder al vermeld is alles in Python een object en is alles wat we doen in Python inherent object-georiënteerd. Het is niet de bedoeling van deze cursus om diep in te gaan op objectoriëntatie, maar het is wel handig om te snappen hoe objecten worden opgebouwd in Python om de syntax van bv. zelfs de basistypes beter te begrijpen. 

Objecten zijn instanties van een bepaalde klasse/type. Een klasse is als het ware een blauwdruk en wordt gebruikt om bepaalde concepten voor te stellen, zoals bv. een student, en z'n bijbehorende eigenschappen (attributen, gecapteerd in klasse-specifieke variabelen) en handelingen (methodes, gecapteerd in klasse-specifieke functies). Laten we een simpele voorstelling van een student maken:

```python
class Student:
    
    def __init__(self, naam_in: str, nummer_in: int, richting_in: str):
        self.naam = naam_in
        self.nummer = nummer_in
        self.richting = richting_in
        self.resultaten = {}
        
    def geef_naam(self) -> str:
        return self.naam
    
    def geef_punt_voor_vak(self, vak: str, punt: int):
        self.resultaten[vak] = punt
        
    def geef_gemiddelde_punt(self) -> float:
        punten = self.resultaten.values() # we krijgen alle punten terug als een itereerbaar object           
        print(list(punten)) # dit moeten we nog in een lijst omzetten om de sum()-functie erover te kunnen gebruiken
        if self.resultaten.values(): # de dictionary is niet leeg
            return sum(list(punten)) / len(punten)
        else:
            return 0
```

Ten eerste zien we dat een klasse kan aangemaakt worden met behulp van het keyword `class`, gevolgd door de naam (altijd met hoofdletter) en een dubbelpunt. Daarna komen er methodes bij, functies eigen aan alle objecten die worden gemaakt volgens de blauwdruk van deze klasse, die net als een gewone functie beginnen met `def`. Een opmerkelijk verschil is dat elke functie begint met hetzelfde argument `self`. Deze variabele stelt de instantie/een specifiek object van de klasse voor eens die is aangemaakt. 

Een instantie/object van een klasse aanmaken doen we door gebruik te maken van de constructor methode `__init__()`, die automatisch wordt opgeroepen als we een object aanmaken:

```python
student_1 = Student('Wilfried', 10, 'HIRB')
```

We gebruiken dus de functie `Student()`, die `__init__()` zal oproepen met dezelfde argumenten (`__init__()` wordt gebruikt voor elke klasse in Python) om de student een naam, nummer en richting mee te geven. Merk op dat `self` niet moet worden ingevuld, het wordt in de achtergrond door Python zelf aangemaakt. `self` wordt in de klasse gebruikt om de attributen van de klasse op te slaan, zoals `self.naam`, `self.nummer`, `self.richting` en de lege dictionary `self.resultaten`. Deze zijn later ook op te vragen via het `.` als volgt:

```python
student_1.naam
```

Merk op, de klasse zelf bestaat niet op zichzelf, je kan dus ook de naam ervan niet opvragen, want een blauwdruk heeft geen specifieke naam, enkel de instanties/objecten gevormd naar de blauwdruk:

```python
Student.naam
```

Het is niet noodzakelijk in Python, maar wel best-practice om altijd alle variabelen te initialiseren in de constructor.
We zien ook dat er een aantal methodes zijn aangemaakt. Deze bestaan dus niet buiten de functie:

```python
geef_naam()
```

Ze bestaan wel als methode van een object van de klasse `Student`:

```python
student_1.geef_naam()
```

Dit is dus gelijkaardig aan bv. de methode `.lower()` van een string object, dat volgens de blauwdruk van de klasse `str` is gemaakt:

```python
a = 'stUDENT'
a.lower()
```

Objecten worden het meest gebruikt om specifiek bepaalde data op te slaan. Hieronder geven we een paar studenten punten en berekenen we hun gemiddelde punt:

```python
student_1.geef_gemiddelde_punt()
```

```python
student_1.geef_punt_voor_vak('grondslagen', 10)
student_2 = Student('Monique', 11, 'HIRB')
student_2.geef_punt_voor_vak('BIS', 18)
student_2.geef_punt_voor_vak('grondslagen', 4)
```

De dictionaries `resultaten` van beide studenten zien er nu als volgt uit:

```python
print('Resultaten student 1:', student_1.resultaten)
print('Resultaten student 2:', student_2.resultaten)
```

```python
student_1.geef_gemiddelde_punt()
```

```python
student_2.geef_gemiddelde_punt()
```

Het is vooral belangrijk om te onthouden dat objecten instantiaties van een bepaalde klasse zijn, die elk hun specifieke eigenschappen (attributen) en methodes hebben. Het zal het makkelijker maken om bepaalde code in de les te begrijpen, zeker als we later naar externe modules kijken zoals bij NumPy en Pandas.
