### Oefening 10
Hier wordt er geen fout gemaakt die de code tegenhoudt, maar toch wordt de lijst hieronder niet juist gefilterd, kan jij dit verhelpen?

```python
lijst = [0, 10, 20, 145, 2]

for el in lijst:
    print(lijst)
    if el > 10:
        lijst.remove(el)
print('Lijst op het einde:', lijst)
```

Een comprehension doet dit wel juist (waarom?):

```python
nieuwe_lijst = [el for el in lijst if el > 10]
print('Nieuwe lijst:', nieuwe_lijst) 
```
