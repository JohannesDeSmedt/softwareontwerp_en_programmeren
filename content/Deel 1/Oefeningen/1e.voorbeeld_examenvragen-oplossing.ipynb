## Opgave 1
Controleer of 2 woorden, `woord1` en `woord2`, meegegeven met de functie `is_anagram()`, anagrammen zijn - woorden die uit juist dezelfde letters bestaan, maar eventueel met een andere volgorde (ook het aantal voorkomens van letters moet overeenkomen). Dit doe je door eerst alle letters van het eerste woord in een lijst te stoppen en daarna door het tweede woord te loopen en telkens de letter die je tegenkomt uit de andere lijst te verwijderen. Indien de lijst leeg is na het overlopen van het tweede woord weet je dat de 2 woorden anagrammen zijn.
In de variabele `aantal_checks` houd je bij hoeveel controles je hebt moeten doen. Als de woorden niet evenlang zijn mag je al stoppen na 1 check, anders moet je het tweede woord helemaal overlopen (je moet geen bewijs van het tegendeel gebruiken).

Enkele extra regels:
- Gebruik geen variabelen van buiten de functie `is_anagram()`, je mag er wel binnen de functie bijmaken.

```python
def is_anagram(woord1:str, woord2: str) -> (int, bool):
    aantal_checks = 0
    
    # je return statement moet eerst aantal_checks teruggeven
    # en dan een boolean, bv.
    # return aantal_checks, True
    
    # controleer of de 2 woorden even lang zijn
    # indien niet, kan je al stoppen
    if len(woord1) != len(woord2):
        aantal_checks += 1
        return aantal_checks, False

    # stop alle letters van woord1 in een lijst
    letters = [letter for letter in woord1]
    # overloop woord 2
    for j in woord2:
        print('-> Letters:', letters)
        # verhoog het aantal checks
        aantal_checks += 1
        # als de letter uit het tweede woord
        # in de lijst zit, verwijder die uit de lijst
        if j in letters:
            letters.remove(j)
            print('After letters:', letters)
            
    # als de lijst leeg is, kan je True teruggeven
    # anders False (vergeet ook aantal_checks niet)
    if len(letters) == 0:
        return aantal_checks, True
    else:
        return aantal_checks, False
```

```python
woord1 = 'straat'
woord2 = 'staart'
aantal_controles, resultaat = is_anagram(woord1,woord2)
print(aantal_controles, resultaat)
assert aantal_controles == 6
assert resultaat == True
```

```python
woord1 = 'straaat'
woord2 = 'staart'
aantal_controles, resultaat = is_anagram(woord1,woord2)
print(aantal_controles, resultaat)
assert aantal_controles == 1
assert resultaat == False
```

## Opgave 2
In deze vraag moet een getal omgezet worden naar z'n binaire representatie, i.e., in exponenten van 2. Dit doen we door eerst de exponent van 2 te verhogen tot we op of boven het getal geraakt zijn dat we willen omzetten. Voor 33 bijvoorbeeld (in binaire representatie '100001' van links naar rechts te lezen als $1 \times 32 + 0 \times 16 + 0 \times 8 + 0 \times 4 + 0 \times 2 + 1 \times 1$) zullen we eerst tot exponent 6 lopen vermits $2^6=64$ want $2^5=32$ is nog te klein om het hele getal 33 te dekken. 

Dan gaan we ons binair getal maken van links naar rechts door de exponent die we gevonden hebben te verlagen en te zien of de vorige exponent een 1 verdient op die plaats of niet door restberekening t.o.v. ons begingetal 33. Dit doen we door te kijken of het opgegeven getal groter is dan 2 verheven tot de nieuwe, lagere exponent. Zo zullen we dus 6 verlagen naar 5 en controleren of $33-2^5=1$ groter of gelijk is aan 0. Vermits dit zo is, voegen we een `1` toe aan de binaire representatie, want dit betekent dat $2^5$ nodig is om het getal's waarde te halen. Zo gaan we verder met het nieuwe getal $1=33-2^5$ en controleren we of $1-2^4>=0$. Dit is niet het geval, dus plaatsen we een 0 op deze plaats. Dit doen we tot alle exponenten (0 inclusief want $2^0=1$ is nodig voor de laatste plaats) doorlopen zijn. De voorbeelden hieronder maken dit duidelijk. Je mag ervan uitgaan dat het argument `getal` een integer is groter of gelijk aan 0.

**Let op, de uitkomstvariabele 'binair' is een STRING**

Enkele extra regels:
- Voeg geen cellen toe.
- Gebruik geen variabelen van buiten de functie `vertaal_naar_binair()`, je mag er wel binnen de functie bijmaken.

```python
def vertaal_naar_binair(getal: int) -> str:
    exponent = 0
    binair = ''
    
    # vind eerst de juiste waarde voor 'exponent' door die telkens te verhogen
    # tot je op of boven 'getal' bent gekomen
#     while 2**exponent <= getal:
#         exponent += 1
    for i in range(0,100):
        if 2**i>getal:
            exponent = i
            break
        
    # wanneer deze gevonden is kunnen we deze gebruiken om terug te tellen
    print('Beginexponent:', exponent, f'(2^{exponent})=', 2**exponent)
        
    # verlaag de exponent voor je begint
    exponent -= 1
    # verlaag nu de exponent tot je exponent kleiner is dan 0
    while exponent >=0:
        print('\nHuidige exponent', exponent, '=', 2**exponent)
        print('Getal:', getal)
        print('Getal - (2**exponent):', getal - (2**exponent))
        # kijk of het verschil van het huidige 'getal' - de huidige exponent van 2
        # groter of gelijk aan 0 is en voeg het juiste cijfer aan 'binair' toe
        if (getal - (2**exponent)) >= 0:
            print('oud getal', getal)
            binair = binair + '1'
            getal = getal - (2**exponent)
            print('nieuw getal', getal)
        else:
            binair = binair + '0'
        exponent -= 1
        # geef de huidige binaire voorstelling
        print('Binair:', binair)
        
    if binair == '':
        binair = '0'
    return binair
```

```python
getal = 33
resultaat = vertaal_naar_binair(getal)
print('\nResultaat:', resultaat)
assert resultaat == '100001'
```

```python
getal = 14
resultaat = vertaal_naar_binair(getal)
print('\nResultaat:', resultaat)
assert resultaat == '1110'
```

## Opgave 3

Je krijgt een dictionary met als sleutels de naam van een aandeel en als waarden een lijst met prijsobservaties van dat aandeel. Programmeer een algoritme dat de aandelen rangschikt op hun gemiddelde prijs. Hiervoor loop je door de aandelen, bereken je het gemiddelde van de prijs (bv. met NumPy `np.mean()` - **afgerond tot een integer**) en voeg je het aandeel op de juiste plaats toe in de lijst `uitkomst`. Je slaat het aandeel op als een tupel met 2 elementen: de naam van het aandeel en het gemiddelde (bv. `(aandeel_naam, gem)`). Hiervoor loop je door de lijst tot je een aandeel tegenkomt met een lager gemiddelde en voeg je het huidige aandeel op die plaats toe (daarna moet je niet meer verderlopen door de lijst, want die is toch gesorteerd).

Je mag er vanuit gaan dat er minstens 1 aandeel in de dictionary `aandelen` zit en dat er minstens 1 prijsobservatie is per aandeel.

```python
import numpy as np

def controleer_aandelen(aandelen: dict) -> list:
     
    # hierin houden we de oplossing bij
    uitkomst = []
    
    # loop door de aandelen
    for aandeel, prijzen in aandelen.items():
        print('Aandeel:', aandeel)
        
        # bereken het gemiddelde
        gem = round(np.mean(prijzen))
        print('Gem:', gem)
        
        # voeg het aandeel op de juiste plaats in 
        # in `uitkomst`. Je kan hiervoor bv, insert gebruiken
        if len(uitkomst) == 0:
            uitkomst.append((aandeel, gem))
        else:
            index = 0
            for a, l in uitkomst:
                if gem > l:
                    uitkomst.insert(index, (aandeel, gem))
                    break
                elif index == len(uitkomst) - 1:
                    uitkomst.append((aandeel, gem))
                    break
                index += 1
        print('Uitkomst na invoegen:', uitkomst)
    
    return uitkomst
```

```python
aandelen = {'samsong': [10, 20, 20, 21, 9, 39], 
            'oppel': [50, 10, 50, 11, 19, 139],
            'heiuaw': [10, 20, 20, 201, 90, 39],
            'uawhei': [100, 200, 200, 201, 90, 39]}
oplossing = controleer_aandelen(aandelen)
print('Oplossing:', oplossing)
assert oplossing == [('uawhei', 138), ('heiuaw', 63), ('oppel', 46), ('samsong', 20)]
```

```python
aandelen = {'samsong': [10, 39], 
            'oppel': [50, 139],
            'heiuaw': [10, 39],
            'uawhei': [100, 201, 90, 39]}
oplossing = controleer_aandelen(aandelen)
print('Oplossing:', oplossing)
assert oplossing == [('uawhei', 108), ('oppel', 94), ('samsong', 24), ('heiuaw', 24)]
```
