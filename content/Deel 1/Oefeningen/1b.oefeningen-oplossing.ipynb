## Vraag 1

Bereken de samengestelde interest van een bedrag tegen een bepaald percentage over 10 jaar met een for-loop:

```python
bedrag = 1000
rate = 0.1

for i in range(0,10):
    bedrag += bedrag * rate
    
print('Finaal bedrag:', bedrag)
```

Verwijder alle sleutels in een dictionary die een even waarde hebben. Let op, je moet hiervoor eerst de sleutels vinden die verwijderd moeten worden en mag ze dan pas verwijderen:

## Vraag 4

Er zijn 3 scenario's die moeten bekeken worden, gegeven 3 strings:
- Als string 1 enkel letters bevat en string 3 zonder white spaces aan de linkerkant enkel letters en cijfers bevat, print 'Scenario 1'
- Als die conditie niet houdt, check dan of string 2 zonder white spaces aan de linkerkant enkel letters bevat en print 'Scenario 2'
- Indien die conditie ook niet voldaan is, print 'Scenario 3'.

```python
def controleer_scenarios(string_1: str, string_2: str, string_3:str) -> str:
    if string_1.isalpha() and string_3.lstrip().isalnum():
        print('Scenario 1')
    elif string_2.lstrip().isalpha():
        print('Scenario 2')
    else:
        print('Scenario 3')

string_1 = '7@@hall0'
string_2 = '  lslslssl'
string_3 = 'js$(22!!  '
print('\nControle 1:')
controleer_scenarios(string_1, string_2, string_3)
print('\nControle 2:')
controleer_scenarios(string_1, string_3, string_1)
```

## Vraag 7

Vind het aantal keer dat 'a' voorkomt in een paswoord zonder gebruik te maken van de count() methode (met een for-loop dus):

```python
paswoord = 'alakjdaflgkj@LKjaldfkgj;b;b;b;aadfg'

aantal_keer = 0
for letter in paswoord:
    if letter == 'a':
        aantal_keer += 1
        
print('a komt', aantal_keer, 'voor')
# controle met count()
print('a komt', paswoord.count('a'), 'voor')
```

## Vraag 8

Voer verschillende controles uit op woorden en print ze anders naar gelang hun inhoud:
- Als er een 'a' in het woord staat: maak de eerste letter een hoofdletter
- Als er geen 'a' maar een 'd' in het woord staat, geef dan het aantal 'd's in het woord terug
- Indien er geen 'a' en 'd' voorkomen, controleer of het woord enkel letters bevat

```python
woorden = ['programmeren', 'studeren', 'eten', 'herhaal', 'et$en']
for woord in woorden:
    print('Woord:', woord)
    if 'a' in woord:
        print('Met hoofdletter aan begin:', woord.capitalize())
    elif 'd' in woord:
        print('Aantal keer dat \'d\' voorkomt: %d' % (woord.count('d')))
    else:
        print('Bestaat het woord enkel uit letters? %r' % (woord.isalpha()))
```

## Vraag 9

Stop alle string-gebaseerde en integer-gebaseerde sleutels van een dictionary in aparte dictionaries:

```python
orginele_dict = {'giovanni': 'pizza', 1: 'pasta', 'windows': 10}
string_dict = {}
int_dict = {}

for sleutel in orginele_dict.keys():
    if type(sleutel) == str:
        string_dict[sleutel] = orginele_dict[sleutel]
    elif type(sleutel) == int:
        int_dict[sleutel] = orginele_dict[sleutel]
        
print('String dictionary:', string_dict)
print('Integer dictioanry:', int_dict)
```

## Vraag 10

Zet alle getallen om naar gehele getallen en voeg hun kwadraat toe aan een set:

```python
lijst_met_getallen = [1, 2.4, 4, 5, 4, 1]
resultaat_set = set()

for getal in lijst_met_getallen:
    getal_nu = int(getal)
    getal_nu = getal_nu ** 2
    resultaat_set.add(getal_nu)

print('Resultaat set:', resultaat_set)
```

## Vraag 11

Voeg om de 3 plaatsen in een lijst een extra '@' toe:

```python
lijst = ['b', 'a', 'd', 'e', 'e', 'd', '4', 't', 'y']
lijst.sort()

index = 0  # kan eventueel aangepast worden om ergens anders te beginnen
while index < len(lijst):
    print(index)
    lijst.insert(index, '@')
    index += 3

print('Uitgebreide lijst:', lijst)
```

Bereken de cosinus van het hoogste getal in een lijst (gebruik hiervoor NumPy).
