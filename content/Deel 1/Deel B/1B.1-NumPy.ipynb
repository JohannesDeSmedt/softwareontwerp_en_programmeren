# NumPy

NumPy is een Python codebibliotheek die verscheidene wiskundige, voornamelijk algebraïsche, operaties implementeert. De onderliggende architectuur is gebaseerd op het gebruik van rijen (arrays) en matrices die vele malen sneller kunnen behandeld worden dan standaard lijsten omdat deze onderling vertaald worden naar C++ code. De matrices die opgesteld worden vormen de basis voor andere pakketten zoals pandas, scikit-learn, enz. Het is het Python antwoord op MatLab en deelt veel van de syntax. Meestal wordt NumPy als volgt geïmporteerd:

````python
import numpy as np
````

## Aanmaken en basisoperaties

Een nieuwe, lege (als in, vol met nullen) rij/matrix kunnen als volgt gemaakt worden:

````python
lege_rij = np.zeros(5)

print('Lege rij:', lege_rij)
print('Grootte van de rij:', lege_rij.shape)
print('Type van de rij:', type(lege_rij))
````

````python
lege_matrix = np.zeros((5,2))

print('Lege matrix:', lege_matrix)
print('Grootte van de dimensies van de matrix:', lege_matrix.shape)
print('Type van de matrix:', type(lege_matrix))
````

Deze matrices kunnen dan makkelijk ingevuld worden:

````python
lege_rij[2] = 4
lege_matrix[3,1] = 4

print('lege_rij:', lege_rij)
print('lege_matrix:', lege_matrix)
````

Er zijn nog andere manieren om matrices aan te maken, bv.:
- `zeros()`: maakt een rij/matrix met allemaal 0-waardes
- `ones()`: maakt een rij/matrix met allemaal 1-waardes
- `empty()`: maakt een rij/matrix met willekeurige waardes
- `arange()`: maakt een rij/matrix volgens de logica van de `range()`-functie
- `linspace()`: maakt een reeks van getallen in een interval op dezelfde afstand naar gelang het aantal elementen gevraagd
- `random()`: maakt willekeurige reeksen

Deze worden hieronder geïllustreerd. Merk op dat je vaak een enkel getal kan opgeven voor een rij, of een tupel om het aantal rijen/kolommen van de matrix die je wilt maken. Daarnaast zijn er nog andere argumenten per functie, zoals te zien in de voorbeelden.

## Berekeningen

### Snelheid: vectorisatie versus lussen

In de inleiding stond dat NumPy-berekeningen veel sneller kunnen verlopen dan met standaard Python-lijsten. Dit komt door *vectorisatie*: in plaats van element per element in een Python-lus te itereren, voert NumPy de berekening in één keer uit op de volledige array (onderliggend in C). Hieronder vergelijken we de som van een grote reeks getallen berekend met een Python-lus versus met de gevectoriseerde `sum()`-methode van NumPy:

````python
import time

n = 5_000_000
lijst = list(range(n))
array = np.arange(n)

start = time.time()
totaal_lijst = 0
for getal in lijst:
    totaal_lijst += getal
tijd_lijst = time.time() - start

start = time.time()
totaal_array = array.sum()
tijd_array = time.time() - start

print(f'Som met Python-lus: {totaal_lijst} in {tijd_lijst:.4f} seconden')
print(f'Som met NumPy (gevectoriseerd): {totaal_array} in {tijd_array:.4f} seconden')
print(f'NumPy is hier ongeveer {tijd_lijst / tijd_array:.0f}x sneller')
````

NumPy laat toe snel verschillende matrix-operaties uit te voeren. Merk hoe indexering wordt gecombineerd met standaard gedefinieerde attributen en attributen en methodes zoals ```T```, ```shape``` enz. Net zoals altijd telt NumPy ook vanaf 0 en niet vanaf 1:

````python
mat = np.array([[1,3,3],[4,5,6],[7,8,9]])

print('Matrix:\n', mat)
print('Cell 2,2:', mat[1,1])
print('Cell 2,3:', mat[1,2])
print('Eerste 2 rijen:', mat[0:2])
# mat[0:2,] kan ook gebruikt worden voor de eerste 2 rijen
print('Eerste 2 kolommen:', mat[:,0:2])
print('Getransponeerde matrix:\n', mat.T)
````

Daarnaast bevat `numpy.linalg` functies voor lineaire algebra, zoals de inverse en de determinant van een (vierkante) matrix:

````python
print('Inverteren (kan enkel op vierkante matrices):\n', np.linalg.inv(mat))
print('Determinant:\n', np.linalg.det(mat))
````

Het `axis`-argument kan bij bijna alle berekeningen gebruikt worden om een bepaalde projectie over een as te maken. Zo kan je bv. de som maken enkel langs de rij-dimensie met `axis=0` (dus voor de kolommen) of langs de kolom-dimensie met `axis=1` (dus voor de rijen):

````python
willekeurige_gehele_matrix = np.random.randint(5,10, (2,4))
print('Matrix:\n', willekeurige_gehele_matrix)

print('Totale som:',willekeurige_gehele_matrix.sum())
print('Som langs de rijen:', willekeurige_gehele_matrix.sum(axis=0))
print('Som langs de kolommen:', willekeurige_gehele_matrix.sum(axis=1))
````

## Statistieken

````python
mat_stat = np.array([[1,3,3],[4,5,6],[7,8,9]])
````

````python
print('Totale som:', np.sum(mat_stat))
print('Hoogste getal:', np.max(mat_stat))
print('Hoogste getal per kolom (langs de rijen gaan):', np.max(mat_stat, axis=0))
print('Hoogste getal per rij (langs de kolommen gaan):', np.max(mat_stat, axis=1))
print('Laagste getal:', np.min(mat_stat))
print('Gemiddelde:', np.mean(mat_stat))
print('Mediaan:', np.median(mat_stat))
print('Standaard deviatie:', np.std(mat_stat))
print('Variantie:', np.var(mat_stat))
````

Naast de waarde van het maximum/minimum kunnen we met `argmax()` en `argmin()` ook de *index* opvragen waar dat maximum/minimum zich bevindt. Zonder `axis`-argument wordt de matrix eerst plat (1D) gemaakt en gaat het over die platte index. `argsort()` geeft dan weer de indices terug die de reeks zouden sorteren, zonder de reeks zelf te sorteren:

````python
print('Index van het hoogste getal (platte matrix):', np.argmax(mat_stat))
print('Index van het laagste getal per kolom:', np.argmin(mat_stat, axis=0))
print('Indices die de eerste rij zouden sorteren:', np.argsort(mat_stat[0]))
````

Deze functies combineren ook goed met indexering, bv. om de som van een specifieke rij of kolom te berekenen:

````python
print('Som rij 1:', np.sum(mat_stat[0]))
print('Som rij 2:', np.sum(mat_stat[1]))
print('Som kolom 3:', np.sum(mat_stat[:,2]))
````

## Matrix operaties

We maken vier arrays aan met verschillende vormen (`shape`) om te tonen hoe NumPy optelt en vermenigvuldigt: een 3x3-matrix (`mat_1`), een vector (`mat_2`), een andere 3x3-matrix (`mat_3`) en een 1x3-matrix (`mat_4`):

````python
mat_1 = np.array([
    [1, 3, 3],
    [4, 5, 6],
    [7, 8, 9]])
mat_2 = np.array([4, 5, 6])
mat_3 = np.array([
    [5, 1, 1],
    [1, 0, 1],
    [5, 3, 4]])
mat_4 = np.array([[4, 5, 6]])
print('#rijen en #kolommen mat_1:', np.shape(mat_1))
print('#rijen en #kolommen mat_2 (vector):', np.shape(mat_2))
print('#rijen en #kolommen mat_4 (1x3 matrix):', np.shape(mat_4))
print('Som 1 & 4:', mat_1 + mat_4)
print('Som 1 & 3:', mat_1 + mat_3)
````

`mat_1 + mat_4` werkt ondanks de verschillende vormen dankzij *broadcasting*: de 1x3-rij van `mat_4` wordt herhaald over de 3 rijen van `mat_1`. Verwar dit niet met vermenigvuldigen: `*` en `np.multiply()` werken element per element (met dezelfde broadcasting-regels), terwijl `np.dot()` (of het `@`-symbool) het echte matrixproduct uit de lineaire algebra berekent:

````python
print('Elements-gewijze vermenigvuldiging 1 & 4:', mat_1 * mat_4)
print('Matrix product 1 & 2 ((3x3) * (3,) -> (3,)):', np.dot(mat_1, mat_2))
print('Elements-gewijze vermenigvuldiging 1 & 2:', np.multiply(mat_1, mat_2))
print('Matrix product 1 & 3 ((3x3) * (3x3) -> (3x3)):', np.dot(mat_1, mat_3))
# dit is equivalent:
print('Matrix product (met @) 1 & 3:', mat_1 @ mat_3)
print('Elements-gewijze vermenigvuldiging 1 & 3:', np.multiply(mat_1, mat_3))
````

Voor een matrixproduct moeten de dimensies overeenkomen: het aantal kolommen van de eerste matrix moet gelijk zijn aan het aantal rijen (of de lengte) van de tweede. `mat_2` (shape `(3,)`) voldoet hieraan, maar `mat_2.reshape((1,3))` (shape `(1,3)`) niet meer — dat geeft een `ValueError`, wat we hieronder opvangen om te tonen wat er dan gebeurt:

````python
mat_2 = np.array([4, 5, 6])
print(mat_2.shape)
print(mat_2.reshape((1,3)).shape)
print('Matrix product 1 & 2:', np.dot(mat_1, mat_2))

# dit kan niet volgens de wiskunde (dimensies komen niet overeen):
try:
    np.dot(mat_1, mat_2.reshape((1,3)))
except ValueError as e:
    print('Fout:', e)
````

## Hervormen en samenvoegen

Verder kunnen we ook makkelijk matrices bij elkaar voegen of hervormen:

````python
reeks_1 = np.array([1,2,3,4,5,6])
reeks_2 = np.array([7,8,9,10,11,12])
reeks_1_2x3 = np.reshape(reeks_1, (2, 3))
reeks_2_2x3 = np.reshape(reeks_2, (2, 3))

print('Rij 1 als 1x6 matrix:\n', reeks_1)
print('Rij 1 als 2x3 matrix:\n', reeks_1_2x3)
print('Rij 2 als 2x3 matrix:\n', reeks_2_2x3)
````

Deze reeksen/matrices kunnen we vervolgens samenvoegen. Voor 1D-reeksen kan dit met `np.concatenate()`, maar voor 2D-matrices is het vaak intuïtiever om `np.vstack()` (verticaal, rijen onder elkaar) en `np.hstack()` (horizontaal, kolommen naast elkaar) te gebruiken — dat bespaart je het opzoeken van het juiste `axis`-argument:

````python
print('Samengevoegd met concatenate (1D):', np.concatenate((reeks_1, reeks_2)))

print('\nSamengevoegd met vstack (rijen onder elkaar):\n', np.vstack((reeks_1_2x3, reeks_2_2x3)))
print('\nSamengevoegd met hstack (kolommen naast elkaar):\n', np.hstack((reeks_1_2x3, reeks_2_2x3)))
````

## Sorteren

Sorteren kan met de ingebouwde `np.sort()`-functie (dit is niet de gewone Python `sort()`-functie) die wordt toegepast langs de verschillende assen:

````python
mat_sorteer = np.array([
    [10, 3, 3],
    [4, 5, 8],
    [9, 8, 7],
    [0, -2, 7]])

arr1 = np.sort(mat_sorteer, axis = 0)   
print('Gesorteerd langs de rijen (per kolom):', arr1)
arr2 = np.sort(mat_sorteer, axis = 1)   
print('Gesorteerd langs de kolommen (per rij):', arr2)
arr3 = np.sort(mat_sorteer, axis = None)    
print('Gesorteerd zonder as:', arr3)
````

Merk op hoe de uitkomsten verschillende eindvormen kunnen aannemen. Dit kan bv. handig zijn als bv. de punten van verschillende vakken (kolommen) moeten worden gesorteerd tegelijk, of als machines (kolommen) moeten ingedeeld worden volgens hun omsteltijd voor bepaalde taken (per rij).

Om te weten welke unieke waarden in een reeks/matrix voorkomen, gebruik je `np.unique()`. Met het argument `return_counts=True` krijg je meteen ook het aantal keer dat elke waarde voorkomt (vergelijkbaar met `value_counts()` in Pandas, dat we later zien):

````python
mat_uniek = np.array([1, 2, 2, 3, 3, 3, 4])

waarden = np.unique(mat_uniek)
print('Unieke waarden:', waarden)

waarden, aantallen = np.unique(mat_uniek, return_counts=True)
print('Unieke waarden:', waarden)
print('Aantal per waarde:', aantallen)
````

## Filteren

NumPy biedt ook een reeks aan lambda-achtige functies om te filteren. Dit gebeurt via de `where()`-functie die wederom eerst een argument meekrijgt om mee te filteren (te behouden), bv. `mat_filter>2` om in de matrix `mat_filter` alle elementen groter dan 2 te behouden. Daarna kunnen we arumenten meegeven om aan te duiden wat te doen met de elementen die wel/niet aan deze conditie voldoen. We maken eerst een matrix aan:

````python
mat_filter = np.array([1,2,3,4,5,6])

mat_filter = np.reshape(mat_filter, (2,3))
mat_filter
````

De eenvoudigste manier om te filteren is door de matrix direct te indexeren met een booleaanse conditie. `mat_filter > 2` geeft een matrix van dezelfde vorm terug met `True`/`False`-waarden; gebruik je die als index, dan krijg je enkel de elementen terug waar de conditie `True` is (als platte, 1D-reeks). Je kan dit ook gebruiken om waarden aan te passen die aan de conditie voldoen:

````python
print('Booleaans masker:\n', mat_filter > 2)
print('Enkel de elementen > 2:', mat_filter[mat_filter > 2])

mat_filter_kopie = mat_filter.copy()
mat_filter_kopie[mat_filter_kopie > 2] = 0
print('Elementen > 2 vervangen door 0:\n', mat_filter_kopie)
````

Hier passen we `where()` toe met de conditie als eerste argument. Het tweede en derde argument duiden aan wat er moet komen op de plaats van de elementen waaraan de conditie voldaan is. Hier worden de elementen die niet voldoen aan de conditie vervangen door -10 en degene die voldoen vervangen door 10:

````python
nieuwe_mat = np.where(mat_filter > 2, 10, -10)
nieuwe_mat
````

Je kan ook de oorspronkelijke elementen behouden:

````python
nieuwe_mat = np.where(mat_filter > 2, mat_filter, 10)
nieuwe_mat
````

Als je enkel de conditie meegeeft zal je enkel de indices van de elementen die voldoen terugkrijgen:

````python
reeks_filter = np.array([7,8,9,10,11,12])
indices_reeks_filter = np.where(reeks_filter > 8)
indices_reeks_filter
````

Deze kan je dan terug uit de originele `ndarray` kunnen halen via `take()`:

````python
nieuwe_reeks_filter = np.take(reeks_filter, indices_reeks_filter)
nieuwe_reeks_filter
````

Voor een matrix krijg je de indices per dimensie mee, dus per rij en kolom:

````python
indices_mat_filter = np.where(mat_filter > 2)
indices_mat_filter
````

Hierbij zijn dus de eerste elementen van deze lijsten de (0,2) voor rij 0, kolom 2 (daar staat 3), het tweede element is (1,0) voor rij 1, kolom 0 (daar staat 4), enz.

Dit kan NIET met `take()` worden gebruikt, maar gewoon `mat_filter[indices_mat_filter]`:

````python
nieuwe_mat_filter = np.take(mat_filter, indices_mat_filter)
nieuwe_mat_filter
````
