## Oefening 1: haal_geslaagden

Een student is geslaagd voor een vak als de score **minstens** (dus ≥) de `grens` haalt. Deze functie moet enkel de geslaagde studenten teruggeven.

```python
import pandas as pd

studenten = ['otis', 'maeve', 'eric', 'adam', 'aimee']
data = {'vak1': [16, 11, 7, 8, 18], 'vak2': [11, 7, 17, 13, 10]}
df = pd.DataFrame(data, index=studenten)

def haal_geslaagden(df, vak, grens):
    return df[df[vak] > grens]

geslaagden = haal_geslaagden(df, 'vak2', 10)
print(geslaagden)
assert list(geslaagden.index) == ['otis', 'eric', 'adam', 'aimee']
```

## Oefening 2: gemiddelde_per_klas

Deze functie moet het **gemiddelde** van de kolom `'score'` teruggeven, gegroepeerd per `'klas'`.

```python
import pandas as pd

data = {'klas': ['A', 'A', 'B', 'B', 'B'], 'score': [10, 14, 8, 12, 13]}
df = pd.DataFrame(data)

def gemiddelde_per_klas(df):
    return df.groupby('klas')['score'].sum()

print(gemiddelde_per_klas(df))
assert gemiddelde_per_klas(df)['A'] == 12.0
```

## Oefening 3: tel_categorie

Deze functie moet teruggeven hoe vaak een specifieke `waarde` voorkomt in een kolom.

```python
import pandas as pd

data = {'vervoer': ['fiets', 'auto', 'fiets', 'fiets', 'trein', 'auto']}
df = pd.DataFrame(data)

def tel_categorie(df, kolom, waarde):
    return df[kolom].value_counts().index[0]

aantal_auto = tel_categorie(df, 'vervoer', 'auto')
print(aantal_auto)
assert aantal_auto == 2
```
