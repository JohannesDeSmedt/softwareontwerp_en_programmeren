### Spyder
Spyder is een IDE van het klassieke type, zoals Eclipse voor Java, of Visual Studio voor de C/Basic talen. Voor Python gebruikten we vroeger PyCharm, veruit 1 van de meest gebruikte IDEs, maar deze heeft als nadeel dat de connectie met Anaconda nog moet gemaakt worden. Verder bevat PyCharm veel extra toeters en bellen die we niet meteen gaan gebruiken. Spyder daarentegen is redelijk low-key, maar bevat net zoals andere klassieke IDEs een debugger. Die laatste is uitermate belangrijk om te leren programmeren omdat die toelaat om stapsgewijs door de code te lopen en bepaalde breakpoints te zetten om bv. for-lussen te debuggen, i.e., vrij te maken van bugs/fouten. Verder kan je ook heel mooi de variabelen die je hebt visualiseren om te zien naar welke objecten met hun bijbehorende waarden ze verwijzen. De code loopt (als je geen tussenscheidingen maakt) ook gewoon van boven (de eerste lijn code) naar onder (de laatste lijn code). De code wordt gewoon uitgevoerd waarbij het programma begint als je op 'Run file' klikt en eindigt als alle code is uitgevoerd. De uitvoer van het programma komt rechtstreeks in de console terecht, bv. alles wat je afprint via de `print()` functie.

### Jupyter notebooks
Jupyter notebooks daarentegen zijn iets anders opgevat. Ze bevatten verschillende cellen waarin zowel code als tekst en figuren kunnen zitten. Deze cell die je nu aan het lezen bent, bijvoorbeeld, is 'Markdown', dus gewone tekst. Jupyter notebooks zijn handig om code en tekst te mengen en wordt daarom gebruikt om een volledig interactieve cursus op te bouwen zodat je de code meteen kan uitvoeren en de output kan genereren per cell. Deze verschijnt onder de cel, er is dus geen console  (eigenlijk geeft elke cell de console output van de code die erin gedraaid heeft). Ook blijft de code opgeslagen in het geheugen en kan je bv. eerst de laatste cel uitvoeren en dan de eerste, zonder dat dit wordt gecontroleerd. Zo kan het zijn dat er variabelen worden aangemaakt in de eerste cel die nog niet beschikbaar zijn in de laatse, waardoor je een foutmelding krijgt. Of het kan zijn dat de variabele meteen overschreven wordt zonder dat dat de bedoeling was. Maw, het is altijd oppassen geblazen bij het uitvoeren van notebooks, maar langs de andere kant zijn notebooks heel handig om bv. in 1 lijn een dataset in te lezen en te tonen in een tabel, daarna een grafiek van enkele variabelen te plotten, enz. Vermits alles in geheugen blijft moet je dan ook niet telkens alle data terug inladen en kan je bv. verschillende grafieken aanmaken of blijven veranderen en verschillende analyses uitvoeren in verschillende cellen. Let op, je kan wel maar 1 cel tegelijk uitvoeren en het nummer van uitvoering komt ernaast staan tussen vierkante haken. Hou deze zeker in het hoog om te zien welke cellen in welke volgorde zijn uitgevoerd.

Verder bieden notebooks ook geen enkele vorm van debugging aan. Je zal dus geen syntaxfouten zien zoals een verkeerd geschreven woord (bv. `prin()` ipv `print()`), of de mogelijkheid om in een 'Variable explorer', zoals bij Spyder, de variabelen hun inhoud te bekijken terwijl je lijn voor lijn door de code gaat. Wel kan je natuurlijk om te debuggen zoveel mogelijk variabelen afprinten om hun huidige inhoud te bekijken. Dit is een meer rudimentaire vorm van debuggen, zoals we later zullen zien.

Jupyter notebooks worden opgeslagen onder de extensie '.ipynb', maar via 'File > Download as' linksboven kan je ze ook exporteren naar een gewone '.py' file en uitvoeren in bv. Spyder. Daarnaast kan je ze ook opslaan als pdf (maar dit gaat dikwijls mis) of als webpagina (.html). Zo zijn ze ook printbaar en dus te gebruiken als (papieren) cursus.

### Compilatie vs. interpretatie
Python is een taal die, in tegenstelling tot C(++) of Java niet telkens gecompileerd wordt. Compilatie zet code om in machinetaal die het besturingssysteem kan uitvoeren. Python (en R) worden on the spot geinterpreteerd en kunnen dus, zoals bij notebooks, in het geheugen blijven en telkens verder uitgevoerd (alvast bij Jupyter notebooks, Spyder kan dit ook maar gebruiken we zo niet voor didactische redenen). Kort gezegd maakt compilatie een programma dat bij uitvoering een output geeft terwijl interpretatie aan een bestaand programma vraagt om een output te geven. Dit heeft 2 grote implicaties voor dit vak:
- Bij een gecompileerd programma moeten alle fouten uit de code worden gehaald alvorens die uitvoerbaar kan worden gemaakt. Bij een interpretatie, bv. het uitvoeren van Python, kan je tegen fouten lopen die anders vermijdbaar waren (bv. een woord dat wordt gebruikt in een functie die afrondt waar er eigenlijk een getal moest staan).
- Een gecompileerd programma zit al rechtstreeks in machinetaal en loopt meestal veel sneller. Daarom gebruiken veel modules zoals NumPy en Pandas in de achtergrond rechtstreeks gecompileerde C code. Python is dan enkel een gebruiksvriendelijke schil rond die snellere code.

Op zich moeten jullie je hier voor de rest niet veel van aantrekken.

### Dynamische typering
Het allerlastigste aan Python is vooral de dynamische typering. Dit klinkt nu zeker nog abstract, maar je gaat er ongetwijfeld fouten tegen maken. In Python is alles een object van een bepaald type (ook klasse genoemd) en variabelen kunnen wisselen van type tijdens het uitvoeren van het programma. Zo kan bijvoorbeeld een integer aan een variabele `a` toegewezen worden en later aan een stuk tekst. Dat maakt dat code die die variabele gebruikt, bijvoorbeeld een stuk code die `a` wil laten delen door een ander getal, niet zal werken wanneer `a` ondertussen een stuk tekst is toegewezen (bijvoorbeeld per ongeluk in een andere cel). Daarom is het belangrijk altijd goede variabelenamen te kiezen, bv. `getal_a` ipv `a`. Later zullen we ook nog bepaalde conventies zien om variabelen, functies, ed. te benoemen.

### Eerste code
Maar nu dus een eerste stuk code. Hierin maken we 3 variabelen aan, of juister, we wijzen 3 variabelen, namelijk `a`, `b` en `c` een bepaalde waarde toe via de toewijzingsoperator `=`. `a` en `b` krijgen gehele getallen toegewezen en `c` krijgt een string (stuk tekst) toegewezen. De `print()` functie schrijft strings weg naar de console. De gehele getallen worden gelukkig automatisch omgezet naar strings, waardoor ze dus ook kunnen geprint worden.

```python
a = 5
b = 3
c = 'SOEP1'

print(a)
print(b)
print(c)
```

We kunnen `b` nu een andere waarde toewijzen:

```python
b = 5
print(b)
```

Let dus op dat variabelen andere waarden aannemen naar gelang de volgorde van uitvoering.
