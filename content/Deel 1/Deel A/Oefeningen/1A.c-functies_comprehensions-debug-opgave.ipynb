## Fout 6: filter() zonder list()

Dit moet het aantal niet-lege strings in `data` printen.

```python
data = ['a', '', 'b', '']
schoon = filter(lambda s: s != '', data)
print(len(schoon))
```

## Fout 7: waarom lost een comprehension dit op?

Beide fragmenten filteren `lijst` op getallen ≤ 10. Voorspel de output van **beide** — waarom geeft de eerste (lus-gebaseerde) versie een fout resultaat, en de tweede (comprehension) wel het juiste?

*(uit 1D.oefeningen_op_fouten, oefening 10)*
