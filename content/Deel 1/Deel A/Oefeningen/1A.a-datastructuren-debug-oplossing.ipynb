## Fout 1: KeyError bij tellen

Dit moet per element in `mijn_lijst` bijhouden hoe vaak het voorkomt.

**Antwoord:** `tellingen[element] += 1` vereist dat `element` al een sleutel is met een bestaande waarde. Bij het eerste getal (`1`) bestaat die sleutel nog niet, dus dat geeft een `KeyError`. Gebruik `tellingen.get(element, 0) + 1` (of `.setdefault()` / `collections.Counter`) om een default van 0 te voorzien.

```python
mijn_lijst = [1, 2, 3, 1, 2, 5]
tellingen = {}
for element in mijn_lijst:
    tellingen[element] = tellingen.get(element, 0) + 1
print(tellingen)
```

## Fout 2: dictionary aanpassen tijdens het overlopen

Dit moet alle sleutels met een even waarde verwijderen.

**Antwoord:** Je mag een dictionary niet wijzigen (sleutels toevoegen/verwijderen) terwijl je er met een `for`-lus rechtstreeks overheen itereert — dat geeft een `RuntimeError`. Overloop een kopie van de sleutels (`list(woordenboek.keys())`), of verzamel eerst de te verwijderen sleutels in een aparte lijst.

```python
woordenboek = {'a': 20, 'b': 13, 'c': 933, 'd': 100}
for sleutel in list(woordenboek.keys()):
    if woordenboek[sleutel] % 2 == 0:
        del woordenboek[sleutel]
print(woordenboek)
```

## Fout 3: paren zonder set

Dit moet elk uniek paar van `characters` precies 1 keer teruggeven (paar (a, b) == paar (b, a)).

**Antwoord:** Door beide personen over dezelfde volledige lijst te laten lopen, komt elk paar twee keer voor (('otis','maeve') én ('maeve','otis')) — hier dus 6 in plaats van 3 paren. Laat de tweede lus pas starten na de eerste (`range(i + 1, ...)`) zodat elk paar maar in één volgorde ontstaat, en/of gebruik een `set` zodat toevallige duplicaten er sowieso uitvallen.

```python
characters = ['otis', 'maeve', 'eric']
alle_paren = set()
for i in range(len(characters)):
    for j in range(i + 1, len(characters)):
        alle_paren.add((characters[i], characters[j]))
print(alle_paren)
print('Aantal paren:', len(alle_paren))
```

## Fout 4: een tuple aanpassen

Dit moet de eerste waarde van `punt` op 10 zetten.

**Antwoord:** Tuples zijn **immutable**: je kan een element niet rechtstreeks herschrijven, dat geeft een `TypeError`. Zet de tuple om naar een `list`, pas die aan, en zet ze terug om naar een `tuple` (of bouw meteen een nieuwe tuple, bv. `(10,) + punt[1:]`).

```python
punt = (3, 7)
punt_lijst = list(punt)
punt_lijst[0] = 10
punt = tuple(punt_lijst)
print(punt)
```

## Fout 5: dictionary verkeerd omgedraaid

Dit moet `REVERSE` opbouwen als code → letter, op basis van `MORSE` (letter → code).

**Antwoord:** `REVERSE[letter] = code` bouwt gewoon een kopie van `MORSE` op (zelfde sleutel/waarde-richting) in plaats van hem om te draaien. Daardoor bestaat de sleutel `'.-'` niet in `REVERSE` en geeft `.get()` de default terug. Het moet `REVERSE[code] = letter` zijn.

```python
MORSE = {'A': '.-', 'B': '-...', 'C': '-.-.'}
REVERSE = {}
for letter, code in MORSE.items():
    REVERSE[code] = letter
print(REVERSE)
print(REVERSE.get('.-', 'NIET GEVONDEN'))
```

## Fout 6: een lijst toevoegen aan een set

Dit moet alle personen uit `bewogen_ronde` toevoegen aan de set `stoute_kindjes`.

**Antwoord:** `.add()` voegt één enkel element toe aan de set — en dat element moet hashable zijn. Een `list` is niet hashable (hij kan nog veranderen), dus dat geeft een `TypeError`. Gebruik `.update()` om alle elementen van een lijst afzonderlijk toe te voegen.

```python
bewogen_ronde = [1, 4, 10]
stoute_kindjes = set()
stoute_kindjes.update(bewogen_ronde)
print(stoute_kindjes)
```

## Fout 7: een niet-bestaand element verwijderen

Dit moet element `3` (niet `4`) uit `lijst` verwijderen.

*(uit 1D.oefeningen_op_fouten, oefening 3)*

**Antwoord:** `4` zit niet in `lijst`, dus `remove(4)` geeft een `ValueError`. Merk trouwens op dat `remove()` zelf niets teruggeeft (`None`) — de methode past de lijst aan, ze geeft geen nieuwe lijst terug.

```python
lijst = [1, 2, 3]
print(lijst.remove(3))
print(lijst)
```

## Fout 8: een getal bij een lijst optellen

Dit moet `2` toevoegen aan de lijst `a`.

*(uit 1D.oefeningen_op_fouten, oefening 5)*

**Antwoord:** `+` tussen een `list` en een `int` bestaat niet (je kan enkel twee lijsten of twee getallen optellen), dat geeft een `TypeError`. Maak van `2` een lijst (`[2]`) om ze te kunnen samenvoegen.

```python
a = [1, 2, 3]
b = a + [2]
print(b)
```

## Fout 9: een lijst aanpassen tijdens het overlopen

Dit moet alle getallen groter dan 10 uit `lijst` verwijderen.

*(uit 1D.oefeningen_op_fouten, oefening 10)*

**Antwoord:** Net als bij een dictionary (fout 2 hierboven) mag je een `list` niet aanpassen terwijl je er met een `for`-lus over itereert: na het verwijderen van `20` schuift alles op, waardoor `145` wordt overgeslagen. Loop over een kopie (`lijst.copy()`), of verzamel eerst de te verwijderen elementen apart.

```python
lijst = [0, 10, 20, 145, 2]

for el in lijst.copy():
    if el > 10:
        lijst.remove(el)
print('Lijst op het einde:', lijst)
```
