# 1A.6 Operatoren

In Python maken we gebruik van verschillende operatoren die ons het leven makkelijk maken. In plaats van bijvoorbeeld het optellen van 2 getallen via de `sum()` functie, kunnen we ook gewoon de plusoperator gebruiken (`+`). Zoals we later zullen zien, wordt die naast een invulling bij getallen ook bv. gebruikt om lijsten uit te breiden, of strings aan elkaar te plakken. Daarom is het belangrijk altijd goed op te passen met operatoren aangezien hun gebruik per (combinatie) van types verschilt.

## Rekenkundige operatoren

De meest voorkomende operatoren zijn uiteraard de wiskundige. Hieronder worden alle operatoren overlopen voor het optellen, aftrekken, vermenigvuldigen, etc. van 2 (of meer) getallen (dus zowel integers als floats):

```python
a = 3
b = -10
print(a + b)
print(a - b)
print(a * b)
print(a / b) # gewone deling
print(50 // 3) # restdeling
print(50 % 3) # rest van deling (50-48)
print(b ** a)
try:
    print(a + 'johannes') # dit geeft een `TypeError` want integers en strings kunnen niet opgeteld worden
except TypeError as e:
    print(f"Fout: {e}")
```

Nog enkele extra voorbeelden:

```python
a = 2
b = 3
print(2 / 3)
print(2 // 3)
print(2 % 3)
print(2.0 / 3)
print(2 / 3.0)
print(2.0 // 3.0)
```

## Vergelijkende operatoren

Om de waarden van variabelen te vergelijken worden de volgende symbolen gebruikt (de output is telkens een boolean). Deze zijn voorlopig toegepast op getallen, hieronder worden ook gelijkheidsoperatoren op andere types toegepast.

```python
a = 10
b = -10
c = 5
d = 10
print(a == d)
print(a == b)
print(a != b)
print(a < b)
print(a >= b)
```

## Logische operatoren

De meest voorkomende operatoren in condities van bv. `if`-statements zijn de logische operatoren. Hieronder worden ze geïllustreerd met 2 boolean waarden, die bv. het resultaat kunnen zijn van vergelijkende en/of gelijkheidsoperatoren zoals in de laatste lijnen.

```python
a = True
b = False
print('a and b', a and b)
print('a or b', a or b)
print('a is b', a is b)
print('a is not b', a is not b)
print('not a', (not a))
print('not b', (not b))
print(5>3)
print((4>3) and (5>6))
```

## Gelijkheidsoperatoren

Hieronder wordt de gelijkheidsoperator meegegeven (de `is`-operator). Deze verschilt van de `==` operator omdat die niet de waarden, maar de objecten zelf gaat vergelijken. Dit vormt een groot verschil en soms een makkelijke manier om fouten te maken (hoewel misschien nog niet meteen binnen deze cursus). De objecten hebben een type, een waarde en een geheugenplaats. We weten als uit de vorige notebooks dat de waarde en de geheugenplaats van basistypes samenvallen.

In ieder geval, om het verschil te zien worden hier zowel basistypes zoals integers, als composiete types zoals lijsten vergeleken. Merk op dat, hoewel de inhoud van de lijsten hetzelfde is (de waarden), de lijsten zelf niet dezelfde zijn aangezien ze een ander object vormen (waarde + type + geheugenplaats). 

Bij de basistypes zit dit iets anders in elkaar. Denk daarbij terug aan onze eerdere uiteenzetting waarbij de `id()` functie gebruikt werd om de geheugenplaats op te vragen in de notebook rond variabelen. Bij integers en zelfs strings wordt bij een identieke verwijzing (bv. allebei naar het nummer 10 of de string 'johannes') geen nieuwe geheugenplaats gebruikt, omdat er maar 1 10 is. 
Bij lijsten wordt er telkens expliciet de constructor aangeroepen bij een toewijzing om een nieuw object te maken. Vandaar dat hieronder `a` en `b` bij `==` wel dezelfde waarden hebben, maar dus niet hetzelfde object zijn, wat we checken via `a is b`. Bij de integers en strings is dit wel het geval. Dit linkt natuurlijk ook terug aan onze eerdere bevindingen rond deep copy en het aanmaken van variabelen.

```python
a = [1, 2, 3]
b = [1, 2, 3]
c = 10
d = 10
e = 'Johannes'
f = 'Johannes'

print(a == b)
print(a is b)
print(c == d)
print(c is d)
print(e == f)
print(e is f)
```

## Volgorde van bewerkingen

Python volgt de standaard communicatieve, associatieve en distributieve eigenschappen van standaard wiskundige vergelijkingen:
- Communicatief eigenschap:	&nbsp;$a + b = b + a$ en $a \times b = b \times a$
- Associatieve eigenhschap:	&nbsp;&nbsp;$(a + b) + c = a + (b + c)$ en $(a \times b) \times c = a \times (b \times c)$
- Distributieve eigenschap:	&nbsp;&nbsp;$a \times (b + c) = a \times b + a \times c$
