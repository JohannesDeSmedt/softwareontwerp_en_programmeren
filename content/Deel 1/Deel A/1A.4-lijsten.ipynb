# 1A.4 Lijsten

Lijsten zijn een geordende collectie van elementen. De elementen worden gewoon in de volgorde van toevoeging bijgehouden en kunnen van eender welk type zijn (dus ook lijsten, sets en zelfs objecten van zelfgemaakte types). Zo kunnen we bv. ook lijsten in lijsten stoppen om (2-dimensionele) matrices te maken.

Lijsten zijn volledig geïndexeerd, waardoor je een element op een bepaald index kan opvragen. Er kunnen dubbels voorkomen, vermits elementen met dezelfde waarde een andere index kunnen hebben en zo uit elkaar kunnen gehouden worden.

Lijsten worden aangemaakt met de ingebouwde `list()` functie/constructor, of met vierkante haken:

```python
print('Lijst 1:')
lijst_1 = [1, 2, 'johannes', 2]
print(lijst_1)
print('Eerste element:', lijst_1[0])
print('Tweede element:', lijst_1[1])

print('\nLijst 2:')
lijst_2 = list()
print(lijst_2)
lijst_2.append(2)
print(lijst_2)

print('\nLijst 3:')
lijst_3 = []
print(lijst_3) 
lijst_3 = [1, 'johannes', 'johannes']
print(lijst_3)
```

## Indexering

De indexering van lijsten werkt door de vierkante haken te gebruiken achter de naam van de variabele waaraan de lijst is toegewezen. We kunnen zowel een enkele index opvragen met het indexnummer (let op: we tellen van 0, dus in een lijst met $n$ elementen is de laatste index $n-1$) van voren te tellen, of van vanachter, waarbij het laatste element index -1 heeft, het voorlaatste -2, enz. Verder kunnen we ook sublijsten opvragen door een range/bereik op te geven, bv. van het 2de tot (niet tot en met) het 4de element met `1:3`:

```python
lijst_3 = ['a', 'b', 'c', 'f', 1, 3, 4]
print('Lijst:\t\t\t\t\t\t', lijst_3)
print('Eerste element:\t\t\t\t\t', lijst_3[0])
print('Vierde element:\t\t\t\t\t', lijst_3[3])
print('Laatste element:\t\t\t\t', lijst_3[-1])
print('Tweede tot het vierde element:\t\t\t', lijst_3[1:3])
print('Tweede tot het zesde element in stappen van 2:\t', lijst_3[1:5:2])
print('De 3 laatste elementen:\t\t\t\t', lijst_3[-3:])
print('Het vierde laatste tot voorlaatste element:\t', lijst_3[-4:-2])
print('De omgekeerde lijst:\t\t\t\t', lijst_3[::-1])
```

## Lengte en membership

Met de ingebouwde functie `len()` kan je opvragen hoeveel elementen een lijst bevat (dit werkt trouwens ook voor sets, tuples, strings, enz.). Om na te gaan of een bepaalde waarde in een lijst voorkomt, gebruiken we, net zoals bij sets, het `in`-keyword. Dit geeft een boolean terug.

```python
lijst_3 = ['a', 'b', 'c', 'f', 1, 3, 4]
print('Aantal elementen:', len(lijst_3))
print("'c' in lijst_3:", 'c' in lijst_3)
print("'z' in lijst_3:", 'z' in lijst_3)
```

## Lijsten manipuleren

Lijsten kunnen makkelijk gemanipuleerd worden. In tegenstelling tot sets, wordt de `append()` methode gebruikt, niet de `add()` methode (die bestaat niet voor lijsten, dit is een veelgemaakte fout bij studenten). `append()` zal een element gewoon achteraan de lijst toevoegen. Als je een lijst toevoegt aan een lijst wordt die gewoon als element toegevoegd, als je alle elementen van een lijst aan een andere lijst wilt toevoegen gebruik je `extend()`. Daarnaast kan je ook elementen toevoegen op een bepaalde index met `insert()`, waarbij het eerste argument van de functie de plaats in de lijst aanduidt waar je het tweede argument wilt invoegen. We kunnen ook de `+` operator gebruiken om 2 lijsten bij elkaar te voegen (dit is equivalent aan `extend()` en niet `append()`). Let wel, als je dus een enkel element wilt toevoegen met `+` zal je dat ook eerst in een lijst moeten stoppen door er vierkante haken rond te zetten.

Ten slotte kunnen elementen op 2 manieren verwijderd worden. Met `pop()` wordt het laatste element van de lijst verwijderd. Met het keyword `del` (geen functie) kan een bepaalde index of range verwijderd worden. We kunnen ook elementen verwijderen met `remove()`, maar deze geeft wel een foutmelding als het element niet in de lijst bestaat (zoals `remove()` bij sets dus). `remove()` zal ook enkel het eerste voorkomen van een element verwijderen.

```python
lijst_4 = [2, 2, 1, 'johannes']
print('1:', lijst_4)
lijst_4.append('Johanna')
print('2:', lijst_4)
lijst_4.append([2, 'Johannes'])
print('3:', lijst_4)
lijst_4.extend([3, 5, 6])
print('4:', lijst_4)
lijst_4 = lijst_4 + [0, 2]
print('5:', lijst_4)
lijst_4.pop()
print('6:', lijst_4)
lijst_4.insert(3, 'blabla')
print('7:', lijst_4)
del lijst_4[2:8]
print('8:', lijst_4)
lijst_4.remove(2)
print('9:', lijst_4)
try:
    lijst_4.remove('Johannes') # dit geeft een foutmelding
except ValueError as e:
    print('Fout:', e)
    print('Johannes is niet aanwezig in de lijst')
print('10:', lijst_4)
```

Twee handige methodes om een lijst te doorzoeken zijn `index()` en `count()`. `index()` geeft de index van het eerste voorkomen van een waarde terug, en geeft (net zoals `remove()`) een foutmelding als de waarde niet in de lijst zit. `count()` telt hoe vaak een waarde in de lijst voorkomt; dit geeft nooit een foutmelding, ook niet als de waarde niet voorkomt (het resultaat is dan gewoon 0).

```python
lijst_5 = [2, 2, 1, 'johannes', 'Johanna']
print('Index van eerste 2:', lijst_5.index(2))
print('Aantal keer dat 2 voorkomt:', lijst_5.count(2))
print("Aantal keer dat 'blabla' voorkomt:", lijst_5.count('blabla'))
try:
    lijst_5.index('blabla') # dit geeft een foutmelding
except ValueError as e:
    print('Fout:', e)
    print("'blabla' is niet aanwezig in de lijst")
```

## Sorteren

Lijsten kunnen ook gesorteerd worden. De methode `sort()` sorteert de lijst zelf (in-place, de originele lijst verandert dus) in oplopende volgorde. Met het keyword-argument `reverse=True` kan je aflopend sorteren. Als je liever een nieuwe, gesorteerde lijst wil zonder de originele te veranderen, gebruik je de ingebouwde functie `sorted()`, die een nieuwe lijst teruggeeft.

```python
lijst_6 = [3, 1, 4, 1, 5, 9, 2, 6]
lijst_6.sort()
print('Oplopend gesorteerd (in-place):', lijst_6)
lijst_6.sort(reverse=True)
print('Aflopend gesorteerd (in-place):', lijst_6)

lijst_7 = [3, 1, 4, 1, 5]
lijst_8 = sorted(lijst_7)
print('Originele lijst blijft ongewijzigd:', lijst_7)
print('Nieuwe, gesorteerde lijst:', lijst_8)
```

## Lijsten in lijsten

Zoals eerder aangehaald kunnen lijsten eender welk (object)type bevatten, dus ook andere lijsten. Als we in een lijst lijsten stoppen van dezelfde lengte kunnen we zo een vorm van matrices bouwen. Hieronder wordt bv. een 2 bij 3 matrix gemaakt met 2 rijen (2 elementen in de hoofdlijst) en 3 kolommen (3 elementen in de sublijsten/rijen van de matrix). De elementen kunnen ook via de indices opgevraagd worden. De eerste index zal een element van de hoofdlijst opvragen (de rijen dus), de tweede index vraagt dan een element binnen dat eerst element uit de hoofdlijst op. Merk op, dit zou zelfs mogelijk zijn bij lijsten in lijsten in lijsten (3D matrices).

```python
hoofdlijst = list()
sublijst_1 = [1, 2, 3]
sublijst_2 = ['johannes', 5, 6]
hoofdlijst.extend([sublijst_1, sublijst_2])
print('Matrix:', hoofdlijst)
print('Tweede rij:', hoofdlijst[1])
print('Eerste element van de tweede rij:', hoofdlijst[1][0])
print('Derde element van de eerste rij:', hoofdlijst[0][2])
```

## Deep copy

Lijsten kunnen niet zomaar gekopieerd worden. Als we een variabele een lijst toewijzen, dan zal die naar de geheugenplaats van die lijst wijzen, niet naar de waarde(n) van de lijst zelf. Als we een kopij maken van een lijst (mbv `list()` met als argument de originele lijst) dan zullen enkel basistypes juist worden meegekopieerd, maar niet composiete variabelen zoals sets en lijsten. Als we een andere lijst met dezelfde waarden als die lijst willen maken, zullen we dus een deep copy moeten maken (in tegenstelling tot een shallow copy zoals `list()`). Zo'n deep copy zal recursief (om ook lijsten in lijsten in lijsten,... te vatten) alle elementen kopiëren naar een nieuwe geheugenplaats. Om een deep copy te maken gebruiken we de ingebouwde Python module `copy` en de daarin beschikbare functie `deepcopy()`.

Indien geen deep copy zou worden gemaakt, zouden we bij het veranderen van de ene lijst ook de andere lijst veranderen, zoals hieronder geillustreerd:

```python
import copy

eerste_lijst = [1, 2, [3, 4]]
# list kan gebruikt worden om een lijst te kopieren - een shallow copy genoemd
tweede_lijst = list(eerste_lijst)

tweede_lijst.append('extra')
print('Eerste lijst:', eerste_lijst)
print('Tweede lijst:', tweede_lijst)

tweede_lijst[2][0] = 5
tweede_lijst[0] = 'iets anders'
print('Eerste lijst na veranderen tweede lijst:', eerste_lijst)
print('Tweede lijst na veranderen tweede lijst:', tweede_lijst)
# merk op dat 5 in beide lijsten is aangepast vermits dit in een niet diep-gekopieerde
# sublijst zit

derde_lijst = copy.deepcopy(eerste_lijst)
derde_lijst[2][0] = 1
print('Eerste lijst na deep copy en veranderen derde lijst:', eerste_lijst)
print('Derde lijst na deep copy en veranderen derde lijst:', derde_lijst)
# nu is de sublijst enkel in eerste_lijst aangepast en niet in derde_lijst
```

## Tupels

Naast lijsten bestaan er ook nog tuples die erg op lijsten lijken. Het grootste verschil is dat ze niet-veranderbaar zijn. Ze worden aangemaakt met ronde haakjes en werken met dezelfde indexering als lijsten (ze mogen ook veranderbare types zoals sets en lijsten bevatten), maar ze kunnen dus niet aangepast worden met `append()` en `remove()`. Een veelgebruikt voorkomen voor tuples is een paar (tuple van 2 elementen):

Net zoals bij lijsten (`list()`) en sets (`set()`), bestaat er ook een `tuple()`-functie/constructor om een tuple te maken op basis van een andere iterable (bv. een lijst):

```python
lijst_tuple = [1, 2, 3]
tuple_van_lijst = tuple(lijst_tuple)
print(tuple_van_lijst)
print(type(tuple_van_lijst))
```
