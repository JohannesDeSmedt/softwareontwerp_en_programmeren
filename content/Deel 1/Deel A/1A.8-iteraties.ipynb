# 1A.8 Iteraties

Naast conditionele uitvoering wordt het andere grote stuk control flow in programma's bepaald door lussen via iteraties. Er zijn verschillende soorten iteraties, die elk hun nut hebben:
- Wanneer **vooraf het aantal iteraties is geweten**, bv. bij het overlopen van alle elementen uit een set, lijst, etc. kun je het best werken met `for`-loops. Die laten toe om over alle elementen van zulke collecties te lopen. 
- Als je echter een **iteratie wil blijven herhalen zolang een bepaalde voorwaarde niet voldaan is**, gebruik je `while`-lussen. Let wel op, deze zijn extra foutengevoelig omdat je in de while-lus er dan wel voor moet zorgen dat je ooit de conditie bereikt en de while-lus stopt. Anders raak je in **een eindeloze loop**. Daarom is het extra belangrijk dat je altijd goed nakijkt wat een while-lus doet met print-statements en debugging.

## `while`-lussen

Zoals gezegd zijn `while`-lussen dus geschikt als je naar een bepaalde conditie werkt, bv. als je door een rij wilt lopen en wilt stoppen wanneer je een bepaald getal tegenkomt. Je weet vooraf niet wanneer je het getal zal tegenkomen, maar moet er wel voor zorgen dat je stopt met lussen maken wanneer je het getal hebt gevonden (bv. met een variabele die dit bijhoudt als een boolean). De conditie wordt na het keyword `while` geplaatst, gevolgd door een dubbelpunt. Ook hier gebruiken we weer indents om aan te duiden waar de code bijhoort:

```python
a = 3

while a > 0:
    print('a aan het begin van de iteratie:', a)
    a = a - 1
    print('a aan het einde van de iteratie:', a, '\n')
```

Merk op hoe we hier de conditie `a > 0` controleren en daarna `a` telkens 1 verlagen tot `a` 1 is geworden. Aan het einde van de derde lus zal de `while` merken dat `a` 0 is geworden en de code niet meer uitvoeren.

```python
b = [1, 'a', 3, 'd']

while b:
    print(b)
    b.pop()
else:
    print('Lijst is nu leeg')
```

In dit voorbeeld zien we hoe een lijst element voor element wordt leeggemaakt m.b.v. `pop()`. De conditie `while b` (je kan dus een variabele gewoon gebruiken als conditie) in de context van lijsten (en andere collecties zoals sets, dictionaries,...) controleert of die leeg is of niet. Met andere woorden, de `while` blijft lopen zolang de lijst niet leeg is. Moesten we bv. een nieuw element toevoegen in elke stap van de `while`, dan zouden we nooit uit de lus geraken!

Ook merk je hier dat je na de laatste controle van de `while`, voor die stopt met loopen, ook nog een `else` kan uitvoeren. Die komt echter heel weinig voor en wordt amper gebruikt.

## `for`-lussen

Zoals eerder vermeld is een `for`-lus ideaal om over een discreet ingedeeld iets te lopen zoals een collectie. Een `for` kan worden gebruikt met een nieuwe variabele (hieronder `a`) om het element dat momenteel in de lus gebruikt wordt (bv. uit een collectie zoals een lijst die je wilt overlopen) aan te duiden gevolgd door `in` waarna de te overlopen collectie wordt gebruikt:

```python
for a in [1, 2, 3]:
    print(a)
```

Zoals eerder aangegeven, worden er geen enkele variabelen teruggegeven bij dictionaries, maar wel een tuple van lengte 2 (een paar) met daarin de sleutel en waarde:

```python
b = {'a': 10, 'b': 20, 'c': 30}
for (sleutel, waarde) in b.items(): # geeft de key-value paren
    print(sleutel, '-', waarde)
    
print('\nOok mogelijk:')
for sleutel, waarde in b.items(): # geeft de key-value paren
    print(sleutel, '-', waarde)
```

Enkel de waarden en sleutels:

```python
for waarde in b.values(): # geeft de waardes
    print(waarde)
```

```python
for sleutel in b: # gewoon een dict overlopen geeft de sleutels terug
    print(sleutel)
```

Gebruik altijd nieuwe variabelenamen in de `for`, anders krijg je de volgende problemen:

```python
i = 'johannes'

for i in [1, 2, 3]:
    print(i)
print('i na de loop:', i)
try:
    i += ' de smedt' # maar i is ondertussen een integer
except TypeError as e:
    print(f"Fout: {e}")
print(i)
```

Een veelgebruikte manier om `for`-lussen te maken is door een `range` te gebruiken. Die functie geeft een itereerbaar object terug naar gelang de argumenten die meegegeven worden:
- Een enkel getal als argument geeft alle (gehele) getallen van 0 tot (niet tot en met) dat getal, bv. `range(3)` geeft [0,1,2]
- Bij 2 argumenten gaan de resultaten tussen die 2 getallen met 1 ertussen (oplopend), bv. `range(1,5)` geeft [1,2,3,4]
- Bij 3 argumenten is het laatste argument de sprong die moet gemaakt worden (die kan ook negatief zijn), bv. `range(0,4,2)` geeft [0, 2].

Nog enkele voorbeelden:

```python
for i in range(3):
    print(i)
print('--')
for i in range(0, 6):
    print(i)
print('--')
# indien geen sprong wordt meegegeven tellen we op
# waardoor deze range niets teruggeeft:
for i in range(6, 0):
    print(i)
print('--')
for i in range(6, 0, -2):
    print(i) 
# let op: laatste element wordt niet meegenomen, hier 0 dus
```

Merk ook op dat je in de `for`-lus zelf je bv. een lijst of set niet mag uitbreiden - je krijgt dan inconsistenties omdat de  `for`-lus dan niet meer klopt (bv. je begint te loopen over 10 elementen, maar je voegt er 1 bij, waardoor de loop eigenlijk gedateerd is):

```python
mijn_lijst = list(range(0, 10)) # zeg een range om in een lijst
print(mijn_lijst)

for element in mijn_lijst:
    if element == 3:
        del mijn_lijst[4]
print(mijn_lijst)
```

Bij lijsten is dit vooral bad practice, maar bij ongeordende collecties zoals sets/dictionaries met hun key sets is dit helemaal uit den boze vermits we niet zomaar iets kunnen verwijderen op een bepaalde index en de volgorde van de loop behouden. Dit zou bv. niet werken:

```python
mijn_dictionary = {'a': 1, 'b': 2}

try:
    for sleutel, waarde in mijn_dictionary.items():
        if sleutel == 'a':
            del mijn_dictionary[sleutel]
    print(mijn_dictionary)
except Exception as e:
    print(f"Fout: {e}")
```

Indien je een volgnummer wilt toevoegen aan je lussen, kan je dat doen met `enumerate`. Deze functie neemt als argument een itereerbaar object en geeft naast de elementen ook een volgnummer terug zodat de resultaten ipv enkele elementen uit bv. een lijst nu een tupel worden van (volgnummer, element):

```python
for volg_nummer, inhoud in enumerate([3, 4, 5, 6]):
    print(volg_nummer, inhoud, sep=' : ')
print('----')
for volg_nummer, inhoud in enumerate(set([2, 8, 9, 9])):
    print(volg_nummer, inhoud, sep=' : ')
```

## break en continue

Naast de lussen kunnen we ook nog 2 keywords gebruiken die de control flow kunnen sturen. Ten eerste is er `break`. Dit keyword laat het toe om uit een lus te springen, bv. in combinatie met een `if`. Zo kan je een `for` stoppen wanneer je bv. een bepaalde conditie bent tegengekomen (hoewel `while` dan waarschijnlijk de betere oplossing was geweest). Heel handig als je vroegtijdig kan stoppen, want dat moet je uiteraard minder vaak itereren en is je code sneller. Het kan bv. gebruikt worden wanneer je een bewijs van het tegendeel hebt gevonden (zie later), of een uitzondering (bv. wanneer je door een lijst loopt op zoek naar het hoogste positieve getal, maar je komt een string tegen).

Dan is er ook nog `continue`, een keyword dat toelaat om de rest van de lus af te breken en naar de volgende iteratie te gaan. Zo kun je bv. een range doorlopen maar alle even getallen overslaan:

```python
# stoppen bij 2:
for i in range(0, 10):
    if i == 2:
        break
    print('i:', i)
print('----')
```

```python
# even getallen overslaan:
for i in range(0, 10):
    if i % 2 == 0: # indien de rest van een deling met 2 0 is -> getal is even
        continue
    print('i:', i)
print('----')
```

```python
# 3 overslaan:
f = 2
while f < 7:
    f += 1
    if f == 3:
        continue
    print('f:', f)
```
