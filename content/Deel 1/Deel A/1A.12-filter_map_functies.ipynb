# 1A.12 Filter en mapfuncties

Python heeft enkele ingebouwde functies die ons het leven makkelijker kunnen maken bij het bewerken van collecties: de `filter()` en `map()` functies. 

## De filter functie

De filter functie laat toe om collecties, of beter alle itereerbare objecten, te filteren gebaseerd op een zelfgemaakte functie of een lambda functie. Een lambda functie kan kort geschreven worden door het gebruik van het keyword `lambda` gevolgd door enkele variabelen/argumenten (hieronder `i`) en een conditie achter een dubbelpunt. Elk element van het itereerbaar object dat wordt meegeven aan de `filter` funtie zal dan om beurt in `i` gestopt worden (zoals in een for-loop) en gecontroleerd volgens de conditie. Het resultaat van een `filter` functie is een itereerbaar object met daarin de elementen die niet gefilterd werden (niet voldeden aan de conditie).

Hieronder geven we een eerste voorbeeld waarbij alle negatieve getallen uit een lijst overblijven. Eerst doen we dit met een normale for-loop en daarna in 1 lijn met een `filter()` oproep. Merk wel op dat het resultaat een standaard iterator is en dus geen lijst zoals het originele object `lijst`. Het kan wel terug omgezet worden naar een lijst via `list()`.

```python
lijst = [0, 2, -5, 6, -3]
nieuwe_lijst = []
for i in lijst:
    if i < 0:
        nieuwe_lijst.append(i) 
print('Nieuwe lijst:', nieuwe_lijst)

print('-----------')

#  of korter met lambda functie
nieuwe_lijst = filter(lambda i: i < 0, lijst)
print('Nieuwe lijst:', list(nieuwe_lijst))
```

Nu gebruiken we een zelfgemaakte functie in de `filter()` functie. De functie moet zo gemaakt worden dat er minstens 1 argument is en dat argument heeft dan dezelfde functie als de `i` hierboven: de functie wordt op elk element van het itereerbaar object dat wordt meegegeven aan de `filter` functie toegepast.

Hieronder wordt een veel makkelijkere, kortere versie meegegeven gebaseerd op het teruggeven van een boolean. Indien de letter geen 'a' is, geeft de functie `verwijder_a_voor_filter()` `True` terug, wat betekent dat de `filter` de letter behoudt:

```python
def verwijder_a_voor_filter(letter: str) -> bool:
    print('Input van \'verwijder_a_voor_filter\':', letter)
    return letter != 'a'

filteren_2 = filter(verwijder_a_voor_filter, 'programmeren')
print(list(filteren_2))
```

## De map functie

De map functie werkt op dezelfde manier als de `filter()`, met het verschil dat het elk element van het itereerbaar object aanpast in plaats van verwijdert/behoudt. Hieronder wordt elk getal uit een lijst verheven tot de macht $1/2$. We kunnen dit wederom doen met een zelfgemaakt functie, `neem_vierkantswortel()`, of een lambda functie:

```python
def neem_vierkantswortel(getal: int) -> float:
    return getal ** (1/2)

getallen = [1, 4, 16, 25]
uitkomst = map(neem_vierkantswortel, getallen)
print(list(uitkomst))

met_lambda = map(lambda x: x ** (1/2), getallen)
print(list(met_lambda))
```

Nog een korte opmerking rond lambda functies: ze kunnen meerdere argumenten hebben. Dit betekent ook dat er dan evenveel itereerbare objecten als argumenten in de lambda functie moeten volgen. Deze moeten zelfs niet evenveel elementen bevatten. Hieronder worden 2 lijsten meegegeven en stopt de lambda functie vanaf het moment dat 1 van de 2 lijsten volledig overlopen is (in dit geval `getallen_2`):

```python
getallen = [1, 2, 3, 4]
getallen_2 = [2, 4, 5]
uitkomst = map(lambda x, y: x ** y, getallen, getallen_2)
print(list(uitkomst))
```

We kunnen zelfs functies meegeven met `map()` die van 1 element een tuple maken. Hieronder wordt 1 getal omgezet in een tupel van het getal + 2 en het getal + 3:

```python
def dubbele_uitkomst(getal: int) -> (int, int):
    return getal + 2, getal + 3

print(list(map(dubbele_uitkomst, getallen)))
```
